### 1.1  Descripcion del proyecto

Este proyecto se centra en el análisis de señales electroencefalográficas (EEG) provenientes de una base de datos que registra la actividad cortical durante la ejecución y la imaginación motora de las extremidades superiores (mano derecha e izquierda) y estados de reposo. El propósito fundamental es determinar la viabilidad técnica de sistemas BCI no invasivos, fundamentándose en la neurofisiología del control motor para discernir patrones espaciales y espectrales específicos que permitan la clasificación de la intención del usuario [1].

### 1.2  Historia, Uso y Perspectivas Futuras de las BCI

La evolución de las interfaces cerebro-computadora (BCI) ha transitado desde una idea conceptual en los inicios de la tecnología digital hasta sistemas altamente sofisticados para la detección, registro y análisis de bioseñales. En la década de 1960, el uso de la electroencefalografía (EEG) estaba limitado a entornos de laboratorio debido a los requerimientos de equipamiento y a la susceptibilidad a artefactos; sin embargo, los avances en miniaturización electrónica y en amplificadores integrados en electrodos han permitido actualmente la adquisición de señales en entornos no controlados. Durante las últimas cuatro décadas, las BCI evolucionaron desde simples sistemas de comunicación hacia aplicaciones complejas de control asistivo. Los usos actuales de los BCIs incluyen la restauración motora a través de exoesqueletos robóticos y estimulación eléctrica funcional, la mejora cognitiva a través del neurofeedback y el entrenamiento de la atención, y herramientas de comunicación para personas con limitaciones físicas severas, exploradas principalmente dentro de entornos de investigación en lugar de la práctica clínica de rutina. En este contexto, las BCI se proyectan como herramientas de asistencia cada vez más integradas al individuo, con potencial para mitigar el aislamiento social y ampliar las capacidades de interacción entre el cerebro humano y los sistemas tecnológicos [2][3].

### 1.3  Actividad cerebral registrada por EEG durante movimientos e imaginación de movimiento de las manos y cómo se hace su análisis en sistemas BCI.

La imaginación motora (Motor Imagery, MI) es uno de los paradigmas más utilizados en sistemas de interfaz cerebro-computadora basados en EEG. Las MI implican simular movimientos mentalmente sin ejecución real, activando redes neuronales similares al movimiento físico, esto involucra regiones visuales y motoras asociadas, lo que facilita la planificación motora y potencialmente ayuda a la rehabilitación al promover la reorganización cortical y mejorar la función sensoriomotora. Esta activación se refleja en modulaciones de los ritmos sensoriomotores (SMR), que representan cambios en la sincronización de poblaciones neuronales durante la planificación y simulación de movimientos [4].

Los ritmos cerebrales de interés en tareas de MI se concentran principalmente en las bandas Mu (8–12 Hz) y beta (13–30 Hz). Durante la imaginación de movimiento, especialmente de las manos, se observa una desincronización de la banda Mu sobre la corteza sensoriomotora (event-related desynchronization, ERD), asociada a la activación cortical. De manera similar, la banda beta también presenta desincronización durante la planificación e imaginación del movimiento, seguida de fenómenos de resincronización al finalizar la tarea [5].

La adquisición de estas señales se realiza mediante electrodos de EEG ubicados sobre la corteza sensoriomotora, comúnmente en las posiciones C3 y C4 del sistema internacional 10–20, que corresponden a áreas motoras relacionadas con el control de las manos. Estas ubicaciones permiten captar con mayor precisión las modulaciones de los ritmos sensoriomotores asociadas a la imaginación de movimientos específicos [5].

En un sistema BCI basado en MI, el análisis de las señales se desarrolla en tres etapas principales: adquisición de señales, donde los electrodos registran la actividad cerebral; procesamiento de señales, que incluye filtrado y extracción de características relevantes en las bandas Mu y beta; y clasificación, donde algoritmos de aprendizaje automático identifican los patrones asociados a la tarea imaginada. El desempeño del sistema se evalúa mediante métricas como precisión, sensibilidad, especificidad y puntuación F1 [5].

### 1.4 Flujos de procesamiento propuestos para estos tipos de proyectos

Un sistema BCI típico se divide en adquisición, preprocesamiento, extracción de características, clasificación y control

En un sistema de Interfaz Cerebro–Computador (BCI), el proceso comienza con la **adquisición de la señal**, generalmente mediante electroencefalografía (EEG), que es una técnica no invasiva ampliamente utilizada en entornos académicos [6], [7]. Esta etapa implica el uso de electrodos colocados sobre el cuero cabelludo, junto con sistemas de amplificación y digitalización que permiten registrar la actividad eléctrica cerebral. Sin embargo, en esta fase surgen diversos problemas, como la presencia de ruido y artefactos, incluyendo interferencias por parpadeo (EOG), actividad muscular (EMG) y ruido eléctrico del entorno [7].

Posteriormente, se realiza el **preprocesamiento de la señal**, cuyo objetivo principal es mejorar la calidad del EEG eliminando o reduciendo el ruido. Para ello, se emplean técnicas como filtros pasa banda y filtros notch (generalmente a 50 o 60 Hz para eliminar interferencia eléctrica), así como métodos de eliminación de artefactos como el Análisis de Componentes Independientes (ICA) o técnicas de regresión [7], [8]. También es común aplicar procesos de normalización. Esta etapa es crítica debido a la baja relación señal/ruido característica de las señales EEG [6].

Una vez limpia la señal, se procede a la **extracción de características**, donde el EEG se transforma en un conjunto de variables representativas que pueden ser utilizadas por algoritmos de clasificación. Estas características pueden obtenerse en el dominio del tiempo (por ejemplo, media o varianza), en el dominio de la frecuencia (como la Transformada Rápida de Fourier o la densidad espectral de potencia) o mediante transformadas wavelet [7], [9]. Asimismo, es común trabajar con las bandas típicas del EEG (delta, theta, alfa, beta y gamma), las cuales son fundamentales en aplicaciones de BCI debido a su relación con distintos estados cognitivos [8].

A continuación, se lleva a cabo la **selección o reducción de características**, con el fin de disminuir la dimensionalidad de los datos y mejorar el rendimiento de los modelos. Entre las técnicas más utilizadas se encuentran el Análisis de Componentes Principales (PCA) y el Análisis Discriminante Lineal (LDA) [7]. En el caso específico de aplicaciones de imaginación motora, es muy común el uso de Patrones Espaciales Comunes (CSP), que permiten resaltar diferencias relevantes entre clases [8].

En la etapa de **clasificación**, se identifican los patrones cerebrales a partir de las características extraídas. Para ello, se utilizan algoritmos de aprendizaje automático como LDA, Máquinas de Vectores de Soporte (SVM) y K-Vecinos más Cercanos (KNN) [7], [9]. En enfoques más recientes, también se emplean redes neuronales profundas, como las redes convolucionales (CNN), destacándose arquitecturas especializadas como EEGNet [10].

Finalmente, el sistema culmina con la **interfaz o aplicación**, donde los resultados de la clasificación se traducen en acciones concretas. Estas pueden incluir el control de un cursor en pantalla, el manejo de prótesis, la interacción con videojuegos o la identificación de estados cognitivos del usuario [6].

Una vez preprocesadas las señales, se procede a la extracción de características relevantes que permitan cuantificar la actividad cerebral. En este caso, se propone el uso de la densidad espectral de potencia como medida principal, ya que permite analizar la distribución de la energía de la señal en función de la frecuencia. El análisis de las características extraídas se realiza mediante una comparación entre las diferentes condiciones experimentales, tanto a nivel espectral como espacial. En el dominio espectral, se evalúan las diferencias en la potencia de las bandas mu y beta entre movimiento e imaginación, así como entre mano izquierda y derecha. En el dominio espacial, se analizan las variaciones entre los electrodos C3 y C4 para identificar patrones de lateralización. ## **3.1 FUNCION DEL FLUJO DE PROCESAMIENTO**

Se procede a crear la función para el procesamiento de los datos (cargar_datos_eeg). Los bloques importantes del codigo seran explicados a continuación:

Se filtran únicamente los directorios no los archivos y luego se ordenan para mantener consistencia en el procesamiento. Además, se inicializan tres estructuras clave, ***datos_epochs***, donde se guardarán las épocas EEG organizadas, ***conteo***, para resumir cuántas épocas tiene cada sujeto y ***datos_raw***, para guardar los datos obtenidos crudos para hace graficas y comparar.

Para cada uno se crea una estructura jerárquica donde se separan los datos en dos condiciones experimentales, movimiento real y movimiento imaginado, y dentro de cada una se subdividen en eventos T0, T1 y T2, correspondientes a, una señal en reposo, una señal de movimiento real o imaginativo del brazo izquierdo y una señal de movimiento real o imaginativo del brazo derecho, respectivamente.

**3. RECORRIDO DE ARCHIVOS EDF**

Dentro de cada carpeta de sujeto, se recorren todos los archivos .edf, el cual es un formato típico de EEG, debido a esto, estos archivos se trabajaran con la libreria ***MNE***. Se filtran solo aquellos que pertenecen a los runs válidos ('R03','R04','R07','R08','R11','R12'). Además, se extrae el número de run desde el nombre del archivo, lo cual será usado después para clasificar el tipo de tarea teniendo en cuenta la descripcion del experimento, donde ***Task 1*** y ***Task 2***, son las relacionadas con la actividad real e imaginaria, respectivamente; De esta manera, se eligieron los runs a filtrar.

**4. CARGA DEL ARCHIVO EEG**

Como se menciono con anterioridad, se utiliza la librería MNE-Python para cargar el archivo EDF en memoria (preload=True). Esto permite trabajar directamente sobre la señal sin necesidad de acceder constantemente al disco, lo cual mejora eficiencia en procesamiento posterior. Ademas, se guarda una copia de los datos crudos.

**5. LIMPIEZA DE NOMBRES DE CANALES**

Algunos archivos EDF tienen inconsistencias en los nombres de los canales, como algunos espacios o puntos. Aquí se estandarizan eliminando estos espacios y puntos. Esto evita errores posteriores cuando se seleccionen canales específicos por nombre.

**6. SELECCIÓN DE ELECTRODOS DE CORTEZA MOTORA**

Se realiza la selección únicamente de los canales C3, C4 y Cz, que corresponden a la corteza motora según el sistema 10-20, ademas, son los canales sugeridos por el flujo de procesamiento. Por otro lado, esto tiene sentido porque en BCI de imaginación motora estas regiones contienen la información más relevante donde se encuentran los ritmos mu y beta asociados al movimiento.

**7. VERIFICACIÓN Y AJUSTE DE FRECUENCIA DE MUESTREO**

Se revisa la frecuencia de muestreo (fs), para asegurar que sean 160 Hz, que es el valor suministrado por la base de datos. Si no es 160 Hz, se remuestrea a ese valor. Esto se hace para garantizar homogeneidad entre sujetos, algo crítico para análisis comparativo o entrenamiento de modelos. Además, se evita imprimir múltiples veces el mismo mensaje usando un set de control.

**8. FILTRADO DE LA SEÑAL**

Se aplica un filtro pasa banda entre 7 y 40 Hz tipo FIR, el cual es un tipo de filtro digital cuya salida depende solo de un número finito de muestras pasadas de la señal de entrada. Este rango captura principalmente los ritmos mu (8–13 Hz) y beta (13–30 Hz), que son fundamentales en tareas motoras, ademas se escongen estos valores por la caracteristica del filtro, ya que en la frecuencia seleccionada, ya ha disminuido su magnitud a la mitad, por ende, se trabajan con valores un poco antes para evitar la perdida de la señal. 

**9. CONVERSIÓN DE ANOTACIONES A EVENTOS**

Las anotaciones del archivo, los cuales son marcadores temporales, se convierten en eventos discretos usando MNE. Se asignan etiquetas numéricas a T0 = 1, T1 = 2 y T2 = 3. Esto transforma información textual en algo computacionalmente manejable para segmentar la señal.

**10. CREACIÓN DE ÉPOCAS**

Se segmenta la señal continua en ventanas temporales, llamadas épocas, alrededor de cada evento, desde -1 a 4 segundos, esto con le fin de tomar un segundo del segmento anterior y 4 del siguiente para tener mas datos para hacer el PSD. Esto permite analizar respuestas neuronales alineadas al estímulo o tarea. 

**11. CLASIFICACIÓN SEGÚN RUN**

Se clasifica cada archivo según el tipo de tarea:

- Runs 3, 7, 11 = movimiento real.

- Runs 4, 8, 12 = imaginación motora.

Esto es clave porque el mismo tipo de evento (T1, T2) puede representar condiciones diferentes dependiendo del run.

**12. ALMACENAMIENTO DE ÉPOCAS**

Las épocas se almacenan en la estructura ***datos_epochs***, separadas por sujeto, tipo de tarea (real/imaginación) y tipo de evento (T0, T1, T2). Esto deja los datos listos para análisis posterior o entrenamiento de modelos de clasificación.

**13. GUARDAR RAW (ORIGINAL + FILTRADA)**

Se almacena tanto la señal EEG original como la señal ya filtrada para cada archivo procesado. Esto permite reutilizar los datos sin tener que repetir etapas costosas como la carga, el remuestreo y el filtrado. Se guarda una estructura organizada por sujeto y tipo de tarea (real o imaginación), lo que facilita posteriormente la visualización comparativa entre señal cruda, señal filtrada y épocas, manteniendo eficiencia computacional.

**14. CONTEO DE ÉPOCAS POR SUJETO**

Se cuentan cuántas épocas hay por sujeto en cada condición relevante, T1 y T2, tanto real como imaginado. Esto sirve para verificar balance de datos, detectar sujetos con pocos ensayos y validar integridad del dataset, todo con el fin de realizar las mediciones estadisticas necesarias.

**15. CREACIÓN DEL DATAFRAME PRIMARIO**

Se convierte la lista conteo en un DataFrame de pandas. Esto facilita la visualicion de los datos con el fin de tomar las decisiones necesarias.

**16. RETORNO DE RESULTADOS**

La función retorna:

- ***datos_epochs:*** estructura completa con las épocas organizadas.

- ***df_conteo:*** resumen cuantitativo por sujeto.

- ***datos_raw:*** datos crudos para evitar tener que volver a procesar los datos y poder trabajar con estos.

Esto permite separar claramente datos crudos procesados de métricas descriptivas.

```python
def cargar_datos_eeg(ruta_base, runs_validos):

    # ==========================================================
    # 1. IDENTIFICACION DE SUJETOS
    # ==========================================================

    sujetos = [s for s in os.listdir(ruta_base)
               if os.path.isdir(os.path.join(ruta_base, s))]
    sujetos.sort()

    datos_epochs = {}
    datos_raw = {}  
    conteo = []
    # Para no imprimir varias veces el remuestreo del mismo sujeto
    sujetos_resampleados = set()

    # ==========================================================
    # 2. RECORRIDO DE SUJETOS
    # ==========================================================

    for sujeto in sujetos:

        ruta_sujeto = os.path.join(ruta_base, sujeto)

        datos_epochs[sujeto] = {
            'real': {'T0': [], 'T1': [], 'T2': []},
            'imaginacion': {'T0': [], 'T1': [], 'T2': []}
        }

        datos_raw[sujeto] = {
            'real': [],
            'imaginacion': []
        }

        # ==========================================================
        # 3. RECORRIDO DE ARCHIVOS EDF
        # ==========================================================

        for archivo in sorted(os.listdir(ruta_sujeto)):

            if not archivo.endswith('.edf'):
                continue

            if not any(r in archivo for r in runs_validos):
                continue

            run = int(archivo.split('R')[1][:2])
            ruta_archivo = os.path.join(ruta_sujeto, archivo)

            # ==========================================================
            # 4. CARGA DEL ARCHIVO EEG
            # ==========================================================

            raw = mne.io.read_raw_edf(
                ruta_archivo,
                preload=True,
                verbose=False
            )

            raw_original = raw.copy()

            # ==========================================================
            # 5. LIMPIEZA DE NOMBRES DE CANALES
            # ==========================================================

            raw.rename_channels(lambda x: x.strip().replace('.', ''))

            # ==========================================================
            # 6. SELECCION DE ELECTRODOS DE CORTEZA MOTORA
            # ==========================================================

            raw.pick(['C3', 'C4'])

            # ==========================================================
            # 7. VERIFICACION Y AJUSTE DE FRECUENCIA DE MUESTREO
            # ==========================================================

            fs = raw.info['sfreq']

            if fs != 160:

                if sujeto not in sujetos_resampleados:
                    print(f"{sujeto} fue remuestreado de {fs} Hz a 160 Hz")
                    sujetos_resampleados.add(sujeto)

                raw.resample(160, npad="auto")

            # ==========================================================
            # 8. FILTRADO DE LA SEÑAL
            # ==========================================================

            raw.filter(
                l_freq=7,
                h_freq=40,
                method='fir',
                verbose=False
            )

            raw_filtrada = raw.copy()

            # ==========================================================
            # 9. CONVERSION DE ANOTACIONES A EVENTOS
            # ==========================================================

            events, event_id = mne.events_from_annotations(
                raw,
                event_id={'T0': 1, 'T1': 2, 'T2': 3},
                verbose=False
            )

            # ==========================================================
            # 10. CREACION DE EPOCAS
            # ==========================================================

            epochs = mne.Epochs(
                raw,
                events,
                event_id={'T0': 1, 'T1': 2, 'T2': 3},
                tmin=-1.0,
                tmax=4.0,
                baseline=None,
                preload=True,
                verbose=False
            )

            # ==========================================================
            # 11. CLASIFICACION SEGUN RUN
            # ==========================================================

            if run in [3, 7, 11]:
                tipo = 'real'

            elif run in [4, 8, 12]:
                tipo = 'imaginacion'

            else:
                continue

            # ==========================================================
            # 12. ALMACENAMIENTO DE EPOCAS
            # ==========================================================

            datos_epochs[sujeto][tipo]['T0'].append(epochs['T0'])
            datos_epochs[sujeto][tipo]['T1'].append(epochs['T1'])
            datos_epochs[sujeto][tipo]['T2'].append(epochs['T2'])

            # ==========================================================
            # 13. GUARDAR RAW (ORIGINAL + FILTRADA)
            # ==========================================================

            datos_raw[sujeto][tipo].append({
                'archivo': archivo,
                'raw_original': raw_original,
                'raw_filtrada': raw_filtrada
            })



        # ==========================================================
        # 14. CONTEO DE EPOCAS POR SUJETO
        # ==========================================================

        real_T1 = sum(len(e) for e in datos_epochs[sujeto]['real']['T1'])
        real_T2 = sum(len(e) for e in datos_epochs[sujeto]['real']['T2'])

        imag_T1 = sum(len(e) for e in datos_epochs[sujeto]['imaginacion']['T1'])
        imag_T2 = sum(len(e) for e in datos_epochs[sujeto]['imaginacion']['T2'])

        conteo.append({
            'Sujeto': sujeto,
            'Real_T1': real_T1,
            'Real_T2': real_T2,
            'Imag_T1': imag_T1,
            'Imag_T2': imag_T2
        })

    # ==========================================================
    # 15. CREACION DEL DATAFRAME PRIMARIO
    # ==========================================================

    df_conteo = pd.DataFrame(conteo)

    # ==========================================================
    # 16. RETORNO DE RESULTADOS
    # ==========================================================
    print(f"Se han procesado correctamente {len(sujetos)} sujetos.")
    return datos_epochs, df_conteo, datos_raw
```

Se selecciona la lista de epochs correspondiente al sujeto y tipo de tarea, y luego se toma la época del mismo índice que el run actual. Esto mantiene coherencia entre señal continua y segmentos temporales.

**5. EXTRAER DATOS**

Se extraen los vectores de tiempo (times) y las matrices de datos de las señales: original, filtrada y de la época. get_data() devuelve las amplitudes por canal, lo que permite trabajar directamente con arrays numéricos para graficación o análisis posterior. En el caso de la época, se accede al primer trial.

**6. GRÁFICAS COMPARATIVAS**

Se generan tres subgráficas en una misma figura para comparar:

- Señal original (cruda)

- Señal filtrada (7–40 Hz)

- Una época segmentada (T1)

Esto permite visualizar el efecto del filtrado y la segmentación en el dominio temporal. Es especialmente útil en BCI para validar que los patrones relevantes (mu y beta) se mantienen tras el preprocesamiento.

```python
def visualizar(datos_raw, datos_epochs, tipo):

    # ===============================
    # 1. SELECCION DE SUJETOS
    # ===============================
    sujetos = sorted(datos_raw.keys())

    for sujeto in sujetos:

        # ===============================
        # 2. RECORRIDO DE RUNS
        # ===============================
        runs = datos_raw[sujeto][tipo]

        for i, run_data in enumerate(runs):

            archivo = run_data['archivo']
            print(f"Sujeto: {sujeto}")
            print(f"Archivo: {archivo}")

            # ===============================
            # 3. OBTENER RAW
            # ===============================
            raw_original = run_data['raw_original']
            raw_filtrada = run_data['raw_filtrada']

            # Nos aseguramos de usar solo C3 y C4 (igual que antes)
            raw_original = raw_original.copy().rename_channels(lambda x: x.strip().replace('.', '')).pick(['C3', 'C4'])
            raw_filtrada = raw_filtrada.copy().rename_channels(lambda x: x.strip().replace('.', '')).pick(['C3', 'C4'])

            # ===============================
            # 4. OBTENER EPOCAS
            # ===============================
            epochs_list = datos_epochs[sujeto][tipo]['T1']

            if len(epochs_list) == 0:
                continue

            epochs = epochs_list[i]  # mismo índice que el run
            epoca = epochs[0]

            # ===============================
            # 5. EXTRAER DATOS
            # ===============================
            t_raw = raw_original.times
            t_filt = raw_filtrada.times
            t_epoca = epoca.times

            señal_raw = raw_original.get_data()
            señal_filt = raw_filtrada.get_data()
            señal_epoca = epoca.get_data()[0]

            # ===============================
            # 6. GRAFICAS COMPARATIVAS
            # ===============================
            plt.figure(figsize=(12, 8))

            # ORIGINAL
            plt.subplot(3,1,1)
            plt.plot(t_raw, señal_raw[0], label="C3")
            plt.plot(t_raw, señal_raw[1], label="C4")
            plt.title(f"Señal Original (C3 vs C4) - {tipo} - Brazo IZQ")
            plt.legend()

            # FILTRADA
            plt.subplot(3,1,2)
            plt.plot(t_filt, señal_filt[0], label="C3")
            plt.plot(t_filt, señal_filt[1], label="C4")
            plt.title(f"Señal Filtrada (7–40 Hz) - {tipo} - Brazo IZQ")
            plt.legend()

            # EPOCA
            plt.subplot(3,1,3)
            plt.plot(t_epoca, señal_epoca[0], label="C3")
            plt.plot(t_epoca, señal_epoca[1], label="C4")
            plt.title(f"Época T1 (C3 vs C4) - {tipo} - Brazo IZQ")
            plt.legend()

            plt.tight_layout()
            plt.show()

            return
```

## **3.2 CALCULO DE LA PSD**



**1. INICIALIZACIÓN DE ESTRUCTURAS**

Se crean diccionarios y listas vacías para almacenar los resultados del análisis espectral por sujeto y para almacenar información auxiliar utilizada en la generación de gráficas. Esto permite separar claramente los datos numéricos de los datos de visualización.

**2. RECORRIDO DE SUJETOS**

Se itera sobre todos los sujetos disponibles en el dataset. Para cada uno se inicializa una estructura jerárquica que organiza los resultados por tipo de tarea (real o imaginación) y por evento (T1 y T2), asegurando consistencia en la organización de datos.

**3. RECORRIDO DE CONDICIONES EXPERIMENTALES**

Se recorren todas las combinaciones posibles de tipo de movimiento y evento. Esto garantiza que cada condición experimental sea procesada de forma independiente, evitando mezclas entre tareas fisiológicamente distintas.

**4. CONCATENACIÓN Y SELECCIÓN DE CANALES**

Las épocas EEG correspondientes a una condición se concatenan en una sola señal continua y se seleccionan únicamente los canales de interés (C3 y C4). Esto reduce dimensionalidad y enfoca el análisis en regiones motoras relevantes.

**5. CÁLCULO DE PSD (WELCH)**

Se estima la densidad espectral de potencia utilizando el método de Welch, que proporciona una estimación más estable frente al ruido al promediar ventanas solapadas de la señal.

**6. DEFINICIÓN DE BANDAS FRECUENCIALES**

Se delimitan las bandas Mu (8–13 Hz) y Beta (13–30 Hz), que son las más relevantes en tareas de imaginación motora. A partir de estas bandas se calcula la potencia promedio.

**7. ALMACENAMIENTO DE RESULTADOS**

Los resultados se almacenan en una estructura jerárquica por sujeto, tipo de tarea, evento y canal. Para cada canal se guarda la PSD completa y las potencias escalares de las bandas Mu y Beta.

**8. PREPARACIÓN DE DATOS PARA VISUALIZACIÓN**

Se seleccionan únicamente los primeros sujetos para graficación con el fin de evitar sobrecarga visual. Los datos se separan por evento (T1 y T2) para su posterior representación.

**9. GENERACIÓN DE GRÁFICAS FINALES**

Se define una función interna que organiza las gráficas en subplots dinámicos, permitiendo visualizar múltiples sujetos simultáneamente de forma estructurada.

**10. PLOTEO DE PSD POR CANAL**

Se grafican las PSD de cada canal junto con las bandas Mu y Beta resaltadas visualmente, facilitando la interpretación espectral de la señal EEG.

**11. LIMPIEZA DE SUBPLOTS VACÍOS**

Se eliminan los ejes no utilizados en la cuadrícula de gráficos para mantener una visualización limpia y profesional.

**12. EJECUCIÓN DE GRÁFICAS POR CONDICIÓN**

Se generan dos visualizaciones principales: una para el brazo izquierdo (T1) y otra para el brazo derecho (T2), permitiendo comparar condiciones motoras de forma separada.

**13. RETORNO FINAL**

Se devuelve la estructura completa de resultados para su uso en análisis posteriores, clasificación o construcción de modelos.

```python
def Calcular_PSD(datos_epochs, canales=['C3','C4'], graficar=True):

    # ==========================================================
    # 1. INICIALIZACIÓN DE ESTRUCTURAS
    # ==========================================================
    resultados = {}
    plots_T1 = []
    plots_T2 = []

    # ==========================================================
    # 2. RECORRIDO DE SUJETOS
    # ==========================================================
    for idx, sujeto in enumerate(sorted(datos_epochs.keys())):

        resultados[sujeto] = {
            'real': {'T1': {}, 'T2': {}},
            'imaginacion': {'T1': {}, 'T2': {}}
        }

        # ==========================================================
        # 3. RECORRIDO DE CONDICIONES EXPERIMENTALES
        # ==========================================================
        for tipo in ['real', 'imaginacion']:
            for evento in ['T1', 'T2']:

                epochs_list = datos_epochs[sujeto][tipo][evento]

                if len(epochs_list) == 0:
                    continue

                # ==========================================================
                # 4. CONCATENACIÓN Y SELECCIÓN DE CANALES
                # ==========================================================
                epochs = mne.concatenate_epochs(epochs_list).copy().pick(canales)

                data = epochs.get_data()
                sfreq = epochs.info['sfreq']

                # ==========================================================
                # 5. CÁLCULO DE PSD (WELCH)
                # ==========================================================
                psd, freqs = psd_array_welch(
                    data, sfreq, fmin=1, fmax=40, verbose=False
                )

                psd_mean = psd.mean(axis=0)

                # ==========================================================
                # 6. DEFINICIÓN DE BANDAS FRECUENCIALES
                # ==========================================================
                mu_idx = (freqs >= 8) & (freqs <= 13)
                beta_idx = (freqs >= 13) & (freqs <= 30)

                mu = psd_mean[:, mu_idx].mean(axis=1)
                beta = psd_mean[:, beta_idx].mean(axis=1)

                # ==========================================================
                # 7. ALMACENAMIENTO DE RESULTADOS
                # ==========================================================
                resultados[sujeto][tipo][evento] = {
                    canal: {
                        'psd': psd_mean[i],
                        'mu': mu[i],
                        'beta': beta[i]
                    }
                    for i, canal in enumerate(canales)
                }

                # ==========================================================
                # 8. PREPARACIÓN DE DATOS PARA GRÁFICAS
                # ==========================================================
                if graficar and idx < 2:

                    info_plot = {
                        'sujeto': sujeto,
                        'tipo': tipo,
                        'freqs': freqs,
                        'psd': psd_mean
                    }

                    if evento == 'T1':
                        plots_T1.append(info_plot)
                    else:
                        plots_T2.append(info_plot)

    # ==========================================================
    # 9. GENERACIÓN DE GRÁFICAS FINALES
    # ==========================================================
    if graficar:

        def graficar_subplot(lista_plots, titulo):

            n = len(lista_plots)
            cols = 2
            filas = int(np.ceil(n / cols))

            fig, axes = plt.subplots(filas, cols, figsize=(12, 4*filas))
            axes = axes.flatten()

            # ==========================================================
            # 10. PLOTEO DE PSD POR CANAL
            # ==========================================================
            for i, plot in enumerate(lista_plots):

                ax = axes[i]

                for ch_idx in range(len(canales)):
                    ax.plot(plot['freqs'], plot['psd'][ch_idx],
                            label=canales[ch_idx])

                ax.axvspan(8, 13, color='skyblue', alpha=0.3, label="Mu (8–13 Hz)")
                ax.axvspan(13, 30, color='salmon', alpha=0.3, label="Beta (13–30 Hz)")

                ax.set_title(f"{plot['sujeto']} - {plot['tipo']}")
                ax.set_xlabel("Frecuencia (Hz)")
                ax.set_ylabel("PSD")
                ax.legend()

            # ==========================================================
            # 11. LIMPIEZA DE SUBPLOTS VACÍOS
            # ==========================================================
            for j in range(i+1, len(axes)):
                fig.delaxes(axes[j])

            fig.suptitle(titulo)
            plt.tight_layout()
            plt.show()

        # ==========================================================
        # 12. EJECUCIÓN DE GRÁFICAS POR CONDICIÓN
        # ==========================================================
        graficar_subplot(plots_T1, "PSD - Brazo izquierdo (T1)")
        graficar_subplot(plots_T2, "PSD - Brazo derecho (T2)")

    # ==========================================================
    # 13. RETORNO FINAL
    # ==========================================================
    return resultados
```

## Calculo de la densidad promedio

**1. Definición de la función e importaciones**

Este bloque define la función graficar_bandas, la cual recibe como entrada la estructura resultados_psd y una lista de canales (por defecto C3 y C4).

**2. Función interna calcular_promedios**

Este bloque encapsula la lógica de procesamiento de datos.

Su objetivo es:

- Recorrer todos los sujetos
- Extraer valores de potencia en bandas Mu y Beta
- Separar entre condiciones: real e imaginación
- Calcular promedios por sujeto y luego globales

Permite además:

Filtrar por evento específico (T1 o T2)
O trabajar con ambos eventos si no se especifica

**3. Inicialización de acumuladores**

Se crean listas para almacenar los valores de:

Mu real e imaginación
Beta real e imaginación

Estas listas acumulan resultados de todos los sujetos antes del promedio final.

**4. Recorrido por sujetos**

Se itera sobre cada sujeto en resultados_psd.

Dentro de este bloque:

Se inicializan listas temporales por sujeto
Se define qué eventos se van a analizar (uno específico o ambos)

**5. Extracción de datos por evento**

Para cada evento:

Se accede a los datos de condición real e imaginación
Se valida que existan datos

Luego:

Se extraen los valores de banda Mu y Beta por canal
Se almacenan como listas (una por canal)

**6. Promedio por sujeto**

Una vez recolectados los datos de un sujeto:

Se calcula el promedio de Mu y Beta (real e imaginación)
Se agregan a las listas globales

Si el sujeto no tiene datos válidos, se omite.

**7. Promedio global**

Se calculan los promedios finales:

Entre todos los sujetos
Para cada condición y banda

Se retorna una tupla con estos valores.

**8. Función interna graficar**

Este bloque se encarga exclusivamente de la visualización.

Construye:

Dos subgráficas (Mu y Beta)
Barras comparativas entre condición real e imaginación
Etiquetas de canales y títulos

**9. Gráfica para T1 (mano izquierda)**

Se calculan los promedios solo para el evento T1 y se genera su respectiva gráfica.

**10. Gráfica para T2 (mano derecha)**

Se repite el proceso para el evento T2.

**11. Gráfica global**

Se calculan promedios usando ambos eventos (T1 y T2) y se genera una gráfica global.

```python
def graficar_bandas(resultados_psd, canales=['C3','C4']):

    # ======================================================
    # FUNCIÓN: CALCULAR PROMEDIOS
    # ======================================================
    # Calcula promedios de bandas Mu y Beta para:
    # - Condición real
    # - Condición de imaginación
    # Puede filtrar por evento (T1, T2) o usar ambos
    def calcular_promedios(evento=None):

        # Listas globales para acumular datos de todos los sujetos
        mu_real_all = []
        mu_imag_all = []
        beta_real_all = []
        beta_imag_all = []

        # ======================================================
        # RECORRIDO POR SUJETOS
        # ======================================================
        for sujeto in resultados_psd:

            # Listas temporales por sujeto
            mu_real_suj = []
            mu_imag_suj = []
            beta_real_suj = []
            beta_imag_suj = []

            # Definición de eventos a evaluar
            eventos = [evento] if evento else ['T1','T2']

            # ======================================================
            # RECORRIDO POR EVENTOS
            # ======================================================
            for ev in eventos:

                # Acceso a datos de cada condición
                datos_real = resultados_psd[sujeto]['real'][ev]
                datos_imag = resultados_psd[sujeto]['imaginacion'][ev]

                # Validación de existencia de datos
                if not datos_real or not datos_imag:
                    continue

                # Extracción de banda Mu por canal
                mu_real_suj.append([datos_real[c]['mu'] for c in canales])
                mu_imag_suj.append([datos_imag[c]['mu'] for c in canales])

                # Extracción de banda Beta por canal
                beta_real_suj.append([datos_real[c]['beta'] for c in canales])
                beta_imag_suj.append([datos_imag[c]['beta'] for c in canales])

            # Si no hay datos válidos, se omite el sujeto
            if len(mu_real_suj) == 0:
                continue

            # ======================================================
            # PROMEDIO POR SUJETO
            # ======================================================
            mu_real_all.append(np.mean(mu_real_suj, axis=0))
            mu_imag_all.append(np.mean(mu_imag_suj, axis=0))

            beta_real_all.append(np.mean(beta_real_suj, axis=0))
            beta_imag_all.append(np.mean(beta_imag_suj, axis=0))

        # ======================================================
        # PROMEDIO GLOBAL ENTRE SUJETOS
        # ======================================================
        return (
            np.mean(mu_real_all, axis=0),
            np.mean(mu_imag_all, axis=0),
            np.mean(beta_real_all, axis=0),
            np.mean(beta_imag_all, axis=0),
        )

    # ======================================================
    # FUNCIÓN: GRAFICAR RESULTADOS
    # ======================================================
    # Genera gráficas de barras para comparar:
    # - Condición real vs imaginación
    # - Bandas Mu y Beta
    def graficar(mu_r, mu_i, beta_r, beta_i, titulo):

        x = np.arange(len(canales))

        # Creación de figura con dos subgráficas
        fig, axes = plt.subplots(1, 2, figsize=(12,5))

        # ===============================
        # BANDA MU
        # ===============================
        axes[0].bar(x - 0.2, mu_r, width=0.4, label='Real')
        axes[0].bar(x + 0.2, mu_i, width=0.4, label='Imaginación')
        axes[0].set_xticks(x)
        axes[0].set_xticklabels(canales)
        axes[0].set_title("Banda Mu (8–13 Hz)")
        axes[0].legend()

        # ===============================
        # BANDA BETA
        # ===============================
        axes[1].bar(x - 0.2, beta_r, width=0.4, label='Real')
        axes[1].bar(x + 0.2, beta_i, width=0.4, label='Imaginación')
        axes[1].set_xticks(x)
        axes[1].set_xticklabels(canales)
        axes[1].set_title("Banda Beta (13–30 Hz)")
        axes[1].legend()

        # Título general
        fig.suptitle(titulo)

        # Ajuste de layout y visualización
        plt.tight_layout()
        plt.show()

    # ======================================================
    # GRAFICACIÓN POR EVENTO
    # ======================================================

    # ----- T1 (Brazo izquierdo) -----
    mu_r, mu_i, beta_r, beta_i = calcular_promedios('T1')
    graficar(mu_r, mu_i, beta_r, beta_i, "Brazo Izquierdo (T1)")

    # ----- T2 (Brazo derecho) -----
    mu_r, mu_i, beta_r, beta_i = calcular_promedios('T2')
    graficar(mu_r, mu_i, beta_r, beta_i, "Brazo Derecho (T2)")

    # ----- Global (todos los eventos) -----
    mu_r, mu_i, beta_r, beta_i = calcular_promedios()
    graficar(mu_r, mu_i, beta_r, beta_i, "Promedio Global")
```

## **3.3 GENERAR RUTINA Y DATAFRAME**

**1. Definición de la función e importación**

Este bloque define la función construir_dataframe_final, que recibe como entrada un diccionario estructurado (resultados_psd) con información de potencia espectral (PSD) organizada por sujeto, tipo de tarea (real o imaginación) y evento (T1, T2).

También se importa la librería pandas, que se utilizará para construir el DataFrame final.

**2. Inicialización de estructura de almacenamiento**

Aquí se crea una lista vacía llamada filas.

Esta lista va a almacenar cada fila del DataFrame como un diccionario independiente. Luego, todas estas filas se convertirán en un DataFrame estructurado.

**3. Recorrido jerárquico de los datos**

Este es el núcleo del código.

Se recorren los datos en tres niveles:

- Sujetos
- Tipo de tarea (real o imaginacion)
- Evento (T1 o T2)

En cada iteración se accede a un subconjunto específico de datos (datos_evento).

Además, se hace una validación: si no hay datos para ese evento, se omite (continue).

**4. Construcción de cada fila**

Aquí se construye un diccionario llamado fila, que representa una observación del DataFrame.

Incluye:

- Identificación del sujeto
- Tipo de tarea
- Evento

Luego, se recorren los canales (por defecto C3 y C4) y se extraen las bandas de frecuencia:

- Banda Mu
- Banda Beta

Si algún canal no existe en los datos, se maneja el error con try-except y se asigna None.

**5. Almacenamiento de filas**

Cada fila construida se agrega a la lista filas.

Esto permite acumular todas las observaciones antes de construir el DataFrame final.

**6. Creación del DataFrame y retorno**

- Se convierte la lista de diccionarios en un DataFrame de pandas
- Se retorna el DataFrame listo para análisis posterior

```python
def Rutina(ruta_base, runs_validos, canales=['C3','C4']):

    print("Se procesan, filtran y almacenan los datos..........\n")
    datos_epochs, df_conteo, datos_raw = cargar_datos_eeg(ruta_base, runs_validos)
    #----------------------------------------------------------------------------------------------------------
    print("\nSe visualiza la señal en crudo comparando con la filtrada y con C3 y C4 solo para T1.........\n")
    visualizar(datos_raw, datos_epochs, tipo="real")
    visualizar(datos_raw, datos_epochs, tipo="imaginacion")
    #---------------------------------------------------------------------------------------------------------
    print("Se observan las graficas calculadas para la PSD en las canales C3 y C4 para real e imanginario junto con las bandas mu y beta para los sujetos 1 y 2......\n")
    resultados_psd = Calcular_PSD(datos_epochs)
    #--------------------------------------------------------------------------------------------------------
    print("Se observan los graficos de barra de la potencia promedio de cada canal para cada prueba de todos los sujetos......\n")
    graficar_bandas(resultados_psd)
    #--------------------------------------------------------------------------------------------------------

    # Lista donde se almacenarán todas las filas del DataFrame
    filas = []

    # ===============================
    # RECORRIDO PRINCIPAL DE LOS DATOS
    # ===============================
    # Se recorre la estructura jerárquica:
    # Sujeto -> Tipo (real/imaginacion) -> Evento (T1/T2)
    for sujeto in resultados_psd:

        for tipo in ['real', 'imaginacion']:

            for evento in ['T1', 'T2']:

                # Acceso a los datos específicos del evento
                datos_evento = resultados_psd[sujeto][tipo][evento]

                # Si no hay datos, se omite este caso
                if not datos_evento:
                    continue

                # ===============================
                # CREACIÓN DE UNA FILA
                # ===============================
                # Se inicializa el diccionario con información base
                fila = {
                    'Sujeto': sujeto,
                    'Tipo': tipo,
                    'Tarea': evento
                }

                # ===============================
                # EXTRACCIÓN DE FEATURES POR CANAL
                # ===============================
                # Para cada canal se extraen las bandas Mu y Beta
                for canal in canales:
                    fila[f'Mu_{canal}'] = datos_evento[canal]['mu']
                    fila[f'Beta_{canal}'] = datos_evento[canal]['beta']

                # Se agrega la fila a la lista
                filas.append(fila)

    # ===============================
    # CREACIÓN DEL DATAFRAME FINAL
    # ===============================
    df = pd.DataFrame(filas)
    print("Se muestra el dataframe generado con los datos obtenidos de las muestras....\n")
    return df
```

## **4. ANALISIS DE RESULTADOS Y COMPROBACION DE HIPOTESIS**

```python
DF_Hipotesis1 = Dataframe_Completo[(Dataframe_Completo['Tarea'] == "T1")]

DF_Hipotesis1_real_mu_C3 = DF_Hipotesis1[(DF_Hipotesis1['Tipo'] == "real")]['Mu_C3']
DF_Hipotesis1_real_mu_C4 = DF_Hipotesis1[(DF_Hipotesis1['Tipo'] == "real")]['Mu_C4']

DF_Hipotesis1_real_beta_C3 = DF_Hipotesis1[(DF_Hipotesis1['Tipo'] == "real")]['Beta_C3']
DF_Hipotesis1_real_beta_C4 = DF_Hipotesis1[(DF_Hipotesis1['Tipo'] == "real")]['Beta_C4']

DF_Hipotesis1_imaginacion_mu_C3 = DF_Hipotesis1[(DF_Hipotesis1['Tipo'] == "imaginacion")]['Mu_C3']
DF_Hipotesis1_imaginacion_mu_C4 = DF_Hipotesis1[(DF_Hipotesis1['Tipo'] == "imaginacion")]['Mu_C4']

DF_Hipotesis1_imaginacion_beta_C3 = DF_Hipotesis1[(DF_Hipotesis1['Tipo'] == "imaginacion")]['Beta_C3']
DF_Hipotesis1_imaginacion_beta_C4 = DF_Hipotesis1[(DF_Hipotesis1['Tipo'] == "imaginacion")]['Beta_C4']
```

normalidad

```python
Datos_analizar = [
    DF_Hipotesis1_real_mu_C3,
    DF_Hipotesis1_real_mu_C4,
    DF_Hipotesis1_real_beta_C3,
    DF_Hipotesis1_real_beta_C4,
    DF_Hipotesis1_imaginacion_mu_C3,
    DF_Hipotesis1_imaginacion_mu_C4,
    DF_Hipotesis1_imaginacion_beta_C3,
    DF_Hipotesis1_imaginacion_beta_C4
]

nombres = [
    'DF_Hipotesis1_real_mu_C3',
    'DF_Hipotesis1_real_mu_C4',
    'DF_Hipotesis1_real_beta_C3',
    'DF_Hipotesis1_real_beta_C4',
    'DF_Hipotesis1_imaginacion_mu_C3',
    'DF_Hipotesis1_imaginacion_mu_C4',
    'DF_Hipotesis1_imaginacion_beta_C3',
    'DF_Hipotesis1_imaginacion_beta_C4'
]

for data, nombre in zip(Datos_analizar, nombres):

    data = np.array(data)

    # Normalizar
    data_std = (data - np.mean(data)) / np.std(data)

    stat, p = kstest(data_std, 'norm')

    if p > 0.05:
        print(f"Los datos de {nombre} son Normal (p = {p:.4f} > 0.05)")
    else:
        print(f"Los datos de {nombre} son No normal (p = {p:.4f} <= 0.05)")
```

**HIPOTESIS 1**

Se plantea la hipótesis de que la ejecución e imaginación del movimiento de la mano izquierda (T1) generan cambios en la potencia espectral de las señales EEG, particularmente en las bandas mu (8–13 Hz) y beta (13–30 Hz), evidenciando una desincronización en el hemisferio contralateral, es decir, en el electrodo C4. Este comportamiento se fundamenta en la organización contralateral del sistema motor, donde la corteza motora derecha controla el movimiento de la mano izquierda, y en fenómenos de desincronización relacionados con la activación cortical durante tareas motoras [11], [12].
Para comprobar esta hipótesis, se propone comparar la potencia espectral en los electrodos C3 y C4 bajo las condiciones de movimiento real e imaginación, utilizando gráficas de densidad espectral de potencia y análisis comparativos entre canales, los cuales han demostrado ser efectivos para evidenciar patrones de lateralización en estudios de interfaces cerebro–computador [12], [1].

diferenica con U de Mann-Whitney entre C3 y C4

```python
pares = [
    ("Real Mu", DF_Hipotesis1_real_mu_C4, DF_Hipotesis1_real_mu_C3),
    ("Real Beta", DF_Hipotesis1_real_beta_C4, DF_Hipotesis1_real_beta_C3),
    ("Imaginación Mu", DF_Hipotesis1_imaginacion_mu_C4, DF_Hipotesis1_imaginacion_mu_C3),
    ("Imaginación Beta", DF_Hipotesis1_imaginacion_beta_C4, DF_Hipotesis1_imaginacion_beta_C3),
]

for nombre, C4, C3 in pares:

    stat, p = wilcoxon(C4, C3)

    print(f"\n{nombre}")
    print(f"p = {p:.5f}")

    if p < 0.05: 
            print("Hay diferencia significativa")
    else:
            print("No hay diferencia significativa")
```

Este fenómeno se asocia con la desincronización relacionada con el evento (ERD), ampliamente reportada en la literatura como un indicador de activación cortical durante tareas motoras [11], [12].
Para comprobar esta hipótesis, se propone comparar la potencia espectral promedio en ambas bandas entre las condiciones de movimiento real e imaginación, evaluando si existen diferencias consistentes en los canales C3 y C4, los cuales están asociados a la corteza motora y presentan patrones de activación contralateral [12], [1].
