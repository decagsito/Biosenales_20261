### 1.1  Descripcion del proyecto

Este proyecto se centra en el análisis de señales electroencefalográficas (EEG) provenientes de una base de datos que registra la actividad cortical durante la ejecución y la imaginación motora de las extremidades superiores (mano derecha e izquierda) y estados de reposo. El propósito fundamental es determinar la viabilidad técnica de sistemas BCI no invasivos, fundamentándose en la neurofisiología del control motor para discernir patrones espaciales y espectrales específicos que permitan la clasificación de la intención del usuario [1].

### 1.2  Historia, Uso y Perspectivas Futuras de las BCI

La evolución de las interfaces cerebro-computadora (BCI) ha transitado desde una idea conceptual en los inicios de la tecnología digital hasta sistemas altamente sofisticados para la detección, registro y análisis de bioseñales. En la década de 1960, el uso de la electroencefalografía (EEG) estaba limitado a entornos de laboratorio debido a los requerimientos de equipamiento y a la susceptibilidad a artefactos; sin embargo, los avances en miniaturización electrónica y en amplificadores integrados en electrodos han permitido actualmente la adquisición de señales en entornos no controlados. Durante las últimas cuatro décadas, las BCI evolucionaron desde simples sistemas de comunicación hacia aplicaciones complejas de control asistivo. Los usos actuales de los BCIs incluyen la restauración motora a través de exoesqueletos robóticos y estimulación eléctrica funcional, la mejora cognitiva a través del neurofeedback y el entrenamiento de la atención, y herramientas de comunicación para personas con limitaciones físicas severas, exploradas principalmente dentro de entornos de investigación en lugar de la práctica clínica de rutina. En este contexto, las BCI se proyectan como herramientas de asistencia cada vez más integradas al individuo, con potencial para mitigar el aislamiento social y ampliar las capacidades de interacción entre el cerebro humano y los sistemas tecnológicos [2][3].

### 1.3  Actividad cerebral registrada por EEG durante movimientos e imaginación de movimiento de las manos y cómo se hace su análisis en sistemas BCI.

La imaginación motora (Motor Imagery, MI) es uno de los paradigmas más utilizados en sistemas de interfaz cerebro-computadora basados en EEG. Las MI implican simular movimientos mentalmente sin ejecución real, activando redes neuronales similares al movimiento físico, esto involucra regiones visuales y motoras asociadas, lo que facilita la planificación motora y potencialmente ayuda a la rehabilitación al promover la reorganización cortical y mejorar la función sensoriomotora. Esta activación se refleja en modulaciones de los ritmos sensoriomotores (SMR), que representan cambios en la sincronización de poblaciones neuronales durante la planificación y simulación de movimientos [4].

Los ritmos cerebrales de interés en tareas de MI se concentran principalmente en las bandas Mu (8–12 Hz) y beta (13–30 Hz). Durante la imaginación de movimiento, especialmente de las manos, se observa una desincronización de la banda Mu sobre la corteza sensoriomotora (event-related desynchronization, ERD), asociada a la activación cortical. De manera similar, la banda beta también presenta desincronización durante la planificación e imaginación del movimiento, seguida de fenómenos de resincronización al finalizar la tarea [5].

La adquisición de estas señales se realiza mediante electrodos de EEG ubicados sobre la corteza sensoriomotora, comúnmente en las posiciones C3 y C4 del sistema internacional 10–20, que corresponden a áreas motoras relacionadas con el control de las manos. Estas ubicaciones permiten captar con mayor precisión las modulaciones de los ritmos sensoriomotores asociadas a la imaginación de movimientos específicos [5].

En un sistema BCI basado en MI, el análisis de las señales se desarrolla en tres etapas principales: adquisición de señales, donde los electrodos registran la actividad cerebral; procesamiento de señales, que incluye filtrado y extracción de características relevantes en las bandas Mu y beta; y clasificación, donde algoritmos de aprendizaje automático identifican los patrones asociados a la tarea imaginada. El desempeño del sistema se evalúa mediante métricas como precisión, sensibilidad, especificidad y puntuación F1 [5].

### 1.4 Flujos de procesamiento propuestos para estos tipos de proyectos

Un sistema BCI típico se divide en adquisición, preprocesamiento, extracción de características, clasificación y control

En un sistema de Interfaz Cerebro–Computador (BCI), el proceso comienza con la **adquisición de la señal**, generalmente mediante electroencefalografía (EEG), que es una técnica no invasiva ampliamente utilizada en entornos académicos [6], [7]. Esta etapa implica el uso de electrodos colocados sobre el cuero cabelludo, junto con sistemas de amplificación y digitalización que permiten registrar la actividad eléctrica cerebral. Sin embargo, en esta fase surgen diversos problemas, como la presencia de ruido y artefactos, incluyendo interferencias por parpadeo (EOG), actividad muscular (EMG) y ruido eléctrico del entorno [7].

Posteriormente, se realiza el **preprocesamiento de la señal**, cuyo objetivo principal es mejorar la calidad del EEG eliminando o reduciendo el ruido. Para ello, se emplean técnicas como filtros pasa banda y filtros notch (generalmente a 50 o 60 Hz para eliminar interferencia eléctrica), así como métodos de eliminación de artefactos como el Análisis de Componentes Independientes (ICA) o técnicas de regresión [7], [8]. También es común aplicar procesos de normalización. Esta etapa es crítica debido a la baja relación señal/ruido característica de las señales EEG [6].

Una vez limpia la señal, se procede a la **extracción de características**, donde el EEG se transforma en un conjunto de variables representativas que pueden ser utilizadas por algoritmos de clasificación. Estas características pueden obtenerse en el dominio del tiempo (por ejemplo, media o varianza), en el dominio de la frecuencia (como la Transformada Rápida de Fourier o la densidad espectral de potencia) o mediante transformadas wavelet [7], [9]. Asimismo, es común trabajar con las bandas típicas del EEG (delta, theta, alfa, beta y gamma), las cuales son fundamentales en aplicaciones de BCI debido a su relación con distintos estados cognitivos [8].

A continuación, se lleva a cabo la **selección o reducción de características**, con el fin de disminuir la dimensionalidad de los datos y mejorar el rendimiento de los modelos. Entre las técnicas más utilizadas se encuentran el Análisis de Componentes Principales (PCA) y el Análisis Discriminante Lineal (LDA) [7]. En el caso específico de aplicaciones de imaginación motora, es muy común el uso de Patrones Espaciales Comunes (CSP), que permiten resaltar diferencias relevantes entre clases [8].

En la etapa de **clasificación**, se identifican los patrones cerebrales a partir de las características extraídas. Para ello, se utilizan algoritmos de aprendizaje automático como LDA, Máquinas de Vectores de Soporte (SVM) y K-Vecinos más Cercanos (KNN) [7], [9]. En enfoques más recientes, también se emplean redes neuronales profundas, como las redes convolucionales (CNN), destacándose arquitecturas especializadas como EEGNet [10].

Finalmente, el sistema culmina con la **interfaz o aplicación**, donde los resultados de la clasificación se traducen en acciones concretas. Estas pueden incluir el control de un cursor en pantalla, el manejo de prótesis, la interacción con videojuegos o la identificación de estados cognitivos del usuario [6].

### 2.1 Metodología de procesamiento y análisis

En este trabajo se consideran cuatro condiciones principales de estudio: el movimiento de la mano izquierda, el movimiento de la mano derecha, la imaginación del movimiento de la mano izquierda y la imaginación del movimiento de la mano derecha. Estas condiciones permiten explorar tanto la ejecución motora como los procesos cognitivos asociados a la imaginación motora, los cuales han demostrado generar patrones similares en el EEG, aunque con menor intensidad. El análisis se centra en la comparación de estas condiciones con el fin de evidenciar diferencias en los ritmos cerebrales asociados a cada una de ellas.

Para ello, se realiza una selección específica de electrodos ubicados sobre la corteza motora, particularmente C3, C4 y Cz, debido a su relevancia en la representación de movimientos voluntarios. La elección de estos canales se basa en el principio de lateralización cerebral, según el cual la actividad motora de un hemisferio se relaciona con el movimiento del lado contralateral del cuerpo. De esta manera, se espera que el movimiento de la mano derecha genere cambios más notorios en el electrodo C3, mientras que el movimiento de la mano izquierda afecte principalmente al electrodo C4.

El procesamiento de la señal inicia con una etapa de preprocesamiento orientada a mejorar la calidad del EEG. En primer lugar, se aplica un filtro pasa banda entre 8 y 30 Hz con el objetivo de aislar las bandas de interés asociadas a la actividad motora, particularmente los ritmos mu (8–13 Hz) y beta (13–30 Hz). Adicionalmente, se emplea un filtro notch para eliminar la interferencia de la red eléctrica 50/60 Hz. Posteriormente, las señales son segmentadas en ventanas temporales o épocas alineadas con los eventos experimentales, lo cual permite analizar de manera consistente la respuesta cerebral ante cada condición.

Una vez preprocesadas las señales, se procede a la extracción de características relevantes que permitan cuantificar la actividad cerebral. En este caso, se propone el uso de la densidad espectral de potencia como medida principal, ya que permite analizar la distribución de la energía de la señal en función de la frecuencia. A partir de esta representación, se calcula la potencia promedio en las bandas mu (8–13 Hz) y beta (13–30 Hz), las cuales están directamente relacionadas con la actividad del córtex motor.

El análisis de las características extraídas se realiza mediante una comparación entre las diferentes condiciones experimentales, tanto a nivel espectral como espacial. En el dominio espectral, se evalúan las diferencias en la potencia de las bandas mu y beta entre movimiento e imaginación, así como entre mano izquierda y derecha. En el dominio espacial, se analizan las variaciones entre los electrodos C3 y C4 para identificar patrones de lateralización. Este análisis puede complementarse con herramientas estadísticas, como pruebas t o análisis de varianza, con el fin de determinar si las diferencias observadas son significativas.

Adicionalmente, se propone la incorporación de un esquema de clasificación supervisada como mecanismo de validación del análisis. Para ello, se puede implementar un flujo basado en técnicas como Patrones Espaciales Comunes (CSP) para la extracción de características espaciales, seguido de un clasificador como el Análisis Discriminante Lineal (LDA) o Máquinas de Vectores de Soporte (SVM). Este enfoque permite evaluar si las características extraídas son suficientemente discriminativas para diferenciar automáticamente entre las condiciones de estudio, lo cual refuerza la evidencia de que existen diferencias consistentes en los ritmos cerebrales.

```python
import mne
mne.set_log_level('ERROR')
from mne.time_frequency import psd_array_welch
import numpy as np
import pandas as pd
import os
import matplotlib.pyplot as plt
from scipy import signal # Nos permite realizar preprocesmaiento de la señal
from scipy.stats import shapiro, levene, ttest_ind, mannwhitneyu
import seaborn as sns
```

```python
def cargar_datos_eeg(ruta_base, runs_validos):

    # ==========================================================
    # 1. IDENTIFICACION DE SUJETOS
    # ==========================================================

    sujetos = [s for s in os.listdir(ruta_base)
               if os.path.isdir(os.path.join(ruta_base, s))]
    sujetos.sort()

    datos_epochs = {}
    conteo = []

    # Para no imprimir varias veces el remuestreo del mismo sujeto
    sujetos_resampleados = set()



    # ==========================================================
    # 2. RECORRIDO DE SUJETOS
    # ==========================================================

    for sujeto in sujetos:

        ruta_sujeto = os.path.join(ruta_base, sujeto)

        datos_epochs[sujeto] = {
            'real': {'T0': [], 'T1': [], 'T2': []},
            'imaginacion': {'T0': [], 'T1': [], 'T2': []}
        }



        # ==========================================================
        # 3. RECORRIDO DE ARCHIVOS EDF
        # ==========================================================

        for archivo in sorted(os.listdir(ruta_sujeto)):

            if not archivo.endswith('.edf'):
                continue

            if not any(r in archivo for r in runs_validos):
                continue

            run = int(archivo.split('R')[1][:2])
            ruta_archivo = os.path.join(ruta_sujeto, archivo)



            # ==========================================================
            # 4. CARGA DEL ARCHIVO EEG
            # ==========================================================

            raw = mne.io.read_raw_edf(
                ruta_archivo,
                preload=True,
                verbose=False
            )



            # ==========================================================
            # 5. LIMPIEZA DE NOMBRES DE CANALES
            # Algunos EDF tienen puntos o espacios en los nombres
            # ==========================================================

            raw.rename_channels(lambda x: x.strip().replace('.', ''))



            # ==========================================================
            # 6. SELECCION DE ELECTRODOS DE CORTEZA MOTORA
            # ==========================================================

            raw.pick(['C3', 'C4', 'Cz'])



            # ==========================================================
            # 7. VERIFICACION Y AJUSTE DE FRECUENCIA DE MUESTREO
            # Algunos sujetos tienen Fs = 128 Hz
            # Se remuestrea todo a 160 Hz para uniformidad
            # ==========================================================

            fs = raw.info['sfreq']

            if fs != 160:

                if sujeto not in sujetos_resampleados:
                    print(f"{sujeto} fue remuestreado de {fs} Hz a 160 Hz")
                    sujetos_resampleados.add(sujeto)

                raw.resample(160, npad="auto")



            # ==========================================================
            # 8. FILTRADO DE LA SEÑAL
            # Filtro pasa banda 7–40 Hz
            # ==========================================================

            raw.filter(
                l_freq=7,
                h_freq=40,
                method='fir',
                verbose=False
            )



            # ==========================================================
            # 9. CONVERSION DE ANOTACIONES A EVENTOS
            # ==========================================================

            events, event_id = mne.events_from_annotations(
                raw,
                event_id={'T0': 1, 'T1': 2, 'T2': 3},
                verbose=False
            )



            # ==========================================================
            # 10. CREACION DE EPOCAS
            # Segmentacion de la señal (-1 a 4 segundos)
            # ==========================================================

            epochs = mne.Epochs(
                raw,
                events,
                event_id={'T0': 1, 'T1': 2, 'T2': 3},
                tmin=-1.0,
                tmax=4.0,
                baseline=None,
                preload=True,
                verbose=False
            )



            # ==========================================================
            # 11. CLASIFICACION SEGUN RUN
            # ==========================================================

            if run in [3, 7, 11]:
                tipo = 'real'

            elif run in [4, 8, 12]:
                tipo = 'imaginacion'

            else:
                continue



            # ==========================================================
            # 12. ALMACENAMIENTO DE EPOCAS
            # ==========================================================

            datos_epochs[sujeto][tipo]['T0'].append(epochs['T0'])
            datos_epochs[sujeto][tipo]['T1'].append(epochs['T1'])
            datos_epochs[sujeto][tipo]['T2'].append(epochs['T2'])



        # ==========================================================
        # 13. CONTEO DE EPOCAS POR SUJETO
        # ==========================================================

        real_T1 = sum(len(e) for e in datos_epochs[sujeto]['real']['T1'])
        real_T2 = sum(len(e) for e in datos_epochs[sujeto]['real']['T2'])

        imag_T1 = sum(len(e) for e in datos_epochs[sujeto]['imaginacion']['T1'])
        imag_T2 = sum(len(e) for e in datos_epochs[sujeto]['imaginacion']['T2'])

        conteo.append({
            'Sujeto': sujeto,
            'Real_T1': real_T1,
            'Real_T2': real_T2,
            'Imag_T1': imag_T1,
            'Imag_T2': imag_T2
        })



    # ==========================================================
    # 14. CREACION DEL DATAFRAME FINAL
    # ==========================================================

    df_conteo = pd.DataFrame(conteo)



    # ==========================================================
    # 15. MENSAJE FINAL
    # ==========================================================

    print(f"Se han procesado correctamente {len(sujetos)} sujetos.")



    # ==========================================================
    # 16. RETORNO DE RESULTADOS
    # ==========================================================

    return datos_epochs, df_conteo
```

```python
def visualizar(ruta_base, tipo):

    runs_real = ["R03", "R07", "R11"]
    runs_imag = ["R04", "R08", "R12"]
    runs_validos = runs_real if tipo == "real" else runs_imag

    sujetos = [s for s in os.listdir(ruta_base)
               if os.path.isdir(os.path.join(ruta_base, s))]
    sujetos.sort()

    for sujeto in sujetos:

        ruta_sujeto = os.path.join(ruta_base, sujeto)

        for archivo in sorted(os.listdir(ruta_sujeto)):

            if not archivo.endswith('.edf'):
                continue

            if not any(r in archivo for r in runs_validos):
                continue

            ruta_archivo = os.path.join(ruta_sujeto, archivo)

            print(f"Sujeto: {sujeto}")
            print(f"Archivo: {archivo}")

            # ===============================
            # 1. CARGA
            # ===============================
            raw = mne.io.read_raw_edf(ruta_archivo, preload=True, verbose=False)
            raw.rename_channels(lambda x: x.strip().replace('.', ''))
            raw.pick(['C3','C4'])

            raw_original = raw.copy()

            # ===============================
            # 2. RESAMPLE
            # ===============================
            if raw.info['sfreq'] != 160:
                raw.resample(160, npad="auto")

            # ===============================
            # 3. FILTRADO
            # ===============================
            raw.filter(7, 40, method='fir', verbose=False)
            raw_filtrada = raw.copy()

            # ===============================
            # 4. EVENTOS Y EPOCAS
            # ===============================
            events, _ = mne.events_from_annotations(
                raw,
                event_id={'T0': 1, 'T1': 2, 'T2': 3},
                verbose=False
            )

            epochs = mne.Epochs(
                raw,
                events,
                event_id={'T0': 1, 'T1': 2, 'T2': 3},
                tmin=-1.0,
                tmax=4.0,
                baseline=None,
                preload=True,
                verbose=False
            )

            epoca = epochs['T1'][0]

            # ===============================
            # 5. EXTRAER DATOS
            # ===============================
            t_raw = raw_original.times
            t_filt = raw_filtrada.times
            t_epoca = epoca.times

            señal_raw = raw_original.get_data()
            señal_filt = raw_filtrada.get_data()
            señal_epoca = epoca.get_data()[0]

            # ===============================
            # 6. GRAFICAS COMPARATIVAS
            # ===============================
            plt.figure(figsize=(12, 8))

            # ORIGINAL
            plt.subplot(3,1,1)
            plt.plot(t_raw, señal_raw[0], label="C3")
            plt.plot(t_raw, señal_raw[1], label="C4")
            plt.title(f"Señal Original (C3 vs C4) - {tipo}")
            plt.legend()

            # FILTRADA
            plt.subplot(3,1,2)
            plt.plot(t_filt, señal_filt[0], label="C3")
            plt.plot(t_filt, señal_filt[1], label="C4")
            plt.title(f"Señal Filtrada (7–40 Hz) - {tipo}")
            plt.legend()

            # EPOCA
            plt.subplot(3,1,3)
            plt.plot(t_epoca, señal_epoca[0], label="C3")
            plt.plot(t_epoca, señal_epoca[1], label="C4")
            plt.title(f"Época T1 (C3 vs C4) - {tipo}")
            plt.legend()

            plt.tight_layout()
            plt.show()

            return
```

```python
def graficar_psd(epochs, tipo):

    # ===============================
    # 1. EXTRAER DATOS
    # ===============================
    data = epochs.get_data()  # (n_epochs, n_canales, n_muestras)
    sfreq = epochs.info['sfreq']
    canales = epochs.ch_names

    # ===============================
    # 2. CALCULAR PSD (WELCH)
    # ===============================
    psds, freqs = psd_array_welch(
        data,
        sfreq=sfreq,
        fmin=1,
        fmax=40,
        verbose=False
    )

    # Promedio sobre épocas
    psd_mean = psds.mean(axis=0)

    # ===============================
    # 3. GRAFICAR
    # ===============================
    plt.figure(figsize=(10,5))

    for i, ch in enumerate(canales):
        plt.plot(freqs, psd_mean[i], label=ch)

    # Bandas EEG
    plt.axvspan(8, 13, color='skyblue', alpha=0.3, label="Mu (8–13 Hz)")
    plt.axvspan(13, 30, color='salmon', alpha=0.3, label="Beta (13–30 Hz)")

    plt.xlabel("Frecuencia (Hz)")
    plt.ylabel("Potencia")
    plt.title(f"Densidad Espectral de Potencia (PSD) - {tipo}")
    plt.legend()
    plt.show()
```

```python
def comparar_C3_C4(epochs_real, epochs_imag):

    # ===============================
    # 1. EXTRAER DATOS
    # ===============================
    data_r = epochs_real.get_data()
    data_i = epochs_imag.get_data()
    sfreq = epochs_real.info['sfreq']
    canales = epochs_real.ch_names

    # Índices de canales
    idx_C3 = canales.index('C3')
    idx_C4 = canales.index('C4')

    # ===============================
    # 2. PSD
    # ===============================
    psd_r, freqs = psd_array_welch(data_r, sfreq, fmin=1, fmax=40, verbose=False)
    psd_i, _ = psd_array_welch(data_i, sfreq, fmin=1, fmax=40, verbose=False)

    psd_r_mean = psd_r.mean(axis=0)
    psd_i_mean = psd_i.mean(axis=0)

    # ===============================
    # 3. GRAFICA
    # ===============================
    plt.figure(figsize=(10,5))

    # C3
    plt.plot(freqs, psd_r_mean[idx_C3], label="C3 Real")
    plt.plot(freqs, psd_i_mean[idx_C3], linestyle='--', label="C3 Imaginación")

    # C4
    plt.plot(freqs, psd_r_mean[idx_C4], label="C4 Real")
    plt.plot(freqs, psd_i_mean[idx_C4], linestyle='--', label="C4 Imaginación")

    # Bandas
    plt.axvspan(8, 13, color='skyblue', alpha=0.2, label="Mu")
    plt.axvspan(13, 30, color='salmon', alpha=0.2, label="Beta")

    plt.xlabel("Frecuencia (Hz)")
    plt.ylabel("Potencia")
    plt.title("PSD: Comparación C3 vs C4 (Real vs Imaginación)")
    plt.legend()
    plt.show()
```

```python
def calcular_potencia_bandas(epochs):
    """
    Calcula la potencia en bandas Mu (8–13 Hz) y Beta (13–30 Hz)
    a partir de un objeto Epochs de MNE.

    Retorna:
        potencia_mu   -> (n_epochs, n_canales)
        potencia_beta -> (n_epochs, n_canales)
    """

    # ===============================
    # 1. EXTRAER DATOS
    # ===============================
    data = epochs.get_data()  # (n_epochs, n_canales, n_muestras)
    sfreq = epochs.info['sfreq']

    # ===============================
    # 2. FFT
    # ===============================
    freqs = np.fft.rfftfreq(data.shape[2], d=1/sfreq)
    fft_vals = np.fft.rfft(data, axis=2)

    # ===============================
    # 3. PSD
    # ===============================
    psd = np.abs(fft_vals) ** 2

    # ===============================
    # 4. DEFINIR BANDAS
    # ===============================
    banda_mu = (freqs >= 8) & (freqs <= 13)
    banda_beta = (freqs >= 13) & (freqs <= 30)

    # ===============================
    # 5. POTENCIA POR BANDA
    # ===============================
    potencia_mu = psd[:, :, banda_mu].mean(axis=2)
    potencia_beta = psd[:, :, banda_beta].mean(axis=2)

    return potencia_mu, potencia_beta
```

```python
def analizar_todos_sujetos(datos_epochs):

    resultados = []

    for sujeto in datos_epochs:

        try:
            # ===============================
            # 1. CONCATENAR EPOCAS
            # ===============================
            epochs_real = mne.concatenate_epochs(
                datos_epochs[sujeto]['real']['T1']
            )

            epochs_imag = mne.concatenate_epochs(
                datos_epochs[sujeto]['imaginacion']['T1']
            )

            # ===============================
            # 2. CALCULAR POTENCIA
            # ===============================
            mu_real, beta_real = calcular_potencia_bandas(epochs_real)
            mu_imag, beta_imag = calcular_potencia_bandas(epochs_imag)

            # ===============================
            # 3. PROMEDIOS
            # ===============================
            mu_real_mean = mu_real.mean(axis=0)
            mu_imag_mean = mu_imag.mean(axis=0)

            beta_real_mean = beta_real.mean(axis=0)
            beta_imag_mean = beta_imag.mean(axis=0)

            # ===============================
            # 4. GUARDAR RESULTADOS
            # ===============================
            resultados.append({
                'sujeto': sujeto,
                'mu_real': mu_real_mean,
                'mu_imag': mu_imag_mean,
                'beta_real': beta_real_mean,
                'beta_imag': beta_imag_mean
            })

        except Exception:
            continue

    return resultados
```

```python
resultados = analizar_todos_sujetos(datos_epochs)
```

```python
mu_real_all = [] # Para almacenar la potencia de banda Mu en condición real de todos los sujetos
mu_imag_all = [] # Para almacenar la potencia de banda Mu en condición imaginación de todos los sujetos
beta_real_all = [] # Para almacenar la potencia de banda Beta en condición real de todos los sujetos
beta_imag_all = [] # Para almacenar la potencia de banda Beta en condición imaginación de todos los sujetos

for r in resultados:
    mu_real_all.append(r['mu_real'])
    mu_imag_all.append(r['mu_imag'])
    beta_real_all.append(r['beta_real'])
    beta_imag_all.append(r['beta_imag'])

mu_real_all = np.array(mu_real_all) 
mu_imag_all = np.array(mu_imag_all)
beta_real_all = np.array(beta_real_all)
beta_imag_all = np.array(beta_imag_all)

# Promedio entre sujetos
mu_real_mean = mu_real_all.mean(axis=0) # Esto me da un vector con la potencia promedio en banda Mu para cada canal (C3, C4, Cz) en condición real
mu_imag_mean = mu_imag_all.mean(axis=0) # Esto me da un vector con la potencia promedio en banda Mu para cada canal (C3, C4, Cz) en condición imaginación

beta_real_mean = beta_real_all.mean(axis=0) # Esto me da un vector con la potencia promedio en banda Beta para cada canal (C3, C4, Cz) en condición real
beta_imag_mean = beta_imag_all.mean(axis=0) # Esto me da un vector con la potencia promedio en banda Beta para cada canal (C3, C4, Cz) en condición imaginación

canales = ['C3','C4','Cz']
x = np.arange(len(canales))

plt.figure(figsize=(10,5))

plt.bar(x - 0.2, mu_real_mean, width=0.4, label='Mu Real')
plt.bar(x + 0.2, mu_imag_mean, width=0.4, label='Mu Imaginación')

plt.xticks(x, canales)
plt.title("Comparación banda Mu")
plt.legend()
plt.show()


plt.figure(figsize=(10,5))

plt.bar(x - 0.2, beta_real_mean, width=0.4, label='Beta Real')
plt.bar(x + 0.2, beta_imag_mean, width=0.4, label='Beta Imaginación')

plt.xticks(x, canales)
plt.title("Comparación banda Beta")
plt.legend()
plt.show()
```

Se calculó la densidad espectral de potencia de las señales EEG mediante la transformada rápida de Fourier, y posteriormente se estimó la potencia promedio en las bandas mu (8–13 Hz) y beta (13–30 Hz) para todos los sujetos.

```python
def construir_dataframe_psd(datos_epochs):

    registros = []

    for sujeto in datos_epochs:

        for condicion in ['real', 'imaginacion']:

            try:
                lista_epochs = datos_epochs[sujeto][condicion]['T1']

                # Validar que haya datos
                if len(lista_epochs) == 0:
                    continue

                # Unir épocas
                epochs = mne.concatenate_epochs(lista_epochs)

                # Calcular potencia
                pot_mu, pot_beta = calcular_potencia_bandas(epochs)

                # Promedio sobre épocas
                pot_mu_mean = pot_mu.mean(axis=0)
                pot_beta_mean = pot_beta.mean(axis=0)

                canales = epochs.ch_names

                # Guardar por canal
                for i, canal in enumerate(canales):

                    registros.append({
                        'Sujeto': sujeto,
                        'Condicion': condicion,
                        'Canal': canal,
                        'Potencia_Mu': pot_mu_mean[i],
                        'Potencia_Beta': pot_beta_mean[i]
                    })

            except Exception:
                continue

    df = pd.DataFrame(registros)

    return df
```
