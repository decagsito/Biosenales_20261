## **1.1 ¿Qué índices del EEG mejoran la clasificación de estados en BCI?**

En los sistemas de interfaz cerebro-computador o Brain-Computer Interface (BCI), la clasificación de estados mentales o motores a partir de señales de EEG puede mejorarse utilizando índices adicionales a la densidad espectral de potencia (PSD, Power Spectral Density). Aunque la PSD es una de las características más empleadas debido a su capacidad para describir la distribución de energía en diferentes bandas de frecuencia, existen múltiples métricas complementarias que permiten capturar propiedades temporales, espaciales y no lineales de la señal cerebral, aumentando así la precisión y robustez de los clasificadores.

Uno de los índices más utilizados es el de Common Spatial Patterns (CSP), una técnica espacial que maximiza la varianza de una clase mientras minimiza la de otra. CSP resulta especialmente útil en tareas de imaginación motora, ya que resalta diferencias entre regiones corticales activadas durante distintos movimientos imaginados. Este método ha demostrado mejorar significativamente la discriminación entre clases en sistemas BCI basados en EEG, especialmente cuando se combina con características espectrales y clasificadores lineales [1].

Otro índice ampliamente utilizado en BCI corresponde a los fenómenos de desincronización y resincronización relacionados con eventos, conocidos como Event Related Desynchronization y Event Related Synchronization (ERD/ERS). Estas métricas cuantifican los cambios relativos de potencia en bandas específicas respecto a un estado basal o de reposo. En tareas de imaginación motora, es común observar ERD en las bandas mu y beta sobre la corteza motora contralateral al movimiento imaginado, lo que refleja disminución de sincronización neuronal durante la activación cortical. Posteriormente puede presentarse ERS, asociado a procesos de recuperación o inhibición cortical. Debido a su alta interpretabilidad fisiológica y su fuerte relación con la activación motora, ERD/ERS se considera una de las características más importantes en sistemas BCI basados en imaginación motora [2].

Otro grupo importante corresponde a las medidas de conectividad cerebral, como la coherencia, la sincronización de fase (Phase Locking Value, PLV) y la coherencia imaginaria. Estas métricas evalúan la interacción funcional entre distintas regiones cerebrales, permitiendo identificar patrones de comunicación neuronal asociados a diferentes estados cognitivos o motores. La inclusión de información de conectividad ha mostrado mejoras en la clasificación de tareas mentales complejas, ya que el cerebro funciona como una red distribuida y no únicamente mediante activaciones locales [3].

También se utilizan ampliamente índices basados en la entropía y complejidad de la señal, entre ellos la entropía aproximada (Approximate Entropy), la entropía muestral (Sample Entropy) y la entropía espectral. Estas métricas cuantifican el grado de irregularidad o complejidad del EEG, permitiendo distinguir entre diferentes estados mentales incluso cuando las diferencias espectrales son pequeñas. En aplicaciones BCI, las características no lineales han demostrado complementar adecuadamente las medidas clásicas de potencia [4].

Además, las características temporales y tiempo-frecuencia obtenidas mediante transformadas wavelet o transformada corta de Fourier (STFT) permiten analizar cómo evolucionan las frecuencias cerebrales en el tiempo. Las wavelets son particularmente útiles en EEG debido a la naturaleza no estacionaria de las señales cerebrales. Estas representaciones permiten detectar eventos transitorios relacionados con desincronización y resincronización cortical, mejorando la detección de intención motora [5].

Otro índice relevante corresponde a la asimetría hemisférica, calculada a partir de diferencias de potencia o coherencia entre hemisferios cerebrales. Este tipo de métrica es especialmente útil en tareas motoras, ya que durante movimientos reales o imaginados suele producirse desincronización contralateral en las bandas mu y beta. La asimetría puede aumentar la sensibilidad del sistema para identificar lateralidad del movimiento [6].

Finalmente, en investigaciones recientes se han incorporado técnicas basadas en aprendizaje profundo que extraen automáticamente características espaciales y temporales desde el EEG crudo mediante redes convolucionales y recurrentes. Aunque estas técnicas no representan un “índice” tradicional, sí constituyen una alternativa moderna para mejorar la clasificación en BCI, al combinar simultáneamente múltiples representaciones de la señal [7].

En conclusión, aunque la densidad espectral de potencia continúa siendo una herramienta fundamental en el análisis de EEG para BCI, la incorporación de índices espaciales, de conectividad, complejidad, ERD/ERS y tiempo-frecuencia permite capturar información adicional del comportamiento cerebral. La combinación de estas características suele producir sistemas de clasificación más precisos y robustos, especialmente en tareas de imaginación motora y reconocimiento de estados cognitivos.

## **1.2 Tres índices que se puedan implementar en Python**

Se seleccionaron tres índices adicionales que pueden implementarse fácilmente en Python utilizando librerías ampliamente disponibles como NumPy, SciPy, MNE y AntroPy. Los índices escogidos permiten analizar diferentes propiedades del EEG: conectividad funcional, complejidad de la señal y discriminación espacial de la actividad cerebral. Estos índices son muy utilizados en aplicaciones reales de interfaces cerebro-computador debido a su capacidad para mejorar la clasificación de estados motores y cognitivos.

### **a) Coherencia (Coherence)**

La coherencia es una medida de conectividad funcional que evalúa el grado de correlación entre dos señales EEG en una frecuencia específica. Matemáticamente, se calcula a partir de las densidades espectrales cruzadas y las densidades espectrales individuales de las señales. Sus valores oscilan entre 0 y 1, donde valores cercanos a 1 indican alta sincronización entre regiones cerebrales [7].

En sistemas BCI, la coherencia permite estudiar cómo interactúan diferentes áreas corticales durante tareas motoras o de imaginación motora. Por ejemplo, puede analizarse la coherencia entre electrodos C3 y C4 para evaluar cambios en la comunicación interhemisférica durante movimientos imaginados de manos. Esta medida complementa la PSD, ya que no solo analiza energía local sino también relaciones funcionales entre canales [2].

La implementación en Python es sencilla utilizando la función scipy.signal.coherence(), que calcula automáticamente la coherencia entre dos señales EEG.

Su uso en el proyecto permitiría identificar variaciones de sincronización cortical entre condiciones de reposo e imaginación motora, mejorando la separación entre clases.

### **b) Entropía muestral (Sample Entropy)**

La entropía muestral es una medida no lineal que cuantifica la complejidad e irregularidad de una señal temporal. Valores altos indican señales más impredecibles y complejas, mientras que valores bajos corresponden a señales más regulares y repetitivas [8].

En EEG, esta métrica resulta útil porque la actividad cerebral cambia su complejidad dependiendo del estado cognitivo o motor. Durante tareas de concentración o imaginación motora suelen presentarse variaciones en la organización temporal de las señales, lo cual puede ser capturado mediante entropía. Esta característica aporta información complementaria a la PSD, especialmente cuando las diferencias espectrales entre clases son pequeñas [4].

La implementación en Python es muy práctica mediante la librería antropy, utilizando la función sample_entropy().

En el contexto del proyecto, la entropía muestral podría calcularse para cada canal EEG y compararse entre diferentes tareas motoras o períodos de reposo, evaluando cambios en la complejidad cortical asociados a la activación cerebral.

### **c) Event Related Desynchronization/Synchronization (ERD/ERS)**

Event Related Desynchronization y Event Related Synchronization (ERD/ERS) son índices ampliamente utilizados en sistemas BCI para analizar cambios en la actividad cerebral asociados a tareas motoras reales o imaginadas. Estas métricas cuantifican la variación relativa de potencia en bandas específicas del EEG respecto a un estado basal o de reposo, permitiendo identificar procesos de activación e inhibición cortical relacionados con el movimiento [9].

En tareas de imaginación motora, es común observar una disminución de potencia en las bandas mu (8–13 Hz) y beta (13–30 Hz), fenómeno conocido como ERD. Esta desincronización refleja activación neuronal en áreas sensorimotoras, especialmente en el hemisferio contralateral al movimiento imaginado. Posteriormente puede presentarse un aumento de potencia denominado ERS, asociado a procesos de recuperación o reorganización cortical luego de la actividad motora [2].

ERD/ERS es particularmente relevante en aplicaciones BCI debido a su alta interpretabilidad fisiológica y a su relación directa con los mecanismos neuronales involucrados en el control motor. Además, estas métricas permiten analizar diferencias entre imaginación motora derecha e izquierda, ya que la desincronización suele observarse con mayor intensidad en regiones corticales contralaterales, como C3 y C4.

Aunque su cálculo requiere una etapa previa de estimación espectral, generalmente mediante PSD o representaciones tiempo-frecuencia, su implementación en Python resulta relativamente sencilla utilizando librerías como MNE y SciPy.

En este proyecto, ERD/ERS puede utilizarse para diferenciar condiciones de reposo, imaginación de movimiento de mano derecha e imaginación de movimiento de mano izquierda, permitiendo evaluar cambios en los ritmos mu y beta asociados a la activación cortical motora.

La coherencia, la entropía muestral y ERD/ERS representan índices complementarios que aportan información funcional, no lineal y fisiológica del EEG. La combinación de estas características con la densidad espectral de potencia puede mejorar considerablemente el rendimiento de clasificación en aplicaciones BCI.

## **2.1 Metodología de procesamiento y análisis para T0, T1 y T2 en imaginacion**

Para evidenciar las diferencias en los ritmos cerebrales asociados a las condiciones de reposo, imaginación de movimiento de mano derecha e imaginación de movimiento de mano izquierda, se propone una metodología de análisis basada en el procesamiento espectral, temporal y funcional de señales EEG. Esta metodología integra tanto las herramientas desarrolladas en el Proyecto 1, centradas en densidad espectral de potencia, como los índices adicionales seleccionados previamente (coherencia, entropía muestral y sincronización/desincronización relacionada con eventos ERS/ERD), con el fin de obtener una caracterización más completa de la actividad cortical relacionada con imaginación motora.

La metodología propuesta inicia con la adquisición y organización de las señales EEG correspondientes a las tres condiciones experimentales: reposo (T0), imaginación motora de mano izquierda (T1) e imaginación motora de mano derecha (T2). Para cada condición, las señales deben almacenarse separadamente y etiquetarse adecuadamente para facilitar el análisis posterior y el entrenamiento de posibles clasificadores. Debido a que la imaginación motora activa principalmente la corteza sensorimotora, es recomendable priorizar canales ubicados sobre regiones centrales como C3, Cz y C4, ya que estas áreas presentan cambios significativos en las bandas mu y beta durante tareas motoras [5].

Posteriormente, se realiza una etapa de preprocesamiento de las señales EEG. En esta fase se aplican filtros digitales para reducir ruido e interferencias presentes en la señal. Se recomienda utilizar:

- Un filtro pasa banda entre 7 y 40 Hz para conservar las principales bandas fisiológicas del EEG.
- Un filtro notch de 50 o 60 Hz para eliminar interferencia de la red eléctrica.
- Corrección o eliminación de artefactos producidos por parpadeo, movimientos o actividad muscular, utilizando rechazo manual de segmentos o métodos automáticos disponibles en librerías como MNE [10].

Una vez filtradas las señales, se realiza la segmentación temporal en ventanas o epochs correspondientes a cada ensayo experimental. Esto permite analizar de forma independiente cada intento de imaginación motora y calcular características estadísticas robustas para cada condición. Las épocas pueden extraerse desde un segundo antes del evento hasta varios segundos después del inicio de la tarea, permitiendo analizar tanto la actividad basal como la respuesta cortical asociada al movimiento imaginado.

La primera etapa de análisis corresponde al cálculo de la densidad espectral de potencia (PSD) mediante el método de Welch. Este análisis permite identificar cambios en la energía de las bandas mu (8–13 Hz) y beta (13–30 Hz), las cuales están estrechamente relacionadas con tareas motoras e imaginación motora [7]. Durante imaginación de movimiento suele presentarse desincronización relacionada con eventos (Event Related Desynchronization, ERD), observable como una disminución de potencia en dichas bandas sobre el hemisferio contralateral al movimiento imaginado. Por ejemplo:

- Imaginación de mano derecha → disminución de potencia en C3.
- Imaginación de mano izquierda → disminución de potencia en C4.

Asimismo, posterior a ciertos eventos motores puede observarse sincronización relacionada con eventos (Event Related Synchronization, ERS), caracterizada por un aumento transitorio de potencia en determinadas bandas de frecuencia luego de la activación cortical. Tanto ERD como ERS representan fenómenos fundamentales en sistemas BCI, ya que permiten identificar variaciones funcionales de la corteza motora asociadas a intención de movimiento.

El análisis espectral puede realizarse comparando:

- Potencia relativa en bandas mu y beta.

Adicionalmente, se propone utilizar la entropía muestral (Sample Entropy) como índice de complejidad de la señal EEG. Este análisis permitirá evaluar cambios en la irregularidad temporal de la actividad cerebral durante las distintas tareas. Generalmente, estados cognitivos activos presentan variaciones en complejidad respecto al reposo [8]. La entropía puede calcularse para cada canal y compararse estadísticamente entre condiciones.

Como complemento temporal y fisiológico, se plantea analizar directamente los fenómenos de ERD y ERS en las bandas mu y beta. Para ello, la potencia espectral durante imaginación motora puede compararse respecto a un periodo basal asociado al reposo. La evaluación de ERD/ERS constituye una de las herramientas más utilizadas en estudios de imaginación motora debido a su capacidad para representar dinámicamente la activación y recuperación cortical [7].

## **3.1.1 Función: cargar_datos_eeg()**

Esta función se encarga de cargar las señales EEG de todos los sujetos del dataset y prepararlas para los análisis posteriores. Durante el proceso selecciona los canales de interés, ajusta la frecuencia de muestreo cuando es necesario, filtra las señales, genera las épocas correspondientes a cada evento y organiza toda la información para que pueda utilizarse posteriormente en las etapas de extracción de características y clasificación.

**1. IDENTIFICACIÓN DE SUJETOS**

En este bloque se buscan todas las carpetas de sujetos que existen dentro de la ruta principal del dataset. Una vez encontradas, se organizan en orden para procesarlas una por una de manera consistente.

**2. ESTRUCTURAS DE ALMACENAMIENTO**

Aquí se crean las variables donde se irá guardando toda la información obtenida durante el procesamiento. Se preparan estructuras para almacenar las épocas generadas, las señales originales y filtradas, así como un registro con la cantidad de muestras obtenidas para cada sujeto.

**3. RECORRIDO DE SUJETOS**

Una vez identificados los sujetos, se inicia un recorrido sobre cada uno de ellos. Para cada sujeto se crean espacios independientes donde se almacenarán los eventos de reposo, imaginación de mano izquierda e imaginación de mano derecha.

**4. RECORRIDO DE ARCHIVOS EDF**

Dentro de la carpeta de cada sujeto se revisan todos los archivos disponibles. En este paso se conservan únicamente los archivos EDF que pertenecen a los runs definidos para el estudio. Además, se obtiene el número de run asociado a cada archivo para saber posteriormente qué tipo de tarea contiene.

**5. CARGA DEL ARCHIVO EEG**

En este bloque se carga la señal EEG contenida en el archivo. También se conserva una copia de la señal original para poder compararla más adelante con la versión procesada si es necesario.

**6. LIMPIEZA DE NOMBRES DE CANALES**

Algunos archivos pueden contener pequeñas diferencias en la forma en que están escritos los nombres de los canales. Por eso, aquí se realiza una limpieza sencilla para dejar todos los nombres con el mismo formato.

**7. SELECCIÓN DE ELECTRODOS MOTORES**

De todos los canales disponibles se conservan únicamente C3, Cz y C4. Estos electrodos son los más utilizados en tareas de imaginación motora porque se encuentran sobre zonas relacionadas con el movimiento de las extremidades.

**8. AJUSTE DE FRECUENCIA DE MUESTREO**

En este paso se verifica la frecuencia de muestreo de la señal. Si algún registro tiene una frecuencia diferente a la utilizada en el estudio, la señal se ajusta para que todos los sujetos queden trabajando bajo las mismas condiciones.

**9. FILTRADO**

Aquí se aplica un filtro para eliminar componentes de frecuencia que no son relevantes para el análisis. De esta manera se conserva principalmente la información cerebral asociada a las tareas de imaginación motora y se reduce parte del ruido presente en la señal.

**10. CONVERSIÓN DE EVENTOS**

Las anotaciones incluidas dentro del registro EEG se transforman en eventos que posteriormente podrán utilizarse para identificar los distintos momentos del experimento, como reposo o imaginación de movimiento.

**11. CREACIÓN DE ÉPOCAS**

Una vez identificados los eventos, la señal continua se divide en pequeños segmentos temporales alrededor de cada uno de ellos. Cada uno de estos segmentos constituye una época que podrá analizarse de forma independiente.

**12. SELECCIÓN SOLO DE RUNS IMAGINARIOS**

No todos los runs del dataset corresponden a tareas de imaginación motora. Por esta razón, en este bloque se conservan únicamente aquellos que contienen las tareas de interés para el proyecto y se descartan los demás.

**13. ALMACENAMIENTO DE ÉPOCAS**

Las épocas obtenidas para cada evento se guardan dentro de la estructura correspondiente al sujeto. De esta forma toda la información queda organizada según la clase a la que pertenece.

**14. ALMACENAMIENTO DE RAW**

Además de las épocas, también se almacenan las señales originales y las señales filtradas de cada archivo procesado. Esto permite acceder a ellas posteriormente para realizar visualizaciones o verificaciones adicionales.

**15. CONTEO DE ÉPOCAS**

En este bloque se cuenta cuántas épocas fueron obtenidas para cada tipo de evento. Esta información sirve para verificar que los datos se hayan cargado correctamente y conocer la cantidad de muestras disponibles.

**16. CREACIÓN DEL DATAFRAME**

La información del conteo se organiza en un DataFrame para tener una vista resumida de la cantidad de muestras disponibles por sujeto y por clase.

**17. MENSAJE FINAL**

Una vez terminado el procesamiento, se muestra un mensaje indicando cuántos sujetos fueron procesados correctamente.

**18. RETORNO FINAL**

Finalmente, la función devuelve las épocas organizadas por sujeto y evento, el DataFrame con el resumen de muestras y las señales EEG almacenadas durante el procesamiento.


```python
def cargar_datos_eeg(ruta_base, runs_validos):

    """============================================================
    FUNCIÓN:
    ------------------------------------------------------------
    Carga y preprocesa señales EEG del dataset OpenNeuro
    ds004362 enfocándose únicamente en:

    - Reposo (T0)
    - Imaginación mano izquierda (T1)
    - Imaginación mano derecha (T2)

    ============================================================
    RETORNA:
    ------------------------------------------------------------
    datos_epochs : épocas organizadas por sujeto y evento
    df_conteo    : resumen de cantidad de épocas
    datos_raw    : señales originales y filtradas
    ============================================================"""

    # 1. IDENTIFICACIÓN DE SUJETOS

    sujetos = [s for s in os.listdir(ruta_base)
        if os.path.isdir(os.path.join(ruta_base, s))]

    sujetos.sort()

    # 2. ESTRUCTURAS DE ALMACENAMIENTO

    datos_epochs = {}

    datos_raw = {}

    conteo = []

    sujetos_resampleados = set()

    # 3. RECORRIDO DE SUJETOS

    for sujeto in sujetos:

        ruta_sujeto = os.path.join(ruta_base, sujeto)

        # SOLO IMAGINACIÓN MOTORA
        datos_epochs[sujeto] = {
            'T0': [],   # Reposo
            'T1': [],   # Mano izquierda
            'T2': []    # Mano derecha
}

        datos_raw[sujeto] = []

        # 4. RECORRIDO DE ARCHIVOS EDF

        for archivo in sorted(os.listdir(ruta_sujeto)):

            # VALIDACIÓN DE ARCHIVOS
            if not archivo.endswith('.edf'):
                continue

            if not any(r in archivo for r in runs_validos):
                continue

            # EXTRACCIÓN DE RUN
            run = int(archivo.split('R')[1][:2])

            ruta_archivo = os.path.join(ruta_sujeto, archivo)

            # 5. CARGA DEL ARCHIVO EEG

            raw = mne.io.read_raw_edf(ruta_archivo,
                preload=True,
                verbose=False)

            raw_original = raw.copy()

            # 6. LIMPIEZA DE NOMBRES DE CANALES

            raw.rename_channels(lambda x: x.strip().replace('.', ''))

            # 7. SELECCIÓN DE ELECTRODOS MOTORES

            raw.pick(['C3', 'Cz', 'C4'])

            # 8. AJUSTE DE FRECUENCIA DE MUESTREO

            fs = raw.info['sfreq']

            if fs != 160:

                if sujeto not in sujetos_resampleados:

                    print(f"{sujeto} fue remuestreado de {fs} Hz a 160 Hz")

                    sujetos_resampleados.add(sujeto)

                raw.resample(160, npad='auto')

            # 9. FILTRADO

            raw.filter(
                l_freq=7,
                h_freq=40,
                method='fir',
                verbose=False)

            raw_filtrada = raw.copy()

            # 10. CONVERSIÓN DE EVENTOS

            events, event_id = mne.events_from_annotations(raw,
                event_id={'T0': 1, 'T1': 2, 'T2': 3},
                verbose=False)

            # 11. CREACIÓN DE ÉPOCAS

            epochs = mne.Epochs(raw, events,
                event_id=event_id,
                tmin=-1.0,
                tmax=4.0,
                baseline=None,
                preload=True,
                verbose=False)

            # 12. SELECCIÓN SOLO DE RUNS IMAGINARIOS

            # R04, R08 y R12:
            # imaginación motora
            if run not in [4, 8, 12]:
                continue

            # 13. ALMACENAMIENTO DE ÉPOCAS

            datos_epochs[sujeto]['T0'].append(epochs['T0'])

            datos_epochs[sujeto]['T1'].append(epochs['T1'])

            datos_epochs[sujeto]['T2'].append(epochs['T2'])

            # 14. ALMACENAMIENTO DE RAW

            datos_raw[sujeto].append({
                'archivo': archivo,
                'raw_original': raw_original,
                'raw_filtrada': raw_filtrada})

        # 15. CONTEO DE ÉPOCAS

        n_T0 = sum(len(e) for e in datos_epochs[sujeto]['T0'])

        n_T1 = sum(len(e) for e in datos_epochs[sujeto]['T1'])

        n_T2 = sum(len(e) for e in datos_epochs[sujeto]['T2'])

        conteo.append({
            'Sujeto': sujeto,
            'Reposo_T0': n_T0,
            'Imaginacion_Izquierda_T1': n_T1,
            'Imaginacion_Derecha_T2': n_T2})

    # 16. CREACIÓN DEL DATAFRAME

    df_conteo = pd.DataFrame(conteo)

    # 17. MENSAJE FINAL

    print(f"\nSe procesaron correctamente {len(sujetos)} sujetos.")

    # 18. RETORNO FINAL

    return datos_epochs, df_conteo, datos_raw
```

## **3.1.2 Función: extraer_indices_eeg()**

Esta función se encarga de calcular diferentes índices utilizados en el análisis de señales EEG para aplicaciones de Brain-Computer Interface (BCI). A partir de las épocas previamente obtenidas, se extraen medidas relacionadas con la actividad espectral, la conectividad entre canales, la complejidad de la señal y los cambios producidos durante las tareas de imaginación motora. Además, la función permite generar gráficas que facilitan la interpretación de los resultados.

**1. SELECCIÓN DE CANALES**

En este bloque se seleccionan únicamente los canales que serán utilizados durante el análisis. También se extrae la información necesaria de las épocas, como los datos de las señales, la frecuencia de muestreo, los nombres de los canales y el vector temporal asociado a cada muestra.

**2. ESTRUCTURA DE RESULTADOS**

Aquí se crea la estructura donde se almacenarán todos los índices calculados. Los resultados se organizan en diferentes categorías para facilitar posteriormente el acceso a la información obtenida de cada análisis.

**3. PSD Y POTENCIA ESPECTRAL**

En esta sección se calcula la densidad espectral de potencia de las señales EEG. Este análisis permite conocer cómo se distribuye la energía de la señal en las diferentes frecuencias. Posteriormente se obtiene la potencia promedio dentro de las bandas Mu y Beta para cada canal, ya que estas bandas suelen contener información relevante durante tareas de imaginación motora.

Si la opción de graficar está activada, también se genera una representación visual de la densidad espectral de potencia de cada canal, resaltando las bandas Mu y Beta para facilitar su interpretación.

**4. COHERENCIA**

En este bloque se analiza la relación existente entre pares de canales EEG mediante el cálculo de la coherencia. Esta medida permite evaluar qué tan sincronizadas se encuentran dos señales en función de la frecuencia.

Para ello se comparan todas las combinaciones posibles entre los canales seleccionados y posteriormente se calcula la coherencia promedio dentro de las bandas Mu y Beta. Si se habilita la visualización, se genera una gráfica que muestra el comportamiento de la coherencia para cada par de canales.

**5. ENTROPÍA MUESTRAL**

Aquí se calcula la entropía muestral de cada canal. Esta medida proporciona una estimación de la complejidad o irregularidad de la señal EEG, permitiendo identificar diferencias en la dinámica de la actividad cerebral.

Cuando se activa la opción de graficar, los valores obtenidos se representan mediante un gráfico de barras para facilitar la comparación entre los distintos canales analizados.

**6. ERD (EVENT RELATED DESYNCHRONIZATION)**

En esta etapa se calcula el porcentaje de desincronización relacionada con eventos, conocido como ERD. Para ello se compara la actividad cerebral registrada durante un periodo basal con la actividad observada durante la ejecución de la tarea de imaginación motora.

El cálculo se realiza de forma independiente para cada canal y para las bandas Mu y Beta. De esta manera es posible cuantificar cómo cambia la potencia de la señal durante la tarea con respecto a la condición de referencia. Cuando se trata de eventos de imaginación motora, también se genera una gráfica que permite visualizar estos cambios de forma comparativa entre canales.

**7. RETORNO FINAL**

Finalmente, la función devuelve una estructura que contiene todos los índices calculados durante el procesamiento, incluyendo la información relacionada con la densidad espectral de potencia, la coherencia, la entropía muestral y el ERD.


```python
def extraer_indices_eeg(
        epochs,
        canales=['C3', 'Cz', 'C4'],
        bandas={'mu': (8, 13), 'beta': (13, 30)},
        calcular_psd=True,
        calcular_coherencia=True,
        calcular_entropia=True,
        calcular_erd=True,
        graficar=True,
        sujeto=None,
        evento=None):

    """
    ============================================================
    FUNCIÓN:
    ------------------------------------------------------------
    Calcula índices EEG utilizados en BCI:

    1. PSD
    2. Potencia Mu/Beta
    3. Coherencia
    4. Entropía muestral
    5. ERD (Event Related Desynchronization)

    ============================================================
    RETORNA:
    ------------------------------------------------------------
    resultados : dict
    ============================================================
    """

    # 1. SELECCIÓN DE CANALES

    epochs = epochs.copy().pick(canales)

    data = epochs.get_data()

    sfreq = epochs.info['sfreq']

    nombres_canales = epochs.ch_names

    tiempos = epochs.times

    # 2. ESTRUCTURA DE RESULTADOS

    resultados = {
        'PSD': {},
        'Coherencia': {},
        'Entropia': {},
        'ERD': {}
    }

    # NOMBRES DE EVENTOS
    nombres_eventos = {
        'T0': 'Reposo',
        'T1': 'Imaginación Mano Izquierda',
        'T2': 'Imaginación Mano Derecha'
    }

    # FIGURA GENERAL CON SUBPLOTS
    if graficar:

        fig, axs = plt.subplots(
            2,
            2,
            figsize=(15, 10)
        )

        fig.suptitle(
            f"{sujeto} - {nombres_eventos[evento]}",
            fontsize=16,
            fontweight='bold'
        )

        axs = axs.flatten()

        idx_subplot = 0

    # 3. PSD Y POTENCIA ESPECTRAL

    if calcular_psd:

        psd, freqs = psd_array_welch(
            data,
            sfreq=sfreq,
            fmin=1,
            fmax=40,
            n_fft=160,
            n_overlap=80,
            verbose=False
        )

        # PROMEDIO ENTRE ÉPOCAS
        psd_mean = psd.mean(axis=0)

        # RECORRIDO DE CANALES
        for i, canal in enumerate(nombres_canales):

            resultados['PSD'][canal] = {
                'Frecuencias': freqs,
                'PSD': psd_mean[i]
            }

            # POTENCIA EN BANDAS
            for nombre_banda, (fmin, fmax) in bandas.items():

                idx = ((freqs >= fmin) & (freqs <= fmax))

                potencia = psd_mean[i, idx].mean()

                resultados['PSD'][canal][nombre_banda] = potencia

        # GRÁFICA PSD
        if graficar:

            ax = axs[idx_subplot]

            for i, canal in enumerate(nombres_canales):

                ax.plot(
                    freqs,
                    psd_mean[i],
                    label=canal
                )

            # Banda Mu
            ax.axvspan(
                8, 13,
                color='skyblue',
                alpha=0.3,
                label='Mu'
            )

            # Banda Beta
            ax.axvspan(
                13, 30,
                color='salmon',
                alpha=0.3,
                label='Beta'
            )

            ax.set_xlabel("Frecuencia [Hz]")
            ax.set_ylabel("PSD")

            ax.set_title("PSD")

            ax.legend()

            ax.grid()

            idx_subplot += 1

    # 4. COHERENCIA

    if calcular_coherencia and len(canales) >= 2:

        for i in range(len(canales)):

            for j in range(i + 1, len(canales)):

                canal_1 = canales[i]
                canal_2 = canales[j]

                # CONCATENACIÓN DE ÉPOCAS
                señal_1 = data[:, i, :].flatten()
                señal_2 = data[:, j, :].flatten()

                # CÁLCULO DE COHERENCIA
                f_coh, coh = coherence(
                    señal_1,
                    señal_2,
                    fs=sfreq,
                    nperseg=64
                )

                nombre_par = f"{canal_1}-{canal_2}"

                resultados['Coherencia'][nombre_par] = {
                    'Frecuencias': f_coh,
                    'Coherencia': coh
                }

                # COHERENCIA EN BANDAS
                for nombre_banda, (fmin, fmax) in bandas.items():

                    idx = ((f_coh >= fmin) & (f_coh <= fmax))

                    coh_media = coh[idx].mean()

                    resultados['Coherencia'][nombre_par][nombre_banda] = coh_media

        # GRÁFICA COHERENCIA
        if graficar:

            ax = axs[idx_subplot]

            for par in resultados['Coherencia']:

                ax.plot(
                    resultados['Coherencia'][par]['Frecuencias'],
                    resultados['Coherencia'][par]['Coherencia'],
                    label=par
                )

            ax.set_xlabel("Frecuencia [Hz]")
            ax.set_ylabel("Coherencia")

            ax.set_title("Coherencia")

            ax.legend()

            ax.grid()

            idx_subplot += 1

    # 5. ENTROPÍA MUESTRAL

    if calcular_entropia:

        for i, canal in enumerate(nombres_canales):

            señal = data[0, i, :]

            entropia = ant.sample_entropy(señal)

            resultados['Entropia'][canal] = entropia

        # GRÁFICA ENTROPÍA
        if graficar:

            canales_ent = list(resultados['Entropia'].keys())

            valores_ent = list(resultados['Entropia'].values())

            ax = axs[idx_subplot]

            ax.bar(
                canales_ent,
                valores_ent,
                color=['royalblue', 'darkorange', 'seagreen']
            )

            ax.set_ylabel("Sample Entropy")

            ax.set_title("Entropía Muestral")

            ax.grid()

            idx_subplot += 1

    # 6. ERD (EVENT RELATED DESYNCHRONIZATION)

    if calcular_erd:

        # VENTANAS TEMPORALES
        # Basal: -1 a 0 s
        idx_basal = np.where(
            (tiempos >= -1) & (tiempos <= 0)
        )[0]

        # Actividad motora: 0.5 a 3 s
        idx_evento = np.where(
            (tiempos >= 0.5) & (tiempos <= 3)
        )[0]

        # RECORRIDO DE CANALES
        for i, canal in enumerate(nombres_canales):

            resultados['ERD'][canal] = {}

            señal = data[:, i, :]

            # PSD BASAL
            psd_basal, f_basal = psd_array_welch(
                señal[:, idx_basal],
                sfreq=sfreq,
                fmin=8,
                fmax=30,
                n_fft=128,
                verbose=False
            )

            # PSD EVENTO
            psd_evento, f_evento = psd_array_welch(
                señal[:, idx_evento],
                sfreq=sfreq,
                fmin=8,
                fmax=30,
                n_fft=128,
                verbose=False
            )

            # ERD POR BANDA
            for nombre_banda, (fmin, fmax) in bandas.items():

                idx_f_basal = (
                    (f_basal >= fmin) &
                    (f_basal <= fmax)
                )

                idx_f_evento = (
                    (f_evento >= fmin) &
                    (f_evento <= fmax)
                )

                potencia_basal = psd_basal[:, idx_f_basal].mean()

                potencia_evento = psd_evento[:, idx_f_evento].mean()

                # FÓRMULA ERD (%)
                erd = (
                    (potencia_evento - potencia_basal)
                    / potencia_basal
                ) * 100

                resultados['ERD'][canal][nombre_banda] = erd

        # GRÁFICA ERD
        if graficar:

            ax = axs[idx_subplot]

            if evento != 'T0':

                x = np.arange(len(nombres_canales))

                mu_vals = [
                    resultados['ERD'][c]['mu']
                    for c in nombres_canales
                ]

                beta_vals = [
                    resultados['ERD'][c]['beta']
                    for c in nombres_canales
                ]

                ax.bar(
                    x - 0.2,
                    mu_vals,
                    width=0.4,
                    label='Mu',
                    color='skyblue'
                )

                ax.bar(
                    x + 0.2,
                    beta_vals,
                    width=0.4,
                    label='Beta',
                    color='salmon'
                )

                ax.axhline(
                    0,
                    color='black',
                    linestyle='--'
                )

                ax.set_xticks(x)

                ax.set_xticklabels(nombres_canales)

                ax.set_ylabel("ERD (%)")

                ax.set_title("ERD")

                ax.legend()

                ax.grid()

            else:

                ax.axis('off')

            idx_subplot += 1

    # MOSTRAR FIGURA FINAL
    if graficar:

        plt.tight_layout()

        plt.show()

    # 7. RETORNO FINAL

    return resultados
```

## **3.1.3 Función: visualizar()**

Esta función se utiliza para observar de forma gráfica las señales EEG en diferentes etapas del procesamiento. Para una condición experimental determinada, permite comparar la señal original, la señal después del filtrado y una época segmentada correspondiente al evento seleccionado. De esta manera es posible verificar visualmente que el preprocesamiento y la segmentación se realizaron correctamente.

**1. RECORRIDO DE SUJETOS**

En este bloque se obtienen todos los sujetos disponibles y se inicia un recorrido sobre cada uno de ellos. El objetivo es acceder a la información almacenada para cada participante y poder visualizar sus registros EEG.

**2. RECORRIDO DE RUNS**

Para cada sujeto se recorren los diferentes runs o archivos procesados. Además, se muestra en pantalla información básica del sujeto y del archivo que se está analizando para identificar fácilmente de dónde provienen las señales que serán visualizadas.

**3. OBTENCIÓN DE SEÑALES RAW**

Aquí se recuperan las señales EEG originales y las señales que previamente fueron filtradas durante el preprocesamiento. Posteriormente se seleccionan únicamente los canales de interés para que las visualizaciones se enfoquen en los electrodos relevantes para el análisis.

**4. OBTENCIÓN DE ÉPOCAS**

En este bloque se accede a las épocas asociadas al evento seleccionado. Si no existen épocas para esa condición, la ejecución continúa con el siguiente caso disponible. Cuando sí existen datos, se selecciona una época correspondiente al mismo run que se está visualizando.

**5. EXTRACCIÓN DE DATOS**

Una vez obtenidas las señales y la época, se extrae toda la información necesaria para construir las gráficas. Esto incluye los vectores de tiempo y los valores de las señales tanto para el registro original, el registro filtrado y la época segmentada.

**6. CREACIÓN DE FIGURA**

Aquí se crea la figura principal que contendrá todas las visualizaciones. La distribución se organiza de manera vertical para facilitar la comparación entre las diferentes etapas de procesamiento de la señal.

**7. SEÑAL ORIGINAL**

En esta sección se grafica la señal EEG original de los canales seleccionados. Esta visualización permite observar la información tal como fue registrada antes de aplicar cualquier procedimiento de filtrado o procesamiento adicional.

**8. SEÑAL FILTRADA**

En este bloque se representa la señal EEG después del proceso de filtrado. Al compararla con la señal original es posible apreciar cómo se reducen componentes no deseadas y se conserva principalmente la actividad relevante para el análisis.

**9. ÉPOCA EEG**

Aquí se muestra una de las épocas obtenidas a partir del evento seleccionado. Dependiendo de la condición experimental elegida, también se asigna un nombre descriptivo que facilita la interpretación de la gráfica. Esta visualización permite observar directamente el segmento de señal que será utilizado en las etapas posteriores de análisis.

**10. AJUSTE FINAL**

Finalmente, se ajusta la distribución de las gráficas para mejorar su presentación y se muestran en pantalla. Después de generar la visualización, la función termina su ejecución para mostrar únicamente el primer ejemplo disponible, evitando producir una gran cantidad de figuras de manera consecutiva.


```python
def visualizar(
        datos_raw,
        datos_epochs,
        evento='T1',
        canales=['C3', 'Cz', 'C4']):

    """============================================================
    FUNCIÓN:
    ------------------------------------------------------------
    Visualiza:

    1. Señal EEG original
    2. Señal EEG filtrada
    3. Época segmentada

    Para una condición experimental específica:

    - T0 : Reposo
    - T1 : Imaginación mano izquierda
    - T2 : Imaginación mano derecha

    ============================================================
    PARÁMETROS:
    ------------------------------------------------------------
    datos_raw : dict
        Señales originales y filtradas

    datos_epochs : dict
        Épocas EEG segmentadas

    evento : str
        Condición experimental a visualizar

    canales : list
        Canales EEG de interés

    ============================================================
    RETORNA:
    ------------------------------------------------------------
    No retorna valores.
    Solo genera visualizaciones.
    ============================================================"""

    # 1. RECORRIDO DE SUJETOS

    sujetos = sorted(datos_raw.keys())

    for sujeto in sujetos:

        # 2. RECORRIDO DE RUNS

        runs = datos_raw[sujeto]

        for i, run_data in enumerate(runs):

            archivo = run_data['archivo']

            print(f"Sujeto: {sujeto}")

            print(f"Archivo: {archivo}")

            # 3. OBTENCIÓN DE SEÑALES RAW

            raw_original = run_data['raw_original']

            raw_filtrada = run_data['raw_filtrada']

            # SELECCIÓN DE CANALES
            raw_original = (raw_original.copy().rename_channels(lambda x: x.strip().replace('.', '')).pick(canales))

            raw_filtrada = (raw_filtrada.copy().rename_channels(lambda x: x.strip().replace('.', '')).pick(canales))

            # 4. OBTENCIÓN DE ÉPOCAS

            epochs_list = datos_epochs[sujeto][evento]

            if len(epochs_list) == 0:

                continue

            # SE TOMA LA ÉPOCA DEL MISMO RUN
            epochs = epochs_list[i]

            epoca = epochs[0]

            # 5. EXTRACCIÓN DE DATOS

            t_raw = raw_original.times

            t_filt = raw_filtrada.times

            t_epoca = epoca.times

            señal_raw = raw_original.get_data()

            señal_filt = raw_filtrada.get_data()

            señal_epoca = epoca.get_data()[0]

            # 6. CREACIÓN DE FIGURA

            plt.figure(figsize=(13, 9))

            # 7. SEÑAL ORIGINAL

            plt.subplot(3, 1, 1)

            for idx, canal in enumerate(canales):

                plt.plot(t_raw, señal_raw[idx], label=canal)

            plt.title(f"Señal EEG Original - {evento}")
            plt.xlabel("Tiempo [s]")
            plt.ylabel("Amplitud")
            plt.legend()
            plt.grid()

            # 8. SEÑAL FILTRADA

            plt.subplot(3, 1, 2)

            for idx, canal in enumerate(canales):

                plt.plot(t_filt, señal_filt[idx], label=canal)

            plt.title("Señal EEG Filtrada (7–40 Hz)")
            plt.xlabel("Tiempo [s]")
            plt.ylabel("Amplitud")
            plt.legend()
            plt.grid()

            # 9. ÉPOCA EEG

            plt.subplot(3, 1, 3)

            for idx, canal in enumerate(canales):

                plt.plot(t_epoca, señal_epoca[idx], label=canal)

            # NOMBRE DE LA CONDICIÓN
            if evento == 'T0':

                condicion = "Reposo"

            elif evento == 'T1':

                condicion = "Imaginación Mano Izquierda"

            elif evento == 'T2':

                condicion = "Imaginación Mano Derecha"

            else:

                condicion = evento

            plt.title(f"Época EEG - {condicion}")
            plt.xlabel("Tiempo [s]")
            plt.ylabel("Amplitud")
            plt.legend()
            plt.grid()

            # 10. AJUSTE FINAL

            plt.tight_layout()
            plt.show()

            # SOLO SE GRAFICA EL PRIMER EJEMPLO

            return
```

## **3.1.4 Función: graficar_bandas()**

Esta función se utiliza para visualizar la potencia promedio de las bandas Mu y Beta obtenidas a partir de las señales EEG. Para cada condición experimental se generan gráficas de barras que permiten comparar la actividad registrada en los distintos canales analizados. Estas visualizaciones facilitan la identificación de diferencias entre eventos y ayudan a interpretar el comportamiento de las bandas más relevantes en tareas de imaginación motora.

**1. NOMBRES DE EVENTOS**

En este bloque se define la correspondencia entre los códigos de los eventos y sus nombres descriptivos. Esto permite que las gráficas muestren títulos más claros y fáciles de interpretar durante la visualización de los resultados.

**2. RECORRIDO DE EVENTOS**

Aquí se realiza un recorrido por las tres condiciones experimentales disponibles: reposo, imaginación de mano izquierda e imaginación de mano derecha. El objetivo es generar una visualización independiente para cada una de ellas.

**3. FILTRAR EVENTO**

En este paso se seleccionan únicamente los registros correspondientes al evento que se está analizando en ese momento. De esta forma, los cálculos y las gráficas se construyen utilizando solamente la información asociada a esa condición experimental.

**4. INICIALIZACIÓN**

Se crean las listas donde se almacenarán temporalmente los valores promedio de potencia de las bandas Mu y Beta para cada canal. Estas listas servirán posteriormente para construir las gráficas de barras.

**5. RECORRIDO DE CANALES**

En este bloque se recorren los canales seleccionados y se calcula la potencia promedio de las bandas Mu y Beta para cada uno de ellos. Los valores obtenidos se almacenan en las listas creadas anteriormente para facilitar su representación gráfica.

**6. POSICIONES**

Aquí se definen las posiciones que ocuparán las barras dentro de las gráficas. Esto permite organizar correctamente los canales sobre el eje horizontal y mantener una distribución uniforme de los datos.

**7. CREACIÓN DE FIGURA**

En este bloque se crea la figura que contendrá las visualizaciones. La distribución se organiza en dos paneles para mostrar simultáneamente la información correspondiente a la banda Mu y a la banda Beta.

**8. BANDA MU**

En esta sección se construye la gráfica de barras para la banda Mu. Cada barra representa la potencia promedio calculada para uno de los canales analizados, permitiendo comparar fácilmente la actividad observada entre ellos.

**9. BANDA BETA**

Aquí se genera la gráfica correspondiente a la banda Beta. Al igual que en el caso anterior, las barras muestran la potencia promedio de cada canal, facilitando la comparación de la actividad cerebral dentro de este rango de frecuencias.

**10. TÍTULO GENERAL**

Finalmente, se agrega un título general que identifica la condición experimental representada en las gráficas. Después se ajusta la distribución de los elementos visuales y se muestran los resultados en pantalla.


```python
def graficar_bandas(df_indices, canales=['C3', 'Cz', 'C4']):

    """============================================================
    FUNCIÓN:
    ------------------------------------------------------------
    Grafica la potencia promedio en bandas Mu y Beta
    usando gráficas de barras agrupadas.

    Condiciones:
    - T0 → Reposo
    - T1 → Imag. Mano Izquierda
    - T2 → Imag. Mano Derecha

    ============================================================"""

    # 1. NOMBRES DE EVENTOS

    nombres_eventos = {
        'T0': 'Reposo',
        'T1': 'Imag. Izquierda',
        'T2': 'Imag. Derecha'
    }

    # 2. RECORRIDO DE EVENTOS

    for evento in ['T0', 'T1', 'T2']:

        # 3. FILTRAR EVENTO

        df_evento = df_indices[df_indices['Evento'] == evento]

        # 4. INICIALIZACIÓN

        mu_vals = []
        beta_vals = []

        # 5. RECORRIDO DE CANALES

        for canal in canales:

            # Banda Mu
            mu = df_evento[f'Mu_{canal}'].mean() * 1e12
            mu_vals.append(mu)

            # Banda Beta
            beta = df_evento[f'Beta_{canal}'].mean() * 1e12
            beta_vals.append(beta)

        # 6. POSICIONES

        x = np.arange(len(canales))

        # 7. CREACIÓN DE FIGURA

        fig, axes = plt.subplots(1, 2, figsize=(12, 5))

        # 8. BANDA MU

        axes[0].bar(x, mu_vals, width=0.6, color='skyblue')
        axes[0].set_xticks(x)
        axes[0].set_xticklabels(canales)
        axes[0].set_title("Banda Mu (8–13 Hz)")
        axes[0].set_ylabel("Potencia media (µV²/Hz)")
        axes[0].grid(alpha=0.3)

        # 9. BANDA BETA

        axes[1].bar(x, beta_vals, width=0.6, color='salmon')
        axes[1].set_xticks(x)
        axes[1].set_xticklabels(canales)
        axes[1].set_title("Banda Beta (13–30 Hz)")
        axes[1].set_ylabel("Potencia media (µV²/Hz)")
        axes[1].grid(alpha=0.3)

        # 10. TÍTULO GENERAL

        fig.suptitle(
            f"Potencia Promedio de {nombres_eventos[evento]}"
        )

        plt.tight_layout()
        plt.show()
```

## **3.2 Rutina y DataFrame| Función: Rutina()**

Esta función ejecuta todo el flujo de procesamiento del proyecto. A partir de los datos EEG almacenados en el dataset, realiza la carga y el preprocesamiento de las señales, genera visualizaciones, calcula los diferentes índices EEG utilizados en el análisis y organiza los resultados en un DataFrame que posteriormente puede emplearse para análisis estadísticos o modelos de clasificación.

**1. CARGA Y PREPROCESAMIENTO**

En este bloque se inicia el procesamiento de las señales EEG. Se cargan los datos de todos los sujetos, se realiza el preprocesamiento definido previamente y se obtiene un resumen con la cantidad de épocas disponibles para cada condición experimental. Además, se almacenan las señales y las épocas que serán utilizadas durante las siguientes etapas del análisis.

**4. RECORRIDO DE SUJETOS**

En este bloque se recorren todos los sujetos disponibles en el conjunto de datos. Para cada sujeto se procesan los diferentes eventos experimentales y se calculan los índices EEG correspondientes. Durante este proceso también se controla qué sujetos generarán gráficas y cuáles únicamente realizarán el cálculo de características.

**5. CREACIÓN DE FILA**

Una vez calculados los índices para un sujeto y un evento determinado, se construye una fila con toda la información obtenida. En ella se almacenan las medidas relacionadas con las bandas Mu y Beta, la entropía, los valores de ERD y las medidas de coherencia calculadas entre los distintos pares de canales.

**6. DATAFRAME FINAL**

Después de procesar todos los sujetos y todas las condiciones experimentales, la información almacenada se organiza en un DataFrame. Esta estructura permite trabajar de forma más cómoda con los datos y facilita los análisis posteriores.

**7. GUARDAR CSV**

En este bloque el DataFrame generado se exporta a un archivo CSV. De esta manera los resultados quedan almacenados y pueden utilizarse posteriormente sin necesidad de repetir todo el procesamiento.

**8. GRAFICAR POTENCIA PROMEDIO**

Aquí se generan las gráficas de potencia promedio para las bandas Mu y Beta utilizando la información contenida en el DataFrame. Estas visualizaciones permiten comparar el comportamiento de los diferentes canales en cada condición experimental.

**9. FINALIZACIÓN**

En esta etapa se muestra un mensaje indicando que todas las tareas del flujo de procesamiento se completaron correctamente y que los resultados ya se encuentran disponibles para su análisis.

**10. RETORNO FINAL**

Finalmente, la función devuelve tanto el DataFrame con las características extraídas como la estructura completa que contiene todos los índices EEG calculados durante el procesamiento.


```python
def Rutina(ruta_base, runs_validos, canales=['C3', 'Cz', 'C4']):

    """
    ============================================================
    FUNCIÓN:
    ------------------------------------------------------------
    Ejecuta el flujo completo de procesamiento EEG para:

    - Reposo (T0)
    - Imaginación mano izquierda (T1)
    - Imaginación mano derecha (T2)

    Además calcula:

    - PSD
    - Potencia Mu/Beta
    - Coherencia
    - Entropía
    - ERD

    ============================================================
    RETORNA:
    ------------------------------------------------------------
    df_indices
    resultados_indices
    ============================================================
    """

    # 1. CARGA Y PREPROCESAMIENTO

    print(
        "PROCESO 1:\n"
        "Carga y procesamiento EEG..."
    )

    datos_epochs, df_conteo, datos_raw = cargar_datos_eeg(
        ruta_base,
        runs_validos
    )

    print(df_conteo)

    # 2. VISUALIZACIÓN EEG

    print(
        "\nPROCESO 2:\n"
        "Visualización EEG..."
    )

    # T0
    print("\nVisualización T0 -> Reposo\n")

    visualizar(
        datos_raw,
        datos_epochs,
        evento='T0',
        canales=canales
    )

    # T1
    print("\nVisualización T1 -> Mano izquierda\n")

    visualizar(
        datos_raw,
        datos_epochs,
        evento='T1',
        canales=canales
    )

    # T2
    print("\nVisualización T2 -> Mano derecha\n")

    visualizar(
        datos_raw,
        datos_epochs,
        evento='T2',
        canales=canales
    )

    # 3. EXTRACCIÓN DE ÍNDICES EEG

    print(
        "\nPROCESO 3:\n"
        "Extracción de índices EEG..."
    )

    resultados_indices = {}

    filas = []

    sujetos = sorted(datos_epochs.keys())

    # NOMBRES DE EVENTOS
    nombres_eventos = {
        'T0': 'Reposo',
        'T1': 'Imaginación Mano Izquierda',
        'T2': 'Imaginación Mano Derecha'
    }

    # 4. RECORRIDO DE SUJETOS

    for idx_sujeto, sujeto in enumerate(sujetos):

        resultados_indices[sujeto] = {}

        # RECORRIDO DE EVENTOS
        for evento in ['T0', 'T1', 'T2']:

            epochs_list = datos_epochs[sujeto][evento]

            # VALIDACIÓN

            if len(epochs_list) == 0:

                continue

            # CONCATENACIÓN DE ÉPOCAS
            epochs = mne.concatenate_epochs(epochs_list)

            # SOLO GRAFICAR PRIMER SUJETO
            if idx_sujeto == 0:

                graficar = True

            else:

                graficar = False

            # CÁLCULO DE ÍNDICES EEG
            resultados = extraer_indices_eeg(
                epochs,
                canales=canales,
                calcular_psd=True,
                calcular_coherencia=True,
                calcular_entropia=True,
                calcular_erd=True,
                graficar=graficar,
                sujeto=sujeto,
                evento=evento
            )

            resultados_indices[sujeto][evento] = resultados

            # 5. CREACIÓN DE FILA

            fila = {
                'Sujeto': sujeto,
                'Evento': evento
            }

            # PSD + ENTROPÍA + ERD
            for canal in canales:

                # PSD
                fila[f'Mu_{canal}'] = (
                    resultados['PSD'][canal]['mu']
                )

                fila[f'Beta_{canal}'] = (
                    resultados['PSD'][canal]['beta']
                )

                # ENTROPÍA
                fila[f'Entropia_{canal}'] = (
                    resultados['Entropia'][canal]
                )

                # ERD
                if evento != 'T0':

                    fila[f'ERD_Mu_{canal}'] = (
                        resultados['ERD'][canal]['mu']
                    )

                    fila[f'ERD_Beta_{canal}'] = (
                        resultados['ERD'][canal]['beta']
                    )

                else:

                    fila[f'ERD_Mu_{canal}'] = np.nan

                    fila[f'ERD_Beta_{canal}'] = np.nan

            # COHERENCIA
            for par in resultados['Coherencia']:

                fila[f'Coherencia_Mu_{par}'] = (
                    resultados['Coherencia'][par]['mu']
                )

                fila[f'Coherencia_Beta_{par}'] = (
                    resultados['Coherencia'][par]['beta']
                )

            # GUARDAR FILA
            filas.append(fila)
            
    print(
    f"\nSe procesaron correctamente {len(sujetos)} sujetos."
)

    # 6. DATAFRAME FINAL

    df_indices = pd.DataFrame(filas)

    # 7. GUARDAR CSV

    df_indices.to_csv(
        "Resultados_P2.csv",
        index=False
    )

    print(
        "\nPROCESO 4:\n"
        "DataFrame final generado"
    )

    print(df_indices.head())

    print("\nCSV guardado correctamente.")

    # 8. GRAFICAR POTENCIA PROMEDIO

    print(
        "\nPROCESO 5:\n"
        "Gráficas de potencia promedio..."
    )

    graficar_bandas(
        df_indices,
        canales=canales
    )

    # 9. FINALIZACIÓN

    print("\nPROCESAMIENTO FINALIZADO")

    # 10. RETORNO FINAL


    return (
        df_indices,
        resultados_indices
    )
```

```python
# BLOQUE 1. CONFIGURACIÓN GENERAL

df = pd.read_csv("Resultados_P2.csv")

sns.set(style='whitegrid', palette='Set2')

# BLOQUE 2. MAPEO DE CONDICIONES

df['Condicion'] = df['Evento'].map({
    'T0': 'Reposo',
    'T1': 'Mano Izquierda',
    'T2': 'Mano Derecha'
})

orden = [
    'Reposo',
    'Mano Izquierda',
    'Mano Derecha'
]

# BLOQUE 3. DEFINICIÓN DE VARIABLES

potencia_abs = [
    'Mu_C3', 'Beta_C3',
    'Mu_Cz', 'Beta_Cz',
    'Mu_C4', 'Beta_C4'
]

erds = [
    'ERD_Mu_C3',
    'ERD_Beta_C3',
    'ERD_Mu_Cz',
    'ERD_Beta_Cz',
    'ERD_Mu_C4',
    'ERD_Beta_C4'
]

entropias = [
    'Entropia_C3',
    'Entropia_Cz',
    'Entropia_C4'
]

coherencias = [
    'Coherencia_Mu_C3-Cz',
    'Coherencia_Beta_C3-Cz',
    'Coherencia_Mu_C3-C4',
    'Coherencia_Beta_C3-C4',
    'Coherencia_Mu_Cz-C4',
    'Coherencia_Beta_Cz-C4'
]

# BLOQUE 4. TODAS LAS VARIABLES NUMÉRICAS

variables_numericas = (
    potencia_abs +
    erds +
    entropias +
    coherencias
)

# BLOQUE 5. LIMPIEZA DE DATOS

for col in variables_numericas:

    if col in df.columns:

        df[col] = df[col].apply(
            lambda x: x[0]
            if isinstance(x, (list, np.ndarray))
            else x
        )

        df[col] = pd.to_numeric(
            df[col],
            errors='coerce'
        )

# BLOQUE 6. VERIFICACIÓN DE TIPOS

print("\nTipos de datos:\n")
print(df[variables_numericas].dtypes)

# BLOQUE 7. ESTADÍSTICA DESCRIPTIVA

estadisticas = (
    df.groupby('Condicion')[variables_numericas]
    .agg(['mean', 'std', 'median', 'min', 'max'])
)

print("\nESTADÍSTICAS DESCRIPTIVAS:\n")
print(estadisticas)

# BLOQUE 8. FUNCIÓN GENERAL DE BOXPLOTS

def graficar_variables(lista_variables, titulo_general):

    lista_variables = [
        var for var in lista_variables
        if var in df.columns
    ]

    n = len(lista_variables)

    filas = int(np.ceil(n / 2))

    fig, axes = plt.subplots(
        filas,
        2,
        figsize=(14, 4 * filas)
    )

    if filas == 1:
        axes = np.array(axes).reshape(-1)
    else:
        axes = axes.flatten()

    for i, var in enumerate(lista_variables):

        sns.boxplot(
            data=df,
            x='Condicion',
            y=var,
            order=orden,
            ax=axes[i]
        )

        sns.stripplot(
            data=df,
            x='Condicion',
            y=var,
            order=orden,
            color='black',
            alpha=0.5,
            jitter=True,
            ax=axes[i]
        )

        axes[i].set_title(var)
        axes[i].set_xlabel('')
        axes[i].set_ylabel('Valor')

    for j in range(i + 1, len(axes)):
        fig.delaxes(axes[j])

    plt.suptitle(
        titulo_general,
        fontsize=18
    )

    plt.tight_layout()
    plt.show()

# BLOQUE 9. BOXPLOTS PRINCIPALES

graficar_variables(
    potencia_abs,
    'Potencia Absoluta'
)

graficar_variables(
    erds,
    'ERD / ERS'
)

graficar_variables(
    entropias,
    'Entropía Muestral'
)

graficar_variables(
    coherencias,
    'Coherencia'
)

# BLOQUE 10. MATRIZ DE CORRELACIÓN

plt.figure(figsize=(16, 12))

corr = df[variables_numericas].corr()

sns.heatmap(
    corr,
    cmap='coolwarm',
    center=0,
    linewidths=0.5
)

plt.title('Matriz de Correlación')

plt.tight_layout()

plt.show()

# BLOQUE 11. VARIABLES IMPORTANTES

variables_importantes = [
    'Mu_C3',
    'Mu_C4',
    'Beta_C3',
    'Beta_C4',
    'ERD_Mu_C3',
    'ERD_Mu_C4',
    'Entropia_C3',
    'Entropia_C4'
]

variables_importantes = [
    var for var in variables_importantes
    if var in df.columns
]

# BLOQUE 12. GRÁFICA DE MEDIA ± SD

fig, axes = plt.subplots(
    4,
    2,
    figsize=(15, 16)
)

axes = axes.flatten()

for i, var in enumerate(variables_importantes):

    sns.barplot(
        data=df,
        x='Condicion',
        y=var,
        order=orden,
        errorbar='sd',
        ax=axes[i]
    )

    axes[i].set_title(var)
    axes[i].set_xlabel('')
    axes[i].set_ylabel('Media ± SD')

plt.suptitle(
    'Media ± SD de Variables Importantes',
    fontsize=18
)

plt.tight_layout()

plt.show()

# BLOQUE 13. DISTRIBUCIONES


fig, axes = plt.subplots(
    4,
    2,
    figsize=(15, 16)
)

axes = axes.flatten()

for i, var in enumerate(variables_importantes):

    sns.histplot(
        data=df,
        x=var,
        hue='Condicion',
        hue_order=orden,
        kde=True,
        element='step',
        ax=axes[i]
    )

    axes[i].set_title(var)

plt.suptitle(
    'Distribución de Variables Importantes',
    fontsize=18
)

plt.tight_layout()

plt.show()
```

Finalmente, se aplicó una normalización Min-Max para llevar todas las variables a una escala común entre 0 y 1, facilitando la comparación entre características con diferentes magnitudes y mejorando la interpretación de los resultados estadísticos y gráficos.

En este bloque se carga el archivo CSV que contiene todas las características EEG extraídas previamente durante el procesamiento de las señales. La selección explícita de estas variables garantiza que únicamente los índices de interés sean procesados.

En esta etapa se verifica que todas las características seleccionadas se encuentren almacenadas como valores numéricos. Esto es necesario porque algunos datos pueden haber sido guardados como texto o contener caracteres incompatibles con los análisis matemáticos posteriores. La función utilizada convierte automáticamente los valores válidos a formato numérico y reemplaza los datos inválidos por valores faltantes (NaN), permitiendo identificarlos y tratarlos posteriormente.

**4. TRATAMIENTO DE VALORES FALTANTES**

Una vez identificados los valores faltantes, estos se reemplazan por cero para evitar errores durante los procesos de detección de valores atípicos y normalización. Este procedimiento garantiza que todas las observaciones tengan un valor definido en cada característica y permite continuar con el análisis sin eliminar registros completos del conjunto de datos.

**5. DETECCIÓN Y CORRECCIÓN DE OUTLIERS MEDIANTE IQR**

Este bloque tiene como objetivo reducir la influencia de valores atípicos extremos que puedan distorsionar el análisis estadístico. Para cada característica se calculan el primer cuartil (Q1), el tercer cuartil (Q3) y el rango intercuartílico (IQR). A partir de estos valores se establecen límites inferiores y superiores aceptables según el criterio de 1.5 veces el IQR. En lugar de eliminar las observaciones que exceden dichos límites, se utiliza una estrategia de *clipping*, donde los valores extremos son reemplazados por el límite correspondiente más cercano. Esta decisión permite conservar todas las muestras originales, evitando la pérdida de información, mientras se reduce el impacto que los valores anómalos podrían tener sobre el análisis estadístico y los modelos de clasificación.

**6. NORMALIZACIÓN DE LAS CARACTERÍSTICAS**

Después de corregir los valores atípicos, se realiza una normalización utilizando el método Min-Max. Este procedimiento transforma todas las características a una escala común entre 0 y 1. La normalización es especialmente importante porque las variables EEG poseen unidades y rangos de magnitud muy diferentes; por ejemplo, la potencia espectral puede tener valores mucho mayores que la entropía o la coherencia. Al llevar todas las características a una misma escala se evita que algunas variables dominen el análisis únicamente por su magnitud numérica y se facilita la comparación entre ellas.

**7. VERIFICACIÓN DE RESULTADOS**

En esta etapa se muestran dimensiones y ejemplos del conjunto de datos procesado. Se compara el tamaño del DataFrame original con el DataFrame corregido para verificar que no se hayan eliminado observaciones durante el tratamiento de outliers. Además, se visualizan las primeras filas del conjunto de datos normalizado con el fin de comprobar que las transformaciones se realizaron correctamente y que los valores se encuentran dentro del rango esperado.

**8. ALMACENAMIENTO DEL DATASET LIMPIO**

Finalmente, el conjunto de datos corregido y normalizado se exporta a un nuevo archivo CSV. Este archivo constituye la versión final preparada para los análisis estadísticos, visualizaciones y posibles modelos de clasificación. Guardar una copia procesada permite reutilizar los datos en etapas posteriores sin necesidad de repetir todo el procedimiento de limpieza y normalización.


```python
# 1. CARGAR CSV

df = pd.read_csv("Resultados_P2.csv")

# 2. FEATURES A UTILIZAR

features = [
    'Mu_C3', 'Beta_C3',
    'Mu_Cz', 'Beta_Cz',
    'Mu_C4', 'Beta_C4',
    'ERD_Mu_C3', 'ERD_Beta_C3',
    'ERD_Mu_Cz', 'ERD_Beta_Cz',
    'ERD_Mu_C4', 'ERD_Beta_C4',
    'Entropia_C3',
    'Entropia_Cz',
    'Entropia_C4',
    'Coherencia_Mu_C3-Cz',
    'Coherencia_Beta_C3-Cz',
    'Coherencia_Mu_C3-C4',
    'Coherencia_Beta_C3-C4',
    'Coherencia_Mu_Cz-C4',
    'Coherencia_Beta_Cz-C4']

# 3. CONVERTIR A NUMERICO

for col in features:

    df[col] = pd.to_numeric(df[col], errors='coerce')

# 4. RELLENAR NaN CON 0

df[features] = df[features].fillna(0)

# 5. REEMPLAZAR OUTLIERS USANDO IQR Y CLIPPING

df_nuevo = df.copy()

for col in features:

    Q1 = df[col].quantile(0.25)

    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    limite_inferior = Q1 - 1.5 * IQR

    limite_superior = Q3 + 1.5 * IQR

    # CLIPPING
    df_nuevo[col] = np.clip(
        df_nuevo[col],
        limite_inferior,
        limite_superior)

# 6. NORMALIZACION Z-SCORE

scaler = MinMaxScaler()

df_nuevo[features] = scaler.fit_transform(df_nuevo[features])

# 7. RESULTADOS

print("\nDataFrame original:")
print(df.shape)

print("\nDataFrame sin outliers:")
print(df_nuevo.shape)

print("\nPrimeras filas normalizadas:\n")

print(df_nuevo.head())

# 8. GUARDAR CSV LIMPIO

df_nuevo.to_csv(
    "Resultados_P2_Limpio_Normalizado.csv",
    index=False
)

print("\nCSV guardado correctamente")
```

## **DEFINICIÓN DE CLASES Y CARACTERÍSTICAS DE ANÁLISIS**

En este bloque se preparan los elementos fundamentales para el análisis estadístico y la posterior etapa de clasificación. Por un lado, se asignan etiquetas descriptivas a las condiciones experimentales para facilitar la interpretación de los resultados. Por otro, se define el conjunto de características neurofisiológicas que será evaluado, incluyendo medidas espectrales, de desincronización, complejidad y conectividad cerebral. Estas variables constituyen la base sobre la cual se realizarán las pruebas de hipótesis para identificar posibles diferencias entre las clases y determinar cuáles características pueden resultar más relevantes para la discriminación de estados motores.

```python
df_nuevo['Condicion'] = df_nuevo['Evento'].map({
    'T0': 'Reposo',
    'T1': 'Mano Izquierda',
    'T2': 'Mano Derecha'})

features = [
    'Mu_C3', 'Mu_C4',
    'Beta_C3', 'Beta_C4',
    'ERD_Mu_C3', 'ERD_Mu_C4',
    'ERD_Beta_C3', 'ERD_Beta_C4',
    'Coherencia_Mu_C3-Cz', 'Coherencia_Beta_C3-Cz',
    'Coherencia_Mu_Cz-C4', 'Coherencia_Beta_Cz-C4']
```

## **PRUEBA DE NORMALIDAD**

En este bloque se aplica la prueba de Shapiro-Wilk a cada característica dentro de las distintas condiciones experimentales. El objetivo es verificar si los datos siguen una distribución normal, información necesaria para seleccionar las pruebas estadísticas más adecuadas en el análisis posterior.

```python
print("=" * 60)
print("PRUEBA DE NORMALIDAD — SHAPIRO-WILK")
print("=" * 60)

for feature in features:

    print(f"\n{feature}")
    print("-" * 40)

    for cond in df_nuevo['Condicion'].unique():

        datos = df_nuevo[df_nuevo['Condicion'] == cond][feature]

        # remover NaN
        datos = datos.dropna()

        stat, p = shapiro(datos)

        resultado = ("Normal" if p > 0.05 else "No normal")

        print(f"{cond:20s} p = {p:.5f}  → {resultado}")
```

## **PRUEBA DE FRIEDMAN**

En este bloque se aplica la prueba no paramétrica de Friedman para evaluar si existen diferencias significativas en cada característica entre las tres condiciones experimentales. El objetivo es identificar aquellas variables que presentan cambios relevantes entre los estados analizados.

```python
print("=" * 70)
print("PRUEBA DE FRIEDMAN — DIFERENCIAS ENTRE CONDICIONES")
print("=" * 70)

for feature in features:

    reposo = df_nuevo[df_nuevo['Condicion'] == 'Reposo'][feature]
    izquierda = df_nuevo[df_nuevo['Condicion'] == 'Mano Izquierda'][feature]
    derecha = df_nuevo[df_nuevo['Condicion'] == 'Mano Derecha'][feature]

    stat, p = friedmanchisquare(reposo, izquierda, derecha)

    print(f"\n{feature}")
    print("-" * 50)
    print(f"Estadístico χ² = {stat:.4f}")
    print(f"p-value = {p:.5f}")

    if p < 0.05:
        print("Conclusión: Existen diferencias significativas.")
    else:
        print("Conclusión: No existen diferencias significativas.")
```

## **ANÁLISIS POST-HOC CON WILCOXON**

En este bloque se realizan comparaciones pareadas entre las condiciones experimentales utilizando la prueba de Wilcoxon. El objetivo es identificar específicamente entre qué pares de condiciones se encuentran las diferencias significativas detectadas previamente por la prueba de Friedman.

```python
comparaciones = [
    ('Reposo', 'Mano Izquierda'),
    ('Reposo', 'Mano Derecha'),
    ('Mano Izquierda', 'Mano Derecha')]

print("=" * 70)
print("POST-HOC — WILCOXON PAREADO")
print("=" * 70)

for feature in features:

    print(f"\n\n{feature}")
    print("=" * 50)

    for c1, c2 in comparaciones:

        datos1 = df_nuevo[df_nuevo['Condicion'] == c1][feature]
        datos2 = df_nuevo[df_nuevo['Condicion'] == c2][feature]

        stat, p = wilcoxon(datos1, datos2)

        print(f"\n{c1} vs {c2}")
        print(f"W = {stat:.4f}")
        print(f"p = {p:.5f}")

        if p < 0.05:
            print("Diferencia significativa")
        else:
            print("No significativa")
```

## **Conclusión general de la hipótesis**

La hipótesis planteada establecía que existen diferencias significativas en la actividad cerebral asociada a las condiciones de reposo, imaginación de movimiento de mano izquierda e imaginación de movimiento de mano derecha.

Con base en los resultados obtenidos mediante las pruebas de Friedman y Wilcoxon, esta hipótesis fue respaldada para las características espectrales, los índices de desincronización cortical y las medidas de conectividad funcional. En particular, se observaron diferencias significativas en:

Por el contrario, la hipótesis no fue respaldada para la entropía muestral, ya que no se observaron diferencias estadísticamente significativas entre las condiciones experimentales.

## **5.1. Estrategia de entrenamiento basada en k-fold**

K-fold cross validation es una estrategia donde:

el dataset se divide en K partes,
el modelo se entrena con K−1 partes,
y se evalúa con la parte restante.

Esto se repite K veces.

**1. PREPARACIÓN DEL CONJUNTO DE DATOS**

En este bloque se crea una copia del conjunto de datos que será utilizado durante la etapa de clasificación. Además, se eliminan los registros que contienen valores faltantes para garantizar que el entrenamiento y la evaluación de los modelos se realicen únicamente con muestras completas.

**2. DEFINICIÓN DE VARIABLES DE ENTRADA Y SALIDA**

Aquí se separan las características seleccionadas como variables predictoras y la condición experimental asociada a cada muestra. Esta organización permite construir la matriz de entrada del modelo y las etiquetas que se desean clasificar.

**3. CONFIGURACIÓN DE K-FOLD**

En esta sección se define la estrategia de validación cruzada mediante Stratified K-Fold. Este procedimiento divide los datos en cinco particiones manteniendo la proporción de las clases en cada una de ellas, permitiendo obtener una evaluación más robusta y representativa del desempeño del modelo.

**4. CONSTRUCCIÓN DEL PIPELINE**

Se crea un flujo de procesamiento que integra la normalización de las características y el clasificador LDA. De esta manera, cada partición de entrenamiento y prueba recibe exactamente el mismo tratamiento durante el proceso de validación.

**5. ENTRENAMIENTO Y EVALUACIÓN**

En este bloque se ejecuta el proceso de validación cruzada. Para cada partición se entrena el modelo utilizando los datos de entrenamiento y posteriormente se evalúa sobre el conjunto de prueba correspondiente. La exactitud obtenida en cada iteración se almacena para su análisis posterior.

**6. ANÁLISIS DE RESULTADOS**

Finalmente, se presentan los resultados obtenidos durante la validación cruzada. Se muestran las exactitudes individuales de cada partición, así como el promedio y la desviación estándar, permitiendo evaluar la estabilidad y capacidad de generalización del clasificador.

```python
# 1. CARGA DELDATASET

df_ml = df_nuevo.copy()

# eliminar NaN
df_ml = df_ml.dropna(subset=features)

# 2. VARIABLES X e y

X = df_ml[features]

y = df_ml['Condicion']

# 3. K-FOLD

kfold = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

# 4. PIPELINE

pipeline = Pipeline([
    ('scaler', MinMaxScaler()),
    ('lda', LinearDiscriminantAnalysis())])

# 5. ENTRENAMIENTO Y EVALUACIÓN

accuracies = []

for train_idx, test_idx in kfold.split(X, y):

    X_train = X.iloc[train_idx]
    X_test  = X.iloc[test_idx]

    y_train = y.iloc[train_idx]
    y_test  = y.iloc[test_idx]

    # entrenar
    pipeline.fit(X_train, y_train)

    # predecir
    y_pred = pipeline.predict(X_test)

    # accuracy
    acc = accuracy_score(y_test, y_pred)

    accuracies.append(acc)

# 6. RESULTADOS

print("=" * 60)
print("RESULTADOS K-FOLD — LDA")
print("=" * 60)

print(f"Accuracies por fold:")
print(np.round(accuracies, 4))

print("\nAccuracy promedio:")
print(np.mean(accuracies))

print("\nDesviación estándar:")
print(np.std(accuracies))
```

Los resultados obtenidos mediante la validación cruzada muestran que el clasificador LDA alcanzó una exactitud promedio de **47.67%**, con una desviación estándar de **0.06** entre las diferentes particiones. Aunque el desempeño fue superior al azar esperado para un problema de tres clases (aproximadamente 33.3%), la capacidad de discriminación del modelo aún es limitada.

La variación relativamente baja entre los folds indica que el comportamiento del clasificador fue consistente a lo largo de las diferentes particiones del conjunto de datos, lo que ssugiere una buena estabilidad del proceso de entrenamiento. Sin embargo, la exactitud obtenida evidencia que las características seleccionadas, aunque estadísticamente significativas, no son suficientes por sí solas para lograr una separación robusta entre las tres condiciones experimentales mediante un modelo lineal como LDA.

Estos resultados sugieren que la relación entre las características EEG y las clases podría presentar patrones más complejos o no lineales, por lo que resulta ideal explorar modelos más avanzados, como redes neuronales, máquinas de soporte vectorial o algoritmos basados en árboles de decisión, para mejorar el desempeño de la clasificación.

A diferencia del codigo anterior, donde el objetivo principal era estimar el rendimiento promedio del clasificador mediante validación cruzada K-Fold, en esta sección se realiza un análisis más detallado del comportamiento del modelo. Para ello, además de entrenar y evaluar el clasificador en cada partición, se almacenan todas las predicciones generadas durante el proceso de validación. Esto permite construir una matriz de confusión y calcular métricas de desempeño por clase, facilitando la identificación de las condiciones experimentales que el modelo logra distinguir con mayor precisión y aquellas donde existen mayores niveles de confusión.

**1. PREPARACIÓN DEL CONJUNTO DE DATOS**

En esta sección se eliminan los registros con valores faltantes y se separan las características seleccionadas de las etiquetas de clase. Esto permite construir el conjunto de entrada que será utilizado durante el entrenamiento y la evaluación del modelo.

**2. CONFIGURACIÓN DE K-FOLD**

Aquí se define una estrategia de validación cruzada estratificada con cinco particiones. Este procedimiento garantiza que cada subconjunto conserve una distribución similar de las clases presentes en el conjunto de datos original.

**3. CONSTRUCCIÓN DEL PIPELINE**

En este bloque se crea un flujo de procesamiento compuesto por una etapa de normalización y un clasificador LDA. Esto asegura que todas las particiones sean procesadas de manera uniforme durante la validación cruzada.

**4. ALMACENAMIENTO DE PREDICCIONES**

Se inicializan las estructuras donde se almacenarán las etiquetas reales y las predicciones generadas por el modelo a lo largo de todas las iteraciones del proceso de validación.

**5. ENTRENAMIENTO Y PREDICCIÓN**

En esta etapa se ejecuta el proceso de validación cruzada. Para cada fold se entrena el modelo con los datos de entrenamiento y posteriormente se generan las predicciones sobre el conjunto de prueba correspondiente.

**6. CONSTRUCCIÓN DE LA MATRIZ DE CONFUSIÓN**

Una vez finalizadas todas las iteraciones, se consolidan las predicciones obtenidas para construir la matriz de confusión, permitiendo analizar los aciertos y errores de clasificación entre las diferentes condiciones experimentales.

**7. VISUALIZACIÓN DE RESULTADOS**

Aquí se genera una representación gráfica de la matriz de confusión mediante un mapa de calor, facilitando la interpretación visual del desempeño del clasificador.

**8. CÁLCULO DE MÉTRICAS DE DESEMPEÑO**

Finalmente, se calculan métricas como precisión, recall, F1-score y exactitud global, proporcionando una evaluación cuantitativa del rendimiento alcanzado por el modelo.

```python
# 1. PREPARACIÓN DEL CONJUNTO DE DATOS

df_ml = df_ml.dropna(subset=features)

X = df_ml[features]
y = df_ml['Condicion']

# 2. CONFIGURACIÓN DE K-FOLD

kfold = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42)

# 3. CONSTRUCCIÓN DEL PIPELINE

pipeline = Pipeline([
    ('scaler', MinMaxScaler()),
    ('lda', LinearDiscriminantAnalysis())
])

# 4. ALMACENAMIENTO DE PREDICCIONES

y_true_total = []
y_pred_total = []

# 5. ENTRENAMIENTO Y PREDICCIÓN

for train_idx, test_idx in kfold.split(X, y):

    X_train = X.iloc[train_idx]
    X_test  = X.iloc[test_idx]

    y_train = y.iloc[train_idx]
    y_test  = y.iloc[test_idx]

    pipeline.fit(X_train, y_train)

    y_pred = pipeline.predict(X_test)

    y_true_total.extend(y_test)
    y_pred_total.extend(y_pred)

# 6. CONSTRUCCIÓN DE LA MATRIZ DE CONFUSIÓN

cm = confusion_matrix(
    y_true_total,
    y_pred_total,
    labels=['Reposo', 'Mano Izquierda', 'Mano Derecha'])

# 7. VISUALIZACIÓN DE RESULTADOS

plt.figure(figsize=(7,6))

sns.heatmap(
    cm,
    annot=True,
    fmt='d',
    cmap='Blues',
    xticklabels=['Reposo', 'Mano Izquierda', 'Mano Derecha'],
    yticklabels=['Reposo', 'Mano Izquierda', 'Mano Derecha'])

plt.xlabel('Predicción')
plt.ylabel('Clase real')

plt.title('Matriz de Confusión — LDA')

plt.tight_layout()
plt.show()

# 8. CÁLCULO DE MÉTRICAS DE DESEMPEÑO

print("=" * 60)
print("REPORTE DE CLASIFICACIÓN")
print("=" * 60)

print(classification_report(
    y_true_total,
    y_pred_total))

print("\nAccuracy global:")
print(accuracy_score(
    y_true_total,
    y_pred_total))
```

### **Análisis de resultados**

La matriz de confusión muestra que el clasificador LDA logra diferenciar parcialmente las tres condiciones experimentales, aunque presenta confusiones importantes entre las tareas de imaginación motora. Las clases con mayor número de aciertos fueron **Mano Derecha** (61) y **Reposo** (57), mientras que **Mano Izquierda** fue la condición más difícil de clasificar correctamente (38).

Las principales confusiones se observan entre **Mano Izquierda** y las demás clases, lo que sugiere un solapamiento entre los patrones EEG asociados a las distintas tareas motoras imaginadas. Desde el punto de vista fisiológico, este comportamiento es esperable debido a que ambas tareas activan regiones sensorimotoras similares y comparten mecanismos de desincronización en las bandas Mu y Beta.

La matriz de confusión se representa mediante un mapa de calor para observar gráficamente los aciertos y errores de clasificación.

**13. CÁLCULO DE MÉTRICAS DE DESEMPEÑO**

Finalmente, se calculan métricas como *precision*, *recall*, *F1-score* y *accuracy global*, permitiendo evaluar el desempeño de la DNN en la discriminación de las diferentes condiciones experimentales.

```python
# 1. PREPARACIÓN DEL CONJUNTO DE DATOS

df_dnn = df_nuevo.copy()

df_dnn = df_dnn.dropna(subset=features)

X = df_dnn[features]
y = df_dnn['Condicion']

# 2. CODIFICACIÓN DE LAS CLASES

encoder = LabelEncoder()

y_encoded = encoder.fit_transform(y)

# 3. DIVISIÓN EN ENTRENAMIENTO Y PRUEBA

X_train, X_test, y_train, y_test = train_test_split(X, y_encoded,
    test_size=0.3,
    random_state=42,
    stratify=y_encoded
)

# 4. NORMALIZACIÓN DE LAS VARIABLES

scaler = MinMaxScaler()

X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

# 5. CODIFICACIÓN ONE-HOT DE LAS CLASES

y_train_cat = to_categorical(y_train)

# 6. CONSTRUCCIÓN DE LA RED NEURONAL PROFUNDA

model = Sequential()

model.add(Dense(128,
    activation='relu',
    input_shape=(X_train.shape[1],)))

model.add(Dense(64,
    activation='relu'))

model.add(Dense(32,
    activation='relu'))

model.add(Dense(3,
    activation='softmax'))

# 7. COMPILACIÓN DEL MODELO

model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy'])

# 8. ENTRENAMIENTO DEL MODELO

history = model.fit(X_train, y_train_cat,
    epochs=100,
    batch_size=16,
    verbose=1)

# 9. PREDICCIÓN DE LAS CLASES

y_pred = np.argmax(model.predict(X_test), axis=1)

# 10. DECODIFICACIÓN DE ETIQUETAS

y_test_labels = encoder.inverse_transform(y_test)
y_pred_labels = encoder.inverse_transform(y_pred)

# 11. CONSTRUCCIÓN DE LA MATRIZ DE CONFUSIÓN

cm = confusion_matrix(y_test_labels, y_pred_labels,
    labels=['Reposo', 'Mano Izquierda', 'Mano Derecha'])

# 12. VISUALIZACIÓN DE RESULTADOS

plt.figure(figsize=(7,6))

sns.heatmap(cm,
    annot=True,
    fmt='d',
    cmap='Greens',
    xticklabels=['Reposo', 'Mano Izquierda', 'Mano Derecha'],
    yticklabels=['Reposo', 'Mano Izquierda', 'Mano Derecha'])

plt.xlabel('Predicción')
plt.ylabel('Clase real')
plt.title('Matriz de Confusión — DNN')
plt.tight_layout()
plt.show()

# 13. CÁLCULO DE MÉTRICAS DE DESEMPEÑO

print("=" * 60)
print("REPORTE DE CLASIFICACIÓN — DNN")
print("=" * 60)

print(classification_report(y_test_labels, y_pred_labels))

print("\nAccuracy global:")

print(accuracy_score(y_test_labels, y_pred_labels))
```

#### **Análisis de resultados — DNN**

La red neuronal profunda obtuvo una accuracy global de 69.7%, mostrando un desempeño aceptable, aunque inferior al obtenido con el modelo MLP.

La matriz de confusión muestra que la condición de Reposo fue identificada correctamente en todas las muestras, alcanzando un recall de 1.00. Esto indica que el modelo logró diferenciar claramente el estado de reposo de las tareas de imaginación motora.

Las principales confusiones ocurrieron entre Mano Izquierda y Mano Derecha, especialmente para la clase Mano Derecha, que presentó un recall de 0.45. Esto sugiere que la red tuvo dificultades para distinguir completamente la lateralidad del movimiento imaginado.

En general, la DNN logró separar adecuadamente el reposo de las tareas motoras, pero no mejoró la discriminación entre lateralidades respecto a arquitecturas más simples, posiblemente debido al tamaño limitado del conjunto de datos.

## **5.2.3 Red Neuronal con Regularización y Dropout**

La tercera arquitectura incorpora técnicas de regularización orientadas a reducir el sobreajuste y mejorar la capacidad de generalización del modelo. Para ello se emplean mecanismos como Dropout y regularización L2, los cuales ayudan a controlar la complejidad de la red durante el entrenamiento.

### **Discusión del modelo**

Las señales EEG suelen presentar ruido y una elevada variabilidad entre sujetos, por lo que una red neuronal puede llegar a aprender patrones específicos del conjunto de entrenamiento en lugar de características fisiológicas generalizables.

La incorporación de técnicas de regularización busca minimizar este problema. En particular, el Dropout desactiva aleatoriamente una parte de las neuronas durante el entrenamiento, mientras que la regularización L2 penaliza pesos excesivamente grandes, favoreciendo modelos más estables.

En aplicaciones BCI, estas estrategias suelen mejorar la capacidad de generalización frente a nuevos datos y permiten evaluar si características como la potencia espectral, los índices ERD y la conectividad funcional contienen patrones suficientemente robustos para diferenciar las condiciones experimentales.

En este codigo se implementa una red neuronal con técnicas de regularización para reducir el sobreajuste y mejorar la capacidad de generalización. A diferencia de la DNN anterior, esta arquitectura incorpora capas Dropout y regularización L2, permitiendo evaluar si los patrones fisiológicos aprendidos se mantienen robustos frente a mecanismos que limitan la complejidad del modelo.

**1. PREPARACIÓN DEL CONJUNTO DE DATOS**

Se prepara el dataset eliminando registros con valores faltantes y separando las variables predictoras (**X**) de las etiquetas de clasificación (**y**).

**2. CODIFICACIÓN DE LAS CLASES**

Las condiciones experimentales se transforman a valores numéricos mediante **LabelEncoder**, permitiendo que sean utilizadas por la red neuronal.

**3. DIVISIÓN EN ENTRENAMIENTO Y PRUEBA**

El conjunto de datos se divide en subconjuntos de entrenamiento y prueba utilizando una partición estratificada para conservar el equilibrio entre clases.

**4. NORMALIZACIÓN DE LAS VARIABLES**

Las características EEG son escaladas al rango entre 0 y 1 mediante **MinMaxScaler**, favoreciendo la estabilidad del entrenamiento.

**5. CODIFICACIÓN ONE-HOT DE LAS CLASES**

Las etiquetas numéricas se convierten a formato **One-Hot Encoding**, requerido para la clasificación multiclase mediante Softmax.

**6. CONSTRUCCIÓN DE LA RED REGULARIZADA**

Se define una red neuronal compuesta por varias capas densas, incorporando regularización L2 y capas Dropout. Estas técnicas ayudan a reducir el sobreajuste y favorecen el aprendizaje de representaciones más robustas.

**7. COMPILACIÓN DEL MODELO**

Se configura el entrenamiento utilizando el optimizador Adam, la función de pérdida *categorical crossentropy* y la métrica de exactitud (*accuracy*).

**8. ENTRENAMIENTO DEL MODELO**

La red neuronal es entrenada durante 100 épocas ajustando sus parámetros internos para minimizar el error de clasificación.

**9. PREDICCIÓN DE LAS CLASES**

Una vez finalizado el entrenamiento, el modelo genera predicciones sobre el conjunto de prueba.

**10. DECODIFICACIÓN DE ETIQUETAS**

Las predicciones y etiquetas reales se convierten nuevamente a sus nombres originales para facilitar la interpretación de los resultados.

**11. CONSTRUCCIÓN DE LA MATRIZ DE CONFUSIÓN**

Se calcula la matriz de confusión comparando las clases reales con las clases estimadas por la red neuronal.

**12. VISUALIZACIÓN DE RESULTADOS**

La matriz de confusión se representa mediante un mapa de calor que permite observar los aciertos y errores de clasificación para cada condición experimental.

**13. CÁLCULO DE MÉTRICAS DE DESEMPEÑO**

Finalmente, se calculan métricas como *precision*, *recall*, *F1-score* y *accuracy global* para evaluar el desempeño de la red regularizada.

```python
# 1. PREPARACIÓN DEL CONJUNTO DE DATOS

df_ryd = df_nuevo.copy()

df_ryd = df_ryd.dropna(subset=features)

X = df_ryd[features]
y = df_ryd['Condicion']

# 2. CODIFICACIÓN DE LAS CLASES

encoder = LabelEncoder()

y_encoded = encoder.fit_transform(y)

# 3. DIVISIÓN EN ENTRENAMIENTO Y PRUEBA

X_train, X_test, y_train, y_test = train_test_split(X, y_encoded,
    test_size=0.3,
    random_state=42,
    stratify=y_encoded)

# 4. NORMALIZACIÓN DE LAS VARIABLES

scaler = MinMaxScaler()

X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

# 5. CODIFICACIÓN ONE-HOT DE LAS CLASES

y_train_cat = to_categorical(y_train)

# 6. CONSTRUCCIÓN DE LA RED REGULARIZADA

model = Sequential()

model.add(Dense(128,
    activation='relu',
    kernel_regularizer=l2(0.001),
    input_shape=(X_train.shape[1],)))

model.add(Dropout(0.4))

model.add(Dense(64,
    activation='relu',
    kernel_regularizer=l2(0.001)))

model.add(Dropout(0.3))

model.add(Dense(32,
    activation='relu'))

model.add(Dense(3,
    activation='softmax'))

# 7. COMPILACIÓN DEL MODELO

model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy'])

# 8. ENTRENAMIENTO DEL MODELO

history = model.fit(X_train, y_train_cat,
    epochs=100,
    batch_size=16,
    verbose=1)

# 9. PREDICCIÓN DE LAS CLASES

y_pred = np.argmax(model.predict(X_test), axis=1)

# 10. DECODIFICACIÓN DE ETIQUETAS

y_test_labels = encoder.inverse_transform(y_test)
y_pred_labels = encoder.inverse_transform(y_pred)

# 11. CONSTRUCCIÓN DE LA MATRIZ DE CONFUSIÓN

cm = confusion_matrix(y_test_labels, y_pred_labels,
    labels=['Reposo', 'Mano Izquierda', 'Mano Derecha'])

# 12. VISUALIZACIÓN DE RESULTADOS

plt.figure(figsize=(7,6))

sns.heatmap(cm,
    annot=True,
    fmt='d',
    cmap='Reds',
    xticklabels=['Reposo', 'Mano Izquierda', 'Mano Derecha'],
    yticklabels=['Reposo', 'Mano Izquierda', 'Mano Derecha'])

plt.xlabel('Predicción')
plt.ylabel('Clase real')
plt.title('Matriz de Confusión — Regularización y Dropout')
plt.tight_layout()
plt.show()

# 13. CÁLCULO DE MÉTRICAS DE DESEMPEÑO

print("=" * 60)
print("REPORTE DE CLASIFICACIÓN — RED REGULARIZADA")
print("=" * 60)

print(classification_report(y_test_labels, y_pred_labels))

print("\nAccuracy global:")

print(accuracy_score(y_test_labels, y_pred_labels))
```

### **Análisis de resultados — Red Neuronal con Regularización y Dropout**

La red regularizada obtuvo una exactitud global de 72.7%, superior a la DNN (69.7%), pero inferior al MLP (78.8%).

La condición Reposo fue clasificada correctamente en todas las muestras, alcanzando un recall de 1.00, lo que confirma que esta condición se diferencia claramente de las tareas de imaginación motora.

Las principales confusiones ocurrieron entre Mano Izquierda y Mano Derecha, especialmente en la clasificación de mano izquierda, que presentó el menor recall (0.52). Esto indica que la discriminación entre lateralidades sigue siendo el principal desafío del problema.

En general, la incorporación de dropout y regularización L2 permitió mantener un desempeño estable y reducir el riesgo de sobreajuste, aunque sin superar los resultados obtenidos con la arquitectura MLP.
