## **1.** ***Consulta*** 

### 1.1 ¿Qué índices del EEG mejoran la clasificación de estados en BCI?

En los sistemas de interfaz cerebro-computador o Brain-Computer Interface (BCI), la clasificación de estados mentales o motores a partir de señales de EEG puede mejorarse utilizando índices adicionales a la densidad espectral de potencia (PSD, Power Spectral Density). Aunque la PSD es una de las características más empleadas debido a su capacidad para describir la distribución de energía en diferentes bandas de frecuencia, existen múltiples métricas complementarias que permiten capturar propiedades temporales, espaciales y no lineales de la señal cerebral, aumentando así la precisión y robustez de los clasificadores.

Uno de los índices más utilizados es el de Common Spatial Patterns (CSP), una técnica espacial que maximiza la varianza de una clase mientras minimiza la de otra. CSP resulta especialmente útil en tareas de imaginación motora, ya que resalta diferencias entre regiones corticales activadas durante distintos movimientos imaginados. Este método ha demostrado mejorar significativamente la discriminación entre clases en sistemas BCI basados en EEG, especialmente cuando se combina con características espectrales y clasificadores lineales [1].

Otro grupo importante corresponde a las medidas de conectividad cerebral, como la coherencia, la sincronización de fase (Phase Locking Value, PLV) y la coherencia imaginaria. Estas métricas evalúan la interacción funcional entre distintas regiones cerebrales, permitiendo identificar patrones de comunicación neuronal asociados a diferentes estados cognitivos o motores. La inclusión de información de conectividad ha mostrado mejoras en la clasificación de tareas mentales complejas, ya que el cerebro funciona como una red distribuida y no únicamente mediante activaciones locales [2].

También se utilizan ampliamente índices basados en la entropía y complejidad de la señal, entre ellos la entropía aproximada (Approximate Entropy), la entropía muestral (Sample Entropy) y la entropía espectral. Estas métricas cuantifican el grado de irregularidad o complejidad del EEG, permitiendo distinguir entre diferentes estados mentales incluso cuando las diferencias espectrales son pequeñas. En aplicaciones BCI, las características no lineales han demostrado complementar adecuadamente las medidas clásicas de potencia [3].

Además, las características temporales y tiempo-frecuencia obtenidas mediante transformadas wavelet o transformada corta de Fourier (STFT) permiten analizar cómo evolucionan las frecuencias cerebrales en el tiempo. Las wavelets son particularmente útiles en EEG debido a la naturaleza no estacionaria de las señales cerebrales. Estas representaciones permiten detectar eventos transitorios relacionados con desincronización y resincronización cortical, mejorando la detección de intención motora [4].

Otro índice relevante corresponde a la asimetría hemisférica, calculada a partir de diferencias de potencia o coherencia entre hemisferios cerebrales. Este tipo de métrica es especialmente útil en tareas motoras, ya que durante movimientos reales o imaginados suele producirse desincronización contralateral en las bandas mu y beta. La asimetría puede aumentar la sensibilidad del sistema para identificar lateralidad del movimiento [5].

Finalmente, en investigaciones recientes se han incorporado técnicas basadas en aprendizaje profundo que extraen automáticamente características espaciales y temporales desde el EEG crudo mediante redes convolucionales y recurrentes. Aunque estas técnicas no representan un “índice” tradicional, sí constituyen una alternativa moderna para mejorar la clasificación en BCI, al combinar simultáneamente múltiples representaciones de la señal [6].

En conclusión, aunque la densidad espectral de potencia continúa siendo una herramienta fundamental en el análisis de EEG para BCI, la incorporación de índices espaciales, de conectividad, complejidad y tiempo-frecuencia permite capturar información adicional del comportamiento cerebral. La combinación de estas características suele producir sistemas de clasificación más precisos y robustos, especialmente en tareas de imaginación motora y reconocimiento de estados cognitivos.

### 1.2 Tres índices que se puedan implementar en Python

Se seleccionaron tres índices adicionales que pueden implementarse fácilmente en Python utilizando librerías ampliamente disponibles como NumPy, SciPy, MNE y AntroPy. Los índices escogidos permiten analizar diferentes propiedades del EEG: conectividad funcional, complejidad de la señal y discriminación espacial de la actividad cerebral. Estos índices son muy utilizados en aplicaciones reales de interfaces cerebro-computador debido a su capacidad para mejorar la clasificación de estados motores y cognitivos.

    a) Coherencia (Coherence)

La coherencia es una medida de conectividad funcional que evalúa el grado de correlación entre dos señales EEG en una frecuencia específica. Matemáticamente, se calcula a partir de las densidades espectrales cruzadas y las densidades espectrales individuales de las señales. Sus valores oscilan entre 0 y 1, donde valores cercanos a 1 indican alta sincronización entre regiones cerebrales [7].

En sistemas BCI, la coherencia permite estudiar cómo interactúan diferentes áreas corticales durante tareas motoras o de imaginación motora. Por ejemplo, puede analizarse la coherencia entre electrodos C3 y C4 para evaluar cambios en la comunicación interhemisférica durante movimientos imaginados de manos. Esta medida complementa la PSD, ya que no solo analiza energía local sino también relaciones funcionales entre canales [2].

La implementación en Python es sencilla utilizando la función scipy.signal.coherence(), que calcula automáticamente la coherencia entre dos señales EEG.

Su uso en el proyecto permitiría identificar variaciones de sincronización cortical entre condiciones de reposo e imaginación motora, mejorando la separación entre clases.

    b) Entropía muestral (Sample Entropy)

La entropía muestral es una medida no lineal que cuantifica la complejidad e irregularidad de una señal temporal. Valores altos indican señales más impredecibles y complejas, mientras que valores bajos corresponden a señales más regulares y repetitivas [8].

En EEG, esta métrica resulta útil porque la actividad cerebral cambia su complejidad dependiendo del estado cognitivo o motor. Durante tareas de concentración o imaginación motora suelen presentarse variaciones en la organización temporal de las señales, lo cual puede ser capturado mediante entropía. Esta característica aporta información complementaria a la PSD, especialmente cuando las diferencias espectrales entre clases son pequeñas [4].

La implementación en Python es muy práctica mediante la librería antropy, utilizando la función sample_entropy().

En el contexto del proyecto, la entropía muestral podría calcularse para cada canal EEG y compararse entre diferentes tareas motoras o períodos de reposo, evaluando cambios en la complejidad cortical asociados a la activación cerebral.

    c) Common Spatial Patterns (CSP)

Common Spatial Patterns (CSP) es una técnica de filtrado espacial ampliamente utilizada en BCI para maximizar las diferencias entre dos clases de señales EEG. CSP encuentra combinaciones lineales de canales que aumentan la varianza de una clase mientras reducen la de la otra, permitiendo resaltar regiones cerebrales relevantes para la clasificación [9].

Esta técnica es especialmente efectiva en tareas de imaginación motora, ya que enfatiza patrones de desincronización y resincronización en áreas sensorimotoras. CSP suele combinarse con características espectrales y clasificadores como LDA o SVM para mejorar significativamente la precisión de los sistemas BCI [2].

Aunque matemáticamente es más complejo que las medidas anteriores, su implementación en Python es relativamente sencilla gracias a la librería MNE.

En este proyecto, CSP podría utilizarse para diferenciar tareas de imaginación motora derecha e izquierda, resaltando los patrones espaciales asociados a la actividad contralateral observada en las bandas mu y beta.

La coherencia, la entropía muestral y CSP representan tres índices complementarios que aportan información funcional, no lineal y espacial del EEG. La combinación de estas características con la densidad espectral de potencia puede mejorar considerablemente el rendimiento de clasificación en aplicaciones BCI.

### 2.1 Metodología de procesamiento y análisis para T0, T1 y T2 en imaginacion

Para evidenciar las diferencias en los ritmos cerebrales asociados a las condiciones de reposo, imaginación de movimiento de mano derecha e imaginación de movimiento de mano izquierda, se propone una metodología de análisis basada en el procesamiento espectral, espacial y funcional de señales EEG. Esta metodología integra tanto las herramientas desarrolladas en el Proyecto 1, centradas en densidad espectral de potencia, como los índices adicionales seleccionados previamente (coherencia, entropia muestral y CSP), con el fin de obtener una caracterización más completa de la actividad cortical relacionada con imaginación motora.

La metodología propuesta inicia con la adquisición y organización de las señales EEG correspondientes a las tres condiciones experimentales: reposo, imaginación motora de mano derecha e imaginación motora de mano izquierda. Para cada condición, las señales deben almacenarse separadamente y etiquetarse adecuadamente para facilitar el análisis posterior y el entrenamiento de posibles clasificadores. Debido a que la imaginación motora activa principalmente la corteza sensorimotora, es recomendable priorizar canales ubicados sobre regiones centrales como C3, Cz y C4, ya que estas áreas presentan cambios significativos en las bandas mu y beta durante tareas motoras [5].

Posteriormente, se realiza una etapa de preprocesamiento de las señales EEG. En esta fase se aplican filtros digitales para reducir ruido e interferencias presentes en la señal. Se recomienda utilizar:

- Un filtro pasa banda entre 0.5 y 40 Hz para conservar las principales bandas fisiológicas del EEG.
- Un filtro notch de 50 o 60 Hz para eliminar interferencia de la red eléctrica.
- Corrección o eliminación de artefactos producidos por parpadeo, movimientos o actividad muscular, utilizando rechazo manual de segmentos o métodos automáticos disponibles en librerías como MNE [10].

Una vez filtradas las señales, se realiza la segmentación temporal en ventanas o epochs correspondientes a cada ensayo experimental. Esto permite analizar de forma independiente cada intento de imaginación motora y calcular características estadísticas robustas para cada condición.

La primera etapa de análisis corresponde al cálculo de la densidad espectral de potencia (PSD) mediante el método de Welch. Este análisis permite identificar cambios en la energía de las bandas mu (8–13 Hz) y beta (13–30 Hz), las cuales están estrechamente relacionadas con tareas motoras e imaginación motora [7]. Durante imaginación de movimiento suele presentarse desincronización relacionada con eventos (Event Related Desynchronization, ERD), observable como una disminución de potencia en dichas bandas sobre el hemisferio contralateral al movimiento imaginado. Por ejemplo:

- Imaginación de mano derecha → disminución de potencia en C3.
- Imaginación de mano izquierda → disminución de potencia en C4.

El análisis espectral puede realizarse comparando:

- PSD promedio entre condiciones.
- Potencia relativa en bandas mu y beta.
- Diferencias hemisféricas entre C3 y C4.

Para complementar la información espectral, se propone calcular la coherencia entre pares de canales motores, especialmente entre C3 y C4. Esta medida permitirá analizar cambios en la conectividad funcional durante imaginación motora. Se espera que ciertas tareas produzcan modificaciones en la sincronización interhemisférica, reflejando reorganización funcional de la actividad cortical [2]. La coherencia puede calcularse utilizando scipy.signal.coherence() y posteriormente promediarse en las bandas mu y beta.

Adicionalmente, se propone utilizar la entropía muestral (Sample Entropy) como índice de complejidad de la señal EEG. Este análisis permitirá evaluar cambios en la irregularidad temporal de la actividad cerebral durante las distintas tareas. Generalmente, estados cognitivos activos presentan variaciones en complejidad respecto al reposo [8]. La entropía podría calcularse para cada canal y compararse estadísticamente entre condiciones.

Como complemento espacial, se plantea implementar la técnica de Common Spatial Patterns (CSP), la cual permite resaltar diferencias espaciales entre imaginación de mano derecha e izquierda [9]. CSP genera filtros espaciales óptimos que maximizan la varianza de una condición y minimizan la de la otra, facilitando la discriminación entre tareas motoras. Esta técnica resulta particularmente útil si posteriormente se desea entrenar clasificadores automáticos.

Después de la extracción de características, se realizaría un análisis estadístico para determinar si existen diferencias significativas entre condiciones. Dependiendo de la distribución de los datos, podrían utilizarse:

- Prueba t pareada.
- ANOVA de medidas repetidas.
- Pruebas no paramétricas como Wilcoxon.

Estas pruebas permitirían validar si los cambios observados en potencia, coherencia o entropía son estadísticamente significativos.

Finalmente, los resultados pueden representarse mediante:

- Gráficas PSD promedio.
- Mapas topográficos de potencia.
- Curvas de coherencia.
- Diagramas de cajas (boxplots) para comparar índices entre condiciones.

La combinación de PSD, coherencia, entropía y CSP permite obtener una representación más completa de la actividad cerebral, aumentando la capacidad de identificar patrones asociados a imaginación de movimiento de mano derecha e izquierda en sistemas BCI basados en EEG.

```python
#Se importan todas las librerias neceesatias para el desarrollo del codigo que siga el flujo de procesamiento.

import mne
mne.set_log_level('ERROR')
from mne.time_frequency import psd_array_welch
import numpy as np
import pandas as pd
import os
import matplotlib.pyplot as plt
from scipy.stats import wilcoxon, kstest
from mne.decoding import CSP
from scipy.signal import coherence
import antropy as ant
import seaborn as sns
```

### 3.1 FUNCIONES DEL FLUJO DE PROCESAMIENTO

```python
def cargar_datos_eeg(ruta_base, runs_validos):

    """============================================================
    FUNCIÓN:
    ------------------------------------------------------------
    Carga y preprocesa señales EEG del dataset OpenNeuro
    ds004362 enfocándose únicamente en:

    - Reposo (T0)
    - Imaginación mano izquierda (T1)
    - Imaginación mano derecha (T2)

    ============================================================
    RETORNA:
    ------------------------------------------------------------
    datos_epochs : épocas organizadas por sujeto y evento
    df_conteo    : resumen de cantidad de épocas
    datos_raw    : señales originales y filtradas
    ============================================================"""

    # ============================================================
    # 1. IDENTIFICACIÓN DE SUJETOS
    # ============================================================

    sujetos = [s for s in os.listdir(ruta_base)
        if os.path.isdir(os.path.join(ruta_base, s))]

    sujetos.sort()

    # ============================================================
    # 2. ESTRUCTURAS DE ALMACENAMIENTO
    # ============================================================

    datos_epochs = {}

    datos_raw = {}

    conteo = []

    sujetos_resampleados = set()

    # ============================================================
    # 3. RECORRIDO DE SUJETOS
    # ============================================================

    for sujeto in sujetos:

        ruta_sujeto = os.path.join(ruta_base, sujeto)

        # --------------------------------------------------------
        # SOLO IMAGINACIÓN MOTORA
        # --------------------------------------------------------

        datos_epochs[sujeto] = {
            'T0': [],   # Reposo
            'T1': [],   # Mano izquierda
            'T2': []    # Mano derecha
}

        datos_raw[sujeto] = []

        # ========================================================
        # 4. RECORRIDO DE ARCHIVOS EDF
        # ========================================================

        for archivo in sorted(os.listdir(ruta_sujeto)):

            # ----------------------------------------------------
            # VALIDACIÓN DE ARCHIVOS
            # ----------------------------------------------------

            if not archivo.endswith('.edf'):
                continue

            if not any(r in archivo for r in runs_validos):
                continue

            # ----------------------------------------------------
            # EXTRACCIÓN DE RUN
            # ----------------------------------------------------

            run = int(archivo.split('R')[1][:2])

            ruta_archivo = os.path.join(ruta_sujeto, archivo)

            # ====================================================
            # 5. CARGA DEL ARCHIVO EEG
            # ====================================================

            raw = mne.io.read_raw_edf(ruta_archivo,
                preload=True,
                verbose=False)

            raw_original = raw.copy()

            # ====================================================
            # 6. LIMPIEZA DE NOMBRES DE CANALES
            # ====================================================

            raw.rename_channels(lambda x: x.strip().replace('.', ''))

            # ====================================================
            # 7. SELECCIÓN DE ELECTRODOS MOTORES
            # ====================================================

            raw.pick(['C3', 'Cz', 'C4'])

            # ====================================================
            # 8. AJUSTE DE FRECUENCIA DE MUESTREO
            # ====================================================

            fs = raw.info['sfreq']

            if fs != 160:

                if sujeto not in sujetos_resampleados:

                    print(f"{sujeto} fue remuestreado de {fs} Hz a 160 Hz")

                    sujetos_resampleados.add(sujeto)

                raw.resample(160, npad='auto')

            # ====================================================
            # 9. FILTRADO
            # ====================================================

            raw.filter(
                l_freq=7,
                h_freq=40,
                method='fir',
                verbose=False)

            raw_filtrada = raw.copy()

            # ====================================================
            # 10. CONVERSIÓN DE EVENTOS
            # ====================================================

            events, event_id = mne.events_from_annotations(raw,
                event_id={'T0': 1, 'T1': 2, 'T2': 3},
                verbose=False)

            # ====================================================
            # 11. CREACIÓN DE ÉPOCAS
            # ====================================================

            epochs = mne.Epochs(raw, events,
                event_id=event_id,
                tmin=-1.0,
                tmax=4.0,
                baseline=None,
                preload=True,
                verbose=False)

            # ====================================================
            # 12. SELECCIÓN SOLO DE RUNS IMAGINARIOS
            # ====================================================

            # R04, R08 y R12:
            # imaginación motora

            if run not in [4, 8, 12]:
                continue

            # ====================================================
            # 13. ALMACENAMIENTO DE ÉPOCAS
            # ====================================================

            datos_epochs[sujeto]['T0'].append(epochs['T0'])

            datos_epochs[sujeto]['T1'].append(epochs['T1'])

            datos_epochs[sujeto]['T2'].append(epochs['T2'])

            # ====================================================
            # 14. ALMACENAMIENTO DE RAW
            # ====================================================

            datos_raw[sujeto].append({
                'archivo': archivo,
                'raw_original': raw_original,
                'raw_filtrada': raw_filtrada})

        # ========================================================
        # 15. CONTEO DE ÉPOCAS
        # ========================================================

        n_T0 = sum(len(e) for e in datos_epochs[sujeto]['T0'])

        n_T1 = sum(len(e) for e in datos_epochs[sujeto]['T1'])

        n_T2 = sum(len(e) for e in datos_epochs[sujeto]['T2'])

        conteo.append({
            'Sujeto': sujeto,
            'Reposo_T0': n_T0,
            'Imaginacion_Izquierda_T1': n_T1,
            'Imaginacion_Derecha_T2': n_T2})

    # ============================================================
    # 16. CREACIÓN DEL DATAFRAME
    # ============================================================

    df_conteo = pd.DataFrame(conteo)

    # ============================================================
    # 17. MENSAJE FINAL
    # ============================================================

    print(f"\nSe procesaron correctamente {len(sujetos)} sujetos.")

    # ============================================================
    # 18. RETORNO FINAL
    # ============================================================

    return datos_epochs, df_conteo, datos_raw
```

```python
def extraer_indices_eeg(
        epochs,
        canales=['C3', 'Cz', 'C4'],
        bandas={'mu':   (8, 13), 'beta': (13, 30)},
        calcular_psd=True,
        calcular_coherencia=True,
        calcular_entropia=True,
        graficar=True,
        sujeto=None,
        evento=None):

    """============================================================
    FUNCIÓN:
    ------------------------------------------------------------
    Calcula índices EEG utilizados en BCI:

    1. PSD
    2. Potencia Mu/Beta
    3. Coherencia
    4. Entropía muestral

    ============================================================
    RETORNA:
    ------------------------------------------------------------
    resultados : dict
    ============================================================"""

    # ============================================================
    # 1. SELECCIÓN DE CANALES
    # ============================================================

    epochs = epochs.copy().pick(canales)
    data = epochs.get_data()
    sfreq = epochs.info['sfreq']
    nombres_canales = epochs.ch_names

    # ============================================================
    # 2. ESTRUCTURA DE RESULTADOS
    # ============================================================

    resultados = {
        'PSD': {},
        'Coherencia': {},
        'Entropia': {}}
    
    # ============================================================
    # NOMBRES DE EVENTOS
    # ============================================================

    nombres_eventos = {
        'T0': 'Reposo',
        'T1': 'Imaginación Mano Izquierda',
        'T2': 'Imaginación Mano Derecha'}

    # ============================================================
    # 3. PSD Y POTENCIA ESPECTRAL
    # ============================================================

    if calcular_psd:
        psd, freqs = psd_array_welch(data,
            sfreq=sfreq,
            fmin=1,
            fmax=40,
            n_fft=160,
            n_overlap=80,
            verbose=False)

        # --------------------------------------------------------
        # PROMEDIO ENTRE ÉPOCAS
        # --------------------------------------------------------

        psd_mean = psd.mean(axis=0)

        # --------------------------------------------------------
        # RECORRIDO DE CANALES
        # --------------------------------------------------------

        for i, canal in enumerate(nombres_canales):
            resultados['PSD'][canal] = {'Frecuencias': freqs, 'PSD': psd_mean[i]}

            # ----------------------------------------------------
            # POTENCIA EN BANDAS
            # ----------------------------------------------------

            for nombre_banda, (fmin, fmax) in bandas.items():

                idx = ((freqs >= fmin) & (freqs <= fmax))

                potencia = psd_mean[i, idx].mean()

                resultados['PSD'][canal][nombre_banda] = potencia

        # ========================================================
        # GRÁFICA PSD
        # ========================================================

        if graficar:
            plt.figure(figsize=(10, 5))

            for i, canal in enumerate(nombres_canales):

                plt.plot(freqs, psd_mean[i], label=canal)

            # Banda Mu
            plt.axvspan(8, 13, color='skyblue', alpha=0.3, label='Mu')

            # Banda Beta
            plt.axvspan(13, 30, color='salmon', alpha=0.3, label='Beta')

            plt.xlabel("Frecuencia [Hz]")
            plt.ylabel("PSD")
            plt.title(f"Densidad Espectral de Potencia de {sujeto} {nombres_eventos[evento]}")
            plt.legend()
            plt.grid()
            plt.show()

    # ============================================================
    # 4. COHERENCIA
    # ============================================================

    if calcular_coherencia and len(canales) >= 2:

        # --------------------------------------------------------
        # RECORRIDO ENTRE PARES DE CANALES
        # --------------------------------------------------------

        for i in range(len(canales)):

            for j in range(i + 1, len(canales)):
                
                canal_1 = canales[i]
                canal_2 = canales[j]

                # ------------------------------------------------
                # SE USA UNA ÉPOCA EJEMPLO
                # ------------------------------------------------

                señal_1 = data[:, i, :].flatten()

                señal_2 = data[:, j, :].flatten()

                # ------------------------------------------------
                # CÁLCULO DE COHERENCIA
                # ------------------------------------------------

                f_coh, coh = coherence(señal_1, señal_2, fs=sfreq, nperseg=64)

                nombre_par = f"{canal_1}-{canal_2}"

                resultados['Coherencia'][nombre_par] = {'Frecuencias': f_coh, 'Coherencia': coh}

                # ------------------------------------------------
                # COHERENCIA EN BANDAS
                # ------------------------------------------------

                for nombre_banda, (fmin, fmax) in bandas.items():

                    idx = ((f_coh >= fmin) & (f_coh <= fmax))

                    coh_media = coh[idx].mean()

                    resultados['Coherencia'][nombre_par][nombre_banda] = coh_media

        # ========================================================
        # GRÁFICA COHERENCIA
        # ========================================================

        if graficar:

            plt.figure(figsize=(10, 5))

            for par in resultados['Coherencia']:

                plt.plot(
                    resultados['Coherencia'][par]['Frecuencias'],
                    resultados['Coherencia'][par]['Coherencia'],
                    label=par)

            plt.xlabel("Frecuencia [Hz]")
            plt.ylabel("Coherencia")
            plt.title(f"Coherencia entre canales EEG de {sujeto} {nombres_eventos[evento]}")
            plt.legend()
            plt.grid()
            plt.show()

    # ============================================================
    # 5. ENTROPÍA MUESTRAL
    # ============================================================

    if calcular_entropia:

        for i, canal in enumerate(nombres_canales):

            # ----------------------------------------------------
            # SEÑAL DE UNA ÉPOCA
            # ----------------------------------------------------

            señal = data[0, i, :]

            # ----------------------------------------------------
            # SAMPLE ENTROPY
            # ----------------------------------------------------

            entropia = ant.sample_entropy(señal)

            resultados['Entropia'][canal] = entropia

        # ========================================================
        # GRÁFICA ENTROPÍA
        # ========================================================

        if graficar:

            canales_ent = list(resultados['Entropia'].keys())

            valores_ent = list(resultados['Entropia'].values())

            plt.figure(figsize=(8, 5))
            plt.bar(canales_ent,valores_ent, color=['royalblue', 'darkorange', 'seagreen'])
            plt.ylabel("Sample Entropy")
            plt.title(f"Entropía Muestral EEG de {sujeto} {nombres_eventos[evento]}")
            plt.grid()
            plt.show()

    # ============================================================
    # 6. RETORNO FINAL
    # ============================================================

    return resultados
```

```python
def calcular_CSP(
        datos_epochs,
        sujeto,
        evento_1='T1',
        evento_2='T2',
        canales=['C3', 'Cz', 'C4'],
        n_componentes=2,
        graficar=True):

    """============================================================
    FUNCIÓN:
    ------------------------------------------------------------
    Calcula Common Spatial Patterns (CSP) entre dos
    condiciones de imaginación motora.

    ============================================================
    COMPARACIONES POSIBLES:
    ------------------------------------------------------------
    - T0 vs T1  -> Reposo vs Mano izquierda
    - T0 vs T2  -> Reposo vs Mano derecha
    - T1 vs T2  -> Mano izquierda vs Mano derecha

    ============================================================
    PARÁMETROS:
    ------------------------------------------------------------
    datos_epochs : dict
        Estructura generada por cargar_datos_eeg()

    sujeto : str
        Sujeto a analizar

    evento_1 : str
        Primera condición experimental

    evento_2 : str
        Segunda condición experimental

    canales : list
        Canales EEG de interés

    n_componentes : int
        Número de componentes CSP

    graficar : bool
        Permite visualizar separación CSP

    ============================================================
    RETORNA:
    ------------------------------------------------------------
    resultados : dict
        Resultados del análisis CSP
    ============================================================"""

    # ============================================================
    # 1. OBTENCIÓN DE ÉPOCAS
    # ============================================================

    epochs_1 = datos_epochs[sujeto][evento_1]

    epochs_2 = datos_epochs[sujeto][evento_2]

    # ============================================================
    # 2. VALIDACIÓN DE DATOS
    # ============================================================

    if len(epochs_1) == 0 or len(epochs_2) == 0:

        print("No existen suficientes épocas para calcular CSP.")

        return None

    # ============================================================
    # 3. CONCATENACIÓN DE ÉPOCAS
    # ============================================================

    epochs_1 = mne.concatenate_epochs(epochs_1)

    epochs_2 = mne.concatenate_epochs(epochs_2)

    # ============================================================
    # 4. SELECCIÓN DE CANALES
    # ============================================================

    epochs_1 = epochs_1.copy().pick(canales)

    epochs_2 = epochs_2.copy().pick(canales)

    # ============================================================
    # 5. EXTRACCIÓN DE DATOS EEG
    # ============================================================

    X1 = epochs_1.get_data()

    X2 = epochs_2.get_data()

    # ============================================================
    # 6. CREACIÓN DE ETIQUETAS
    # ============================================================

    y1 = np.zeros(len(X1))

    y2 = np.ones(len(X2))

    # ============================================================
    # 7. COMBINACIÓN DE DATOS
    # ============================================================

    X = np.concatenate([X1, X2], axis=0)

    y = np.concatenate([y1, y2], axis=0)

    # ============================================================
    # 8. CREACIÓN DEL MODELO CSP
    # ============================================================

    csp = CSP(
        n_components=n_componentes,
        reg=None,
        log=True,
        norm_trace=False)

    # ============================================================
    # 9. AJUSTE DEL MODELO CSP
    # ============================================================

    X_csp = csp.fit_transform(X, y)

    # ============================================================
    # 10. ALMACENAMIENTO DE RESULTADOS
    # ============================================================

    resultados = {
        'X_csp': X_csp,
        'etiquetas': y,
        'filtros_CSP': csp.filters_,
        'patrones_CSP': csp.patterns_,
        'modelo_CSP': csp,
        'evento_1': evento_1,
        'evento_2': evento_2}

    # ============================================================
    # 11. VISUALIZACIÓN CSP
    # ============================================================

    if graficar:

        plt.figure(figsize=(10, 5))

        # --------------------------------------------------------
        # PRIMERA CLASE
        # --------------------------------------------------------

        plt.scatter(
            X_csp[y == 0, 0],
            X_csp[y == 0, 1],
            alpha=0.7,
            label=evento_1)

        # --------------------------------------------------------
        # SEGUNDA CLASE
        # --------------------------------------------------------

        plt.scatter(
            X_csp[y == 1, 0],
            X_csp[y == 1, 1],
            alpha=0.7,
            label=evento_2)

        # ============================================================
        # NOMBRES DESCRIPTIVOS DE EVENTOS
        # ============================================================

        nombres_eventos = {
            'T0': 'Reposo',
            'T1': 'Imaginación Mano Izquierda',
            'T2': 'Imaginación Mano Derecha'}

        # --------------------------------------------------------
        # CONFIGURACIÓN DE GRÁFICA
        # --------------------------------------------------------

        plt.xlabel("Componente CSP 1")
        plt.ylabel("Componente CSP 2")
        plt.title(f"CSP - {sujeto}\n {nombres_eventos[evento_1]} vs {nombres_eventos[evento_2]}")
        plt.legend()
        plt.grid()
        plt.show()

    # ============================================================
    # 12. RETORNO FINAL
    # ============================================================

    return resultados
```

```python
def visualizar(
        datos_raw,
        datos_epochs,
        evento='T1',
        canales=['C3', 'Cz', 'C4']):

    """============================================================
    FUNCIÓN:
    ------------------------------------------------------------
    Visualiza:

    1. Señal EEG original
    2. Señal EEG filtrada
    3. Época segmentada

    Para una condición experimental específica:

    - T0 : Reposo
    - T1 : Imaginación mano izquierda
    - T2 : Imaginación mano derecha

    ============================================================
    PARÁMETROS:
    ------------------------------------------------------------
    datos_raw : dict
        Señales originales y filtradas

    datos_epochs : dict
        Épocas EEG segmentadas

    evento : str
        Condición experimental a visualizar

    canales : list
        Canales EEG de interés

    ============================================================
    RETORNA:
    ------------------------------------------------------------
    No retorna valores.
    Solo genera visualizaciones.
    ============================================================"""

    # ============================================================
    # 1. RECORRIDO DE SUJETOS
    # ============================================================

    sujetos = sorted(datos_raw.keys())

    for sujeto in sujetos:

        # ========================================================
        # 2. RECORRIDO DE RUNS
        # ========================================================

        runs = datos_raw[sujeto]

        for i, run_data in enumerate(runs):

            archivo = run_data['archivo']

            print(f"Sujeto: {sujeto}")

            print(f"Archivo: {archivo}")

            # ====================================================
            # 3. OBTENCIÓN DE SEÑALES RAW
            # ====================================================

            raw_original = run_data['raw_original']

            raw_filtrada = run_data['raw_filtrada']

            # ----------------------------------------------------
            # SELECCIÓN DE CANALES
            # ----------------------------------------------------

            raw_original = (raw_original.copy().rename_channels(lambda x: x.strip().replace('.', '')).pick(canales))

            raw_filtrada = (raw_filtrada.copy().rename_channels(lambda x: x.strip().replace('.', '')).pick(canales))

            # ====================================================
            # 4. OBTENCIÓN DE ÉPOCAS
            # ====================================================

            epochs_list = datos_epochs[sujeto][evento]

            if len(epochs_list) == 0:

                continue

            # ----------------------------------------------------
            # SE TOMA LA ÉPOCA DEL MISMO RUN
            # ----------------------------------------------------

            epochs = epochs_list[i]

            epoca = epochs[0]

            # ====================================================
            # 5. EXTRACCIÓN DE DATOS
            # ====================================================

            t_raw = raw_original.times

            t_filt = raw_filtrada.times

            t_epoca = epoca.times

            señal_raw = raw_original.get_data()

            señal_filt = raw_filtrada.get_data()

            señal_epoca = epoca.get_data()[0]

            # ====================================================
            # 6. CREACIÓN DE FIGURA
            # ====================================================

            plt.figure(figsize=(13, 9))

            # ====================================================
            # 7. SEÑAL ORIGINAL
            # ====================================================

            plt.subplot(3, 1, 1)

            for idx, canal in enumerate(canales):

                plt.plot(t_raw, señal_raw[idx], label=canal)

            plt.title(f"Señal EEG Original - {evento}")
            plt.xlabel("Tiempo [s]")
            plt.ylabel("Amplitud")
            plt.legend()
            plt.grid()

            # ====================================================
            # 8. SEÑAL FILTRADA
            # ====================================================

            plt.subplot(3, 1, 2)

            for idx, canal in enumerate(canales):

                plt.plot(t_filt, señal_filt[idx], label=canal)

            plt.title("Señal EEG Filtrada (7–40 Hz)")
            plt.xlabel("Tiempo [s]")
            plt.ylabel("Amplitud")
            plt.legend()
            plt.grid()

            # ====================================================
            # 9. ÉPOCA EEG
            # ====================================================

            plt.subplot(3, 1, 3)

            for idx, canal in enumerate(canales):

                plt.plot(t_epoca, señal_epoca[idx], label=canal)

            # ----------------------------------------------------
            # NOMBRE DE LA CONDICIÓN
            # ----------------------------------------------------

            if evento == 'T0':

                condicion = "Reposo"

            elif evento == 'T1':

                condicion = "Imaginación Mano Izquierda"

            elif evento == 'T2':

                condicion = "Imaginación Mano Derecha"

            else:

                condicion = evento

            plt.title(f"Época EEG - {condicion}")
            plt.xlabel("Tiempo [s]")
            plt.ylabel("Amplitud")
            plt.legend()
            plt.grid()

            # ====================================================
            # 10. AJUSTE FINAL
            # ====================================================

            plt.tight_layout()
            plt.show()

            # ----------------------------------------------------
            # SOLO SE GRAFICA EL PRIMER EJEMPLO
            # ----------------------------------------------------

            return
```

```python
def graficar_bandas(df_indices, canales=['C3', 'Cz', 'C4']):

    """============================================================
    FUNCIÓN:
    ------------------------------------------------------------
    Grafica la potencia promedio en bandas Mu y Beta
    usando gráficas de barras agrupadas.

    Condiciones:
    - T0 → Reposo
    - T1 → Imag. Mano Izquierda
    - T2 → Imag. Mano Derecha

    ============================================================"""

    # ============================================================
    # 1. NOMBRES DE EVENTOS
    # ============================================================

    nombres_eventos = {
        'T0': 'Reposo',
        'T1': 'Imag. Izquierda',
        'T2': 'Imag. Derecha'}

    # ============================================================
    # 2. RECORRIDO DE EVENTOS
    # ============================================================

    for evento in ['T0', 'T1', 'T2']:

        # ========================================================
        # FILTRAR EVENTO
        # ========================================================

        df_evento = df_indices[df_indices['Evento'] == evento]

        # ========================================================
        # INICIALIZACIÓN
        # ========================================================

        mu_vals = []
        beta_vals = []

        # ========================================================
        # RECORRIDO DE CANALES
        # ========================================================

        for canal in canales:

            # Banda Mu
            mu = df_evento[f'Mu_{canal}'].mean() * 1e12

            mu_vals.append(mu)

            # Banda Beta
            beta = df_evento[f'Beta_{canal}'].mean() * 1e12

            beta_vals.append(beta)

        # ========================================================
        # POSICIONES
        # ========================================================

        x = np.arange(len(canales))

        # ========================================================
        # CREACIÓN DE FIGURA
        # ========================================================

        fig, axes = plt.subplots(1, 2, figsize=(12, 5))

        # ========================================================
        # BANDA MU
        # ========================================================

        axes[0].bar(x, mu_vals, width=0.6, color='skyblue')
        axes[0].set_xticks(x)
        axes[0].set_xticklabels(canales)
        axes[0].set_title("Banda Mu (8–13 Hz)")
        axes[0].set_ylabel("Potencia media (µV²/Hz)")
        axes[0].grid(alpha=0.3)

        # ========================================================
        # BANDA BETA
        # ========================================================

        axes[1].bar(x,beta_vals,width=0.6,color='salmon')
        axes[1].set_xticks(x)
        axes[1].set_xticklabels(canales)
        axes[1].set_title("Banda Beta (13–30 Hz)")
        axes[1].set_ylabel("Potencia media (µV²/Hz)")
        axes[1].grid(alpha=0.3)

        # ========================================================
        # TÍTULO GENERAL
        # ========================================================

        fig.suptitle(f"Potencia Promedio de {nombres_eventos[evento]}")
        plt.tight_layout()
        plt.show()
```

### 3.2 Rutina y DataFrame

```python
def Rutina(ruta_base, runs_validos, canales=['C3', 'Cz', 'C4']):

    """============================================================
    FUNCIÓN:
    ------------------------------------------------------------
    Ejecuta el flujo completo de procesamiento EEG para:

    - Reposo (T0)
    - Imaginación mano izquierda (T1)
    - Imaginación mano derecha (T2)

    Además calcula:

    - PSD
    - Potencia Mu/Beta
    - Coherencia
    - Entropía
    - CSP

    ============================================================
    RETORNA:
    ------------------------------------------------------------
    df_indices
    resultados_indices
    resultados_csp
    ============================================================
    """

    # ============================================================
    # 1. CARGA Y PREPROCESAMIENTO
    # ============================================================

    print("PROCESO 1:\n""Carga y procesamiento EEG...")

    datos_epochs, df_conteo, datos_raw = cargar_datos_eeg(ruta_base, runs_validos)

    print(df_conteo)

    # ============================================================
    # 2. VISUALIZACIÓN EEG
    # ============================================================

    print("PROCESO 2:\n""Visualización EEG...")

    # ------------------------------------------------------------
    # T0
    # ------------------------------------------------------------

    print("Visualización T0 -> Reposo\n")

    visualizar(datos_raw,datos_epochs, evento='T0', canales=canales)

    # ------------------------------------------------------------
    # T1
    # ------------------------------------------------------------

    print("Visualización T1 -> Mano izquierda\n")

    visualizar(datos_raw, datos_epochs, evento='T1', canales=canales)

    # ------------------------------------------------------------
    # T2
    # ------------------------------------------------------------

    print("Visualización T2 -> Mano derecha\n")

    visualizar(datos_raw, datos_epochs, evento='T2', canales=canales)

    # ============================================================
    # 3. EXTRACCIÓN DE ÍNDICES EEG
    # ============================================================

    print(
        "PROCESO 3:\n""Extracción de índices EEG...")

    resultados_indices = {}

    filas = []

    sujetos = sorted(datos_epochs.keys())

    # ============================================================
    # 4. RECORRIDO DE SUJETOS
    # ============================================================

    for idx_sujeto, sujeto in enumerate(sujetos):

        print(f"Procesando sujeto: {sujeto}")

        resultados_indices[sujeto] = {}

        # --------------------------------------------------------
        # RECORRIDO DE EVENTOS
        # --------------------------------------------------------

        for evento in ['T0', 'T1', 'T2']:

            epochs_list = datos_epochs[sujeto][evento]

            # ----------------------------------------------------
            # VALIDACIÓN
            # ----------------------------------------------------

            if len(epochs_list) == 0:

                continue

            # ----------------------------------------------------
            # CONCATENACIÓN DE ÉPOCAS
            # ----------------------------------------------------

            epochs = mne.concatenate_epochs(epochs_list)

            # ====================================================
            # SOLO GRAFICAR PRIMER SUJETO
            # ====================================================

            if idx_sujeto == 0:

                graficar = True

            else:

                graficar = False

            # ----------------------------------------------------
            # CÁLCULO DE ÍNDICES EEG
            # ----------------------------------------------------

            resultados = extraer_indices_eeg(epochs, canales=canales,
                calcular_psd=True,
                calcular_coherencia=True,
                calcular_entropia=True,
                graficar=graficar,
                sujeto=sujeto,
                evento=evento)

            resultados_indices[sujeto][evento] = resultados

            # ====================================================
            # 5. CREACIÓN DE FILA
            # ====================================================

            fila = {'Sujeto': sujeto,'Evento': evento}

            # ----------------------------------------------------
            # PSD Y ENTROPÍA
            # ----------------------------------------------------

            for canal in canales:

                fila[f'Mu_{canal}'] = (resultados['PSD'][canal]['mu'])

                fila[f'Beta_{canal}'] = (resultados['PSD'][canal]['beta'])

                fila[f'Entropia_{canal}'] = (resultados['Entropia'][canal])

            # ----------------------------------------------------
            # COHERENCIA
            # ----------------------------------------------------

            for par in resultados['Coherencia']:

                fila[f'Coherencia_Mu_{par}'] = (resultados['Coherencia'][par]['mu'])

                fila[f'Coherencia_Beta_{par}'] = (resultados['Coherencia'][par]['beta'])

            # ====================================================
            # CSP
            # ====================================================

            if evento in ['T1', 'T2']:

                try:

                    resultados_csp_temp = calcular_CSP(datos_epochs,sujeto=sujeto,
                        evento_1='T1',
                        evento_2='T2',
                        canales=canales,
                        n_componentes=2,
                        graficar=False)

                    X_csp = resultados_csp_temp['X_csp']

                    fila['CSP_1'] = (X_csp[:, 0].mean())

                    fila['CSP_2'] = (X_csp[:, 1].mean())

                except:

                    fila['CSP_1'] = np.nan

                    fila['CSP_2'] = np.nan

            else:

                fila['CSP_1'] = np.nan

                fila['CSP_2'] = np.nan

            # ----------------------------------------------------
            # GUARDAR FILA
            # ----------------------------------------------------

            filas.append(fila)


    # ============================================================
    # 6. CSP EJEMPLO VISUAL
    # ============================================================

    print(
        "PROCESO 4:\n""Cálculo de CSP")

    sujeto_ejemplo = sujetos[0]

    resultados_csp = calcular_CSP(datos_epochs, sujeto=sujeto_ejemplo,
        evento_1='T1',
        evento_2='T2',
        canales=canales,
        n_componentes=2,
        graficar=True)

    # ============================================================
    # 7. DATAFRAME FINAL
    # ============================================================

    df_indices = pd.DataFrame(filas)

    # ============================================================
    # GUARDAR CSV
    # ============================================================

    df_indices.to_csv("Resultados_P2.csv", index=False)

    print(
        "PROCESO 5:\n DataFrame final generado")

    print(df_indices.head())

    print("\nCSV guardado correctamente.")

    # ============================================================
    # GRAFICAR POTENCIA PROMEDIO
    # ============================================================

    print("\nPROCESO 6: Gráficas de potencia promedio...\n")

    graficar_bandas(

    df_indices,

    canales=canales
)

    # ============================================================
    # 8. FINALIZACIÓN
    # ============================================================

    print("PROCESAMIENTO FINALIZADO")

    # ============================================================
    # 9. RETORNO FINAL
    # ============================================================

    return (df_indices, resultados_indices, resultados_csp)
```

Existen diferencias significativas en los ritmos cerebrales asociados a las condiciones de reposo, imaginación de movimiento de mano izquierda e imaginación de movimiento de mano derecha.

- BANDA MU

*Hipótesis nula (H0)*

No existen diferencias significativas en la potencia de la banda Mu entre las condiciones de reposo, imaginación de mano izquierda e imaginación de mano derecha.

*Hipótesis alternativa (H1)*

Existen diferencias significativas en la potencia de la banda Mu entre al menos dos de las condiciones experimentales.

- BANDA BETA

*Hipótesis nula (H0)*

No existen diferencias significativas en la potencia Beta entre las condiciones experimentales.

*Hipótesis alternativa (H1)*

Existen diferencias significativas en la potencia Beta entre las condiciones experimentales.

- ENTROPÍA

*Hipótesis nula (H0)*

La entropía muestral no cambia entre las diferentes tareas motoras imaginadas y el reposo.

*Hipótesis alternativa (H1)*

La entropía muestral presenta diferencias significativas entre las condiciones experimentales.

- COHERENCIA

*Hipótesis nula (H0)*

La coherencia funcional entre regiones motoras no presenta diferencias entre condiciones.
