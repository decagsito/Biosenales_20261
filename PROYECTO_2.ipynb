### 1.1 ¿Qué índices del EEG mejoran la clasificación de estados en BCI?

En los sistemas de interfaz cerebro-computador o Brain-Computer Interface (BCI), la clasificación de estados mentales o motores a partir de señales de EEG puede mejorarse utilizando índices adicionales a la densidad espectral de potencia (PSD, Power Spectral Density). Aunque la PSD es una de las características más empleadas debido a su capacidad para describir la distribución de energía en diferentes bandas de frecuencia, existen múltiples métricas complementarias que permiten capturar propiedades temporales, espaciales y no lineales de la señal cerebral, aumentando así la precisión y robustez de los clasificadores.

Uno de los índices más utilizados es el de Common Spatial Patterns (CSP), una técnica espacial que maximiza la varianza de una clase mientras minimiza la de otra. CSP resulta especialmente útil en tareas de imaginación motora, ya que resalta diferencias entre regiones corticales activadas durante distintos movimientos imaginados. Este método ha demostrado mejorar significativamente la discriminación entre clases en sistemas BCI basados en EEG, especialmente cuando se combina con características espectrales y clasificadores lineales [1].

Otro grupo importante corresponde a las medidas de conectividad cerebral, como la coherencia, la sincronización de fase (Phase Locking Value, PLV) y la coherencia imaginaria. Estas métricas evalúan la interacción funcional entre distintas regiones cerebrales, permitiendo identificar patrones de comunicación neuronal asociados a diferentes estados cognitivos o motores. La inclusión de información de conectividad ha mostrado mejoras en la clasificación de tareas mentales complejas, ya que el cerebro funciona como una red distribuida y no únicamente mediante activaciones locales [2].

También se utilizan ampliamente índices basados en la entropía y complejidad de la señal, entre ellos la entropía aproximada (Approximate Entropy), la entropía muestral (Sample Entropy) y la entropía espectral. Estas métricas cuantifican el grado de irregularidad o complejidad del EEG, permitiendo distinguir entre diferentes estados mentales incluso cuando las diferencias espectrales son pequeñas. En aplicaciones BCI, las características no lineales han demostrado complementar adecuadamente las medidas clásicas de potencia [3].

Además, las características temporales y tiempo-frecuencia obtenidas mediante transformadas wavelet o transformada corta de Fourier (STFT) permiten analizar cómo evolucionan las frecuencias cerebrales en el tiempo. Las wavelets son particularmente útiles en EEG debido a la naturaleza no estacionaria de las señales cerebrales. Estas representaciones permiten detectar eventos transitorios relacionados con desincronización y resincronización cortical, mejorando la detección de intención motora [4].

Otro índice relevante corresponde a la asimetría hemisférica, calculada a partir de diferencias de potencia o coherencia entre hemisferios cerebrales. Este tipo de métrica es especialmente útil en tareas motoras, ya que durante movimientos reales o imaginados suele producirse desincronización contralateral en las bandas mu y beta. La asimetría puede aumentar la sensibilidad del sistema para identificar lateralidad del movimiento [5].

Finalmente, en investigaciones recientes se han incorporado técnicas basadas en aprendizaje profundo que extraen automáticamente características espaciales y temporales desde el EEG crudo mediante redes convolucionales y recurrentes. Aunque estas técnicas no representan un “índice” tradicional, sí constituyen una alternativa moderna para mejorar la clasificación en BCI, al combinar simultáneamente múltiples representaciones de la señal [6].

En conclusión, aunque la densidad espectral de potencia continúa siendo una herramienta fundamental en el análisis de EEG para BCI, la incorporación de índices espaciales, de conectividad, complejidad y tiempo-frecuencia permite capturar información adicional del comportamiento cerebral. La combinación de estas características suele producir sistemas de clasificación más precisos y robustos, especialmente en tareas de imaginación motora y reconocimiento de estados cognitivos.

### 1.2 Tres índices que se puedan implementar en Python

Se seleccionaron tres índices adicionales que pueden implementarse fácilmente en Python utilizando librerías ampliamente disponibles como NumPy, SciPy, MNE y AntroPy. Los índices escogidos permiten analizar diferentes propiedades del EEG: conectividad funcional, complejidad de la señal y discriminación espacial de la actividad cerebral. Estos índices son muy utilizados en aplicaciones reales de interfaces cerebro-computador debido a su capacidad para mejorar la clasificación de estados motores y cognitivos.

1. Coherencia (Coherence)

La coherencia es una medida de conectividad funcional que evalúa el grado de correlación entre dos señales EEG en una frecuencia específica. Matemáticamente, se calcula a partir de las densidades espectrales cruzadas y las densidades espectrales individuales de las señales. Sus valores oscilan entre 0 y 1, donde valores cercanos a 1 indican alta sincronización entre regiones cerebrales [7].

En sistemas BCI, la coherencia permite estudiar cómo interactúan diferentes áreas corticales durante tareas motoras o de imaginación motora. Por ejemplo, puede analizarse la coherencia entre electrodos C3 y C4 para evaluar cambios en la comunicación interhemisférica durante movimientos imaginados de manos. Esta medida complementa la PSD, ya que no solo analiza energía local sino también relaciones funcionales entre canales [2].

La implementación en Python es sencilla utilizando la función scipy.signal.coherence(), que calcula automáticamente la coherencia entre dos señales EEG.

Su uso en el proyecto permitiría identificar variaciones de sincronización cortical entre condiciones de reposo e imaginación motora, mejorando la separación entre clases.

2. Entropía muestral (Sample Entropy)

La entropía muestral es una medida no lineal que cuantifica la complejidad e irregularidad de una señal temporal. Valores altos indican señales más impredecibles y complejas, mientras que valores bajos corresponden a señales más regulares y repetitivas [8].

En EEG, esta métrica resulta útil porque la actividad cerebral cambia su complejidad dependiendo del estado cognitivo o motor. Durante tareas de concentración o imaginación motora suelen presentarse variaciones en la organización temporal de las señales, lo cual puede ser capturado mediante entropía. Esta característica aporta información complementaria a la PSD, especialmente cuando las diferencias espectrales entre clases son pequeñas [4].

La implementación en Python es muy práctica mediante la librería antropy, utilizando la función sample_entropy().

En el contexto del proyecto, la entropía muestral podría calcularse para cada canal EEG y compararse entre diferentes tareas motoras o períodos de reposo, evaluando cambios en la complejidad cortical asociados a la activación cerebral.

3. Common Spatial Patterns (CSP)

Common Spatial Patterns (CSP) es una técnica de filtrado espacial ampliamente utilizada en BCI para maximizar las diferencias entre dos clases de señales EEG. CSP encuentra combinaciones lineales de canales que aumentan la varianza de una clase mientras reducen la de la otra, permitiendo resaltar regiones cerebrales relevantes para la clasificación [9].

Esta técnica es especialmente efectiva en tareas de imaginación motora, ya que enfatiza patrones de desincronización y resincronización en áreas sensorimotoras. CSP suele combinarse con características espectrales y clasificadores como LDA o SVM para mejorar significativamente la precisión de los sistemas BCI [2].

Aunque matemáticamente es más complejo que las medidas anteriores, su implementación en Python es relativamente sencilla gracias a la librería MNE.

En este proyecto, CSP podría utilizarse para diferenciar tareas de imaginación motora derecha e izquierda, resaltando los patrones espaciales asociados a la actividad contralateral observada en las bandas mu y beta.

La coherencia, la entropía muestral y CSP representan tres índices complementarios que aportan información funcional, no lineal y espacial del EEG. La combinación de estas características con la densidad espectral de potencia puede mejorar considerablemente el rendimiento de clasificación en aplicaciones BCI.
