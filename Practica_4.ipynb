## **Introducción a los biopotenciales y adquisición de señales ECG y EMG**

## 1. Etapas del registro de biopotenciales
 

**1. Generación del biopotencial:** Se origina en células excitables por el movimiento de iones (principalmente sodio, potasio y calcio) a través de la membrana celular, produciendo potenciales eléctricos muy pequeños, generalmente en el rango de microvoltios a milivoltios.

**2. Propagación en los tejidos:** La señal viaja a través del cuerpo (tejidos como músculo, grasa y piel), donde puede atenuarse y mezclarse con otras señales y ruido.

**3. Interfaz electrodo–piel:** Los electrodos hacen contacto con la piel y convierten la señal iónica en electrónica, pudiendo introducir impedancia y ruido si no hay una buena preparación de la piel o si existen movimientos.

**4. Captación de la señal:** Los electrodos detectan las diferencias de potencial generadas por la actividad biológica. Dependiendo de la aplicación, pueden utilizarse electrodos superficiales (como en ECG o EMG) o invasivos.

**5. Amplificación:** La señal se amplifica para hacerla medible, reduciendo al mismo tiempo interferencias externas, permitiendo mejorar la calidad de la señal útil frente al ruido.

**6. Filtrado:** Se eliminan ruidos como el de la red eléctrica (50/60 Hz), artefactos de movimiento y el ruido de alta frecuencia. Para ello se emplean filtros pasa-bajas, pasa-altas y rechaza banda.

**7. Conversión analógica-digital (ADC):** La señal analógica se transforma en digital mediante un convertidor ADC para su procesamiento.

**8. Adquisición y visualización:** La señal se almacena, procesa y muestra en un computador para su análisis. En esta etapa se pueden aplicar algoritmos de análisis para interpretar la señal.

## Componentes de un equipo de adquisición

**1. Electrodos:** Son los encargados de captar la señal eléctrica del cuerpo. Generalmente están hechos de materiales conductores como plata/cloruro de plata (Ag/AgCl), que permiten una buena transmisión de la señal.

**2. Amplificador de instrumentación:** Su función es aumentar la amplitud de las señales biológicas sin distorsionarlas. Además, reduce el ruido externo gracias a su alto rechazo de modo común.

**3. Sistema de filtrado:** Incluye tanto filtros analógicos (antes de la digitalización) como filtros digitales (después del ADC). Su objetivo es mejorar la calidad de la señal eliminando componentes no deseados.

**4. Convertidor analógico-digital (ADC):** Transforma la señal analógica en una señal digital. Es un componente clave que determina la precisión y calidad de la adquisición.

**. Sistema de adquisición (DAQ):** Es el hardware que permite la comunicación entre el sistema de medición y el computador. Puede manejar múltiples canales y coordinar la captura de datos.

**6. Software de procesamiento y visualización:** Permite observar la señal en tiempo real, almacenarla y analizarla. Incluye herramientas para filtrado, detección de eventos y extracción de características.

## 2. Principales fuentes de ruido en el registro de biopotenciales

| Fuente de ruido | Descripción | Tipo | Estrategias de control |
|-----------------|------------|------|------------------------|
| Interferencia de red (50/60 Hz) | Ruido de la red eléctrica que se acopla al sistema | Variante en el tiempo | Filtro notch, blindaje, conexión a tierra, alto CMRR en el amplificador [1] |
| Ruido térmico (Johnson) | Ruido por agitación térmica en componentes | Estático (aleatorio) | Componentes de bajo ruido, buen diseño electrónico [1] |
| Artefactos de movimiento | Movimiento del paciente o electrodos | Variante en el tiempo | Fijación de electrodos, preparación de piel, filtro pasa-altas [1], [2] |
| Interferencia electromagnética (EMI) | Señales externas de dispositivos eléctricos | Variante en el tiempo | Blindaje, cables coaxiales, jaula de Faraday [1] |
| Ruido electrodo–piel | Variaciones en la impedancia de contacto | Variante en el tiempo | Limpieza de piel, gel conductor, electrodos adecuados [1], [2] |
| Ruido muscular (EMG) | Actividad muscular que contamina la señal | Variante en el tiempo | Relajación del paciente, filtrado  [2] |
| Deriva de línea base | Cambios lentos por respiración o sudoración | Variante en el tiempo | Filtros pasa-altas, detrending [2] |
| Ruido de cuantización | Error en la digitalización (ADC) | Estático | Mayor resolución del ADC, ajuste del rango dinámico [2] |

## 3. Problemática asociada al acoplamiento piel-electrodo

El acoplamiento piel–electrodo es una de las principales fuentes de degradación de la señal en el registro de biopotenciales. La problemática radica en que la interfaz entre la piel (medio iónico) y el electrodo (medio electrónico) introduce una impedancia de contacto que puede ser alta, variable y no lineal. Esta impedancia depende de factores como la hidratación de la piel, la presencia de vello, la presión del electrodo y la calidad del gel conductor [1].

Además, en esta interfaz se generan potenciales de media celda y fenómenos de polarización, los cuales pueden introducir ruido y deriva en la señal. También es común la aparición de artefactos de movimiento, que producen variaciones bruscas en la señal debido a cambios en la impedancia de contacto [1], [2].

Para minimizar estos problemas, se emplean diversas estrategias como:

**Preparación de la piel:** Limpieza, exfoliación suave o rasurado para reducir la impedancia.

**Uso de gel conductor:** Mejora la conductividad entre la piel y el electrodo, reduciendo la resistencia de contacto.

**Electrodos adecuados (Ag/AgCl):** Disminuyen la polarización y mejoran la estabilidad de la señal.

**Fijación mecánica correcta:** Evita movimientos relativos entre el electrodo y la piel.

**Ubicación adecuada de electrodos:** Colocarlos en zonas con menor interferencia muscular o movimiento.

**Uso de amplificadores con alto CMRR:** Reduce el efecto de interferencias externas como el ruido de red.

El acoplamiento piel–electrodo se considera la etapa más crítica en el registro de biopotenciales porque constituye el primer punto de contacto entre la señal biológica y el sistema de medición. Cualquier alteración en esta interfaz, como una alta impedancia o inestabilidad en el contacto, introduce ruido, distorsión y variabilidad que afectan directamente la calidad de la señal. Esto es especialmente problemático porque dichas perturbaciones ocurren antes de las etapas de amplificación y procesamiento, por lo que no siempre pueden corregirse completamente mediante filtrado posterior. Además, en señales de superficie como sEMG, EEG y ECG, cuyas amplitudes son muy pequeñas, incluso variaciones mínimas en el acoplamiento pueden deteriorar significativamente la relación señal-ruido y comprometer la interpretación de los datos [1], [2].

## 4. Análisis de la Variabilidad de la frecuencia cardíaca (HRV)

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import neurokit2 as nk
from ecgdetectors import Detectors

Ruta_archivo = r"C:\UdeA\Semestre 20261\Bioseñales\Practica adquisicion\ECG_DAVID.csv"

df = pd.read_csv(Ruta_archivo, header=None) #Se lee el archivo, al no tener encabezado y ser una columna se pone header = None.

# Se tomar la única columna que hay.
ecg = df.iloc[:,0].values

#Un gran problema en este caso es la frecuencia de muestro, la cual no sabemos, pero es posible obtenerla, con el
#Numero de muestras y la duracion de la toma de muestras.
duracion = 13  # segundos
fs = int(len(ecg) / duracion)

print("Frecuencia de muestreo estimada:", fs, "Hz")

#Se procede a obtener los picos por NeuroKit
signals, info = nk.ecg_process(ecg, sampling_rate=fs)
rpeaks_neurokit = info["ECG_R_Peaks"] #Se detectan los picos R

#Se procede a obtener los picos por Detectors
detectors = Detectors(int(fs))
rpeaks_detectors = detectors.pan_tompkins_detector(ecg) #Se detectan los picos R

#Se realiza una grafica comparativa con el fin de mostrar que tan certeras son las funciones para detectar los picos R
plt.figure(figsize=(12,5))
plt.plot(ecg, label="ECG")
plt.scatter(rpeaks_neurokit, ecg[rpeaks_neurokit], label="R peaks NeuroKit", color = 'green') #Los picos hallados por Neurokit en color verde
plt.scatter(rpeaks_detectors, ecg[rpeaks_detectors], label="R peaks Pan-Tompkins",color = 'orange')#Los picos hallados por Detectors en color naranja
plt.legend()
plt.title("Comparación detección de picos R")
plt.xlabel("Muestras")
plt.ylabel("Amplitud")
plt.show()
```

Primero que todo, se debe aclarar que la frecuencia de muestreo fue estimada a partir del número de picos R observados en la señal obtenida y considerando una frecuencia cardíaca promedio de reposo de aproximadamente 75 latidos por minuto. Se observaron 16 picos R en los datos obtenidos, entonces, al considerar que el sujeto se encontraba en reposo, se realiza una regla de tres para estimar que el tiempo de muestreo es de 13s aproximadamente. Para finalizar, se calculó la frecuencia de muestreo al dividir la cantidad de muestras sobre tiempo de muestreo; Se obtuvo un valor de 76 Hz, se recomienda que la frecuencia óptima de muestreo del ECG para el análisis de la variabilidad de la frecuencia cardíaca (HRV) esté entre 250 Hz y 500 Hz, pero para una práctica académica esta aproximación suele ser aceptada [3].

Segundo, a partir de la comparación de la señal ECG con los picos detectados, el algoritmo implementado en NeuroKit presenta un mejor desempeño en la identificación de los picos R. Los puntos detectados por este método se ubican directamente sobre los máximos de amplitud del complejo QRS, lo que indica una detección más precisa de la onda R. Por el contrario, el algoritmo Pan-Tompkins implementado en la librería Detectors identifica varios puntos desplazados respecto al máximo real de la señal e incluso en algunos casos sobre regiones cercanas al complejo QRS que no corresponden exactamente al pico R. En conclusión, NeuroKit muestra mayor exactitud en la localización de los latidos cardíacos, lo que resulta más adecuado para análisis posteriores como la variabilidad de la frecuencia cardíaca (HRV).

```python
# Se realiza el calculo de HRV a partir de picos R

# Se calculan los intervalos RR, que son el tiempo transcurrido entre dos picos R consecutivos, en milisegundos
rr_intervals = np.diff(rpeaks_neurokit) / fs * 1000

# Se calcula el HRV que es la desviación estándar de los intervalos RR
hrv = np.std(rr_intervals)

print("HRV (SDNN):", hrv, "ms,", 'presenta una varianza moderada') #Imprimo los valores encontrados
```

La biblioteca hrvanalysis es un paquete de Python diseñado para analizar la variabilidad de la frecuencia cardíaca (HRV), que sirve como una herramienta no invasiva para evaluar la actividad autónoma cardíaca. Esta biblioteca facilita la conversión de los datos del intervalo entre latidos en diversas funciones relacionadas con la HRV, lo que permite a los investigadores explorar las respuestas fisiológicas, como la detección del estrés, mediante métodos analíticos avanzados. Las capacidades de la biblioteca se mejoran gracias a su integración con técnicas de aprendizaje automático, lo que permite clasificar eficazmente los niveles de estrés en función de las métricas de la HRV [4]. 

La **función get_time_domain_features()**, permite calcular diversas métricas de HRV en el dominio del tiempo a partir de una serie de intervalos RR (NN intervals). Esta función recibe como entrada una lista o arreglo con los intervalos RR en milisegundos y devuelve un diccionario con diferentes características estadísticas que describen la variabilidad del ritmo cardíaco.

Algunas metricas son:

**1. mean_nni =**  Promedio de intervalos NN.
**2. sdnn =** Desviación estándar de intervalos NN.
**3. rmssd =** Raíz media cuadrática de diferencias sucesivas.
**4. nn50 =** Número de diferencias mayores a 50 ms.
**5. pnn50 =** Porcentaje de diferencias mayores a 50 ms.


```python
from hrvanalysis import get_time_domain_features

# Características HRV dominio del tiempo
time_features = get_time_domain_features(rr_intervals)

print("Características HRV (dominio del tiempo):")
for key, value in time_features.items():
    print(key, ":", value)
```

## 5. Aplicaciones clínicas del uso de ECG y EMG

### Aplicaciones clínicas del ECG

- **Diagnóstico de arritmias cardíacas**, donde se analizan alteraciones en el ritmo del corazón como taquicardias o fibrilación auricular. En este caso, se emplea generalmente un registro de 12 derivaciones o monitoreo continuo tipo Holter, y el procesamiento incluye filtrado de la señal, detección de picos R y análisis de intervalos como RR, PR y QT [1] [2].
- **Detección de infarto de miocardio**, donde el ECG permite identificar cambios en el segmento ST y la onda T asociados a isquemia. Este análisis se realiza típicamente con registros en reposo de 12 derivaciones y se basa en filtrado y análisis morfológico de la señal [1] [2].
- **Monitoreo continuo de pacientes en unidades de cuidados intensivos**, donde se realiza un registro en tiempo real con untipo de registro continuo mediante telemetría, aplicando filtrado, detección automática de eventos y generación de alarmas ante cambios significativos en la frecuencia o el ritmo cardíaco [2].

### Aplicaciones clínicas del EMG

- **Diagnóstico de enfermedades neuromusculares**, como neuropatías o miopatías, utilizando registros de EMG de aguja o superficial. En este contexto, el procesamiento incluye filtrado en rangos típicos de 20 a 500 Hz, así como el análisis de la amplitud y frecuencia de la señal para identificar patrones anormales [1] [5].
- **Análisis de la marcha y la rehabilitación**, donde se emplea EMG de superficie multicanal sincronizado con sistemas de movimiento para estudiar la activación muscular. En estos casos, la señal se procesa mediante rectificación, cálculo de la envolvente (por ejemplo, usando el valor RMS) y normalización respecto a una contracción máxima [2] [5].
- **Control de prótesis mioeléctricas**, donde las señales musculares se utilizan para accionar dispositivos. Este tipo de aplicación emplea registros de superficie en músculos residuales y requiere procesamiento más avanzado, incluyendo filtrado, extracción de características como RMS, valor absoluto medio (MAV) y frecuencia media, así como técnicas de clasificación basadas en aprendizaje automático [2], [4].
